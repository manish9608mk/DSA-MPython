# 01. Graphs: Fundamentals

A graph is a set of **vertices** (nodes) connected by **edges**. It models anything with **relationships**.

**Learning order:** REAL LIFE → TERMS → TYPES → REPRESENTATIONS → BUILD → GRID AS A GRAPH → CHOOSING → INTERVIEW THINKING

## 1. Real-life examples

- **Google Maps:** places are vertices, roads are edges (weighted by distance/time).
- **Facebook/LinkedIn:** people are vertices, friendships are edges.
- **The web:** pages are vertices, hyperlinks are directed edges.
- **Course prerequisites, build systems (package dependencies):** directed edges "A must come before B".
- **Network routers, flight routes, recommendation engines.**

Trees and linked lists are special graphs (no cycles / one path).

## 2. Terms

| Term | Meaning |
|---|---|
| vertex / node, edge | thing and connection between two things |
| **directed** / **undirected** | edge has a direction (A → B) or not |
| **weighted** | edge has a cost (distance, time) |
| degree (in / out) | number of edges at a vertex (incoming / outgoing in directed graphs) |
| path / cycle | sequence of connected vertices / a path that returns to its start |
| connected | undirected graph with a path between every pair |
| **DAG** | **D**irected **A**cyclic **G**raph (dependencies, course schedules) |
| sparse / dense | `E ≈ V` / `E ≈ V²` |

`V` = number of vertices, `E` = number of edges. Complexities are written in `V` and `E`.

## 3. Three ways to store a graph

| Representation | Space | "Is there an edge u-v?" | "All neighbours of u" | Best for |
|---|---|---|---|---|
| **Adjacency list** (`dict` or list of lists) | O(V + E) | O(degree) | **O(degree)** | most problems (sparse graphs) |
| Adjacency matrix (V × V) | O(V²) | **O(1)** | O(V) | dense graphs, Floyd-Warshall, small V |
| Edge list | O(E) | O(E) | O(E) | Kruskal, Bellman-Ford |

**Default choice in interviews: adjacency list.**

```python
from collections import defaultdict

edges = [(0, 1), (0, 2), (1, 2), (2, 3)]

def build_adj_list(n, edges, directed=False):
    graph = defaultdict(list)
    for u, v in edges:
        graph[u].append(v)
        if not directed:
            graph[v].append(u)
    for node in range(n):                      # make sure isolated nodes exist
        graph[node]
    return graph

g = build_adj_list(4, edges)
print(dict(g))   # {0: [1, 2], 1: [0, 2], 2: [0, 1, 3], 3: [2]}

matrix = [[0] * 4 for _ in range(4)]
for u, v in edges:
    matrix[u][v] = matrix[v][u] = 1
print(matrix[2])   # [1, 1, 0, 1]

degree = {node: len(nei) for node, nei in g.items()}
print(degree)   # {0: 2, 1: 2, 2: 3, 3: 1}
```

## 4. Weighted and directed graphs

Store `(neighbour, weight)` pairs. For a directed graph add the edge in **one direction only**.

```python
weighted = defaultdict(list)
for u, v, w in [("A", "B", 4), ("A", "C", 1), ("C", "B", 2)]:
    weighted[u].append((v, w))                 # directed A -> B, A -> C, C -> B
print(dict(weighted))   # {'A': [('B', 4), ('C', 1)], 'C': [('B', 2)]}
```

## 5. A grid is a graph

Every cell `(r, c)` is a vertex with up to 4 neighbours (up, down, left, right). You do not build the graph; you **generate neighbours on the fly**. Most "island / maze / flood" problems are graph problems.

```python
DIRECTIONS = [(1, 0), (-1, 0), (0, 1), (0, -1)]

def neighbours(r, c, rows, cols):
    for dr, dc in DIRECTIONS:
        nr, nc = r + dr, c + dc
        if 0 <= nr < rows and 0 <= nc < cols:
            yield nr, nc

print(list(neighbours(0, 0, 3, 3)))   # [(1, 0), (0, 1)]
```

## 6. Sizing the problem (important in interviews)

If edges are given as pairs of **1-indexed** labels, convert or allocate `n + 1`. Check: are there self-loops, duplicate edges, disconnected components, node labels not starting at 0? Are nodes strings or numbers?

Typical limits: `V, E ≤ 10⁵` → you need O(V + E) or O(E log V). `V ≤ 400` → O(V³) (Floyd-Warshall) is fine.

## 7. Interview thinking

**First questions:** directed or undirected? weighted? could there be cycles? is the graph connected? how is it given (edges, matrix, grid)?

**Say:** "I'll model this as a graph where nodes are ... and edges are ... Then ... is a (shortest path / connectivity / ordering) problem."

**Common mistakes:** forgetting both directions for undirected edges, forgetting isolated nodes, confusing 0-indexed and 1-indexed nodes, using a matrix when V is large (memory).

## Quick Revision

- Graph = vertices + edges. Directed/undirected, weighted/unweighted, cyclic/acyclic.
- Adjacency list: O(V + E) space, the default.
- Grid = implicit graph with 4 (or 8) neighbours.
- Start every problem by naming what the nodes and edges are.
