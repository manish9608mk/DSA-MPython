## 1. Decision table

| Question looks like | Pattern |
|---|---|
| connected components, can I reach X | DFS / BFS / DSU |
| islands, flood, regions on a grid | grid DFS/BFS |
| reachable from the border/ocean | reverse search from the boundary |
| fewest steps, unweighted shortest path | BFS (multi-source if many starts) |
| order with prerequisites, is it possible | topological sort (Kahn) |
| cycle in directed graph | DFS 3 states or Kahn |
| cycle in undirected graph | DFS with parent or DSU |
| edges added over time, groups | Union-Find |
| weighted shortest path, non-negative | Dijkstra |
| negative weights / at most K edges | Bellman-Ford |
| all pairs, small V | Floyd-Warshall |
| connect everything cheapest | MST (Kruskal / Prim) |
| two groups, no conflicts | bipartite coloring |
| minimize the maximum edge on a path | Dijkstra with `max`, or DSU + sorted edges, or binary search + BFS |

## 2. Complexities

| Algorithm | Time | Space |
|---|---|---|
| BFS / DFS | O(V + E) | O(V) |
| Topological sort (Kahn) | O(V + E) | O(V) |
| Union-Find | O(α(n)) per op | O(V) |
| Dijkstra (heap) | O((V + E) log V) | O(V) |
| Bellman-Ford | O(V · E) | O(V) |
| Floyd-Warshall | O(V³) | O(V²) |
| Kruskal | O(E log E) | O(V) |
| Prim (heap) | O(E log V) | O(V) |

## 3. Templates to remember

**BFS (shortest path)**
```python
q = deque([src]); seen = {src}
while q:
    for _ in range(len(q)):
        u = q.popleft()
        for v in adj[u]:
            if v not in seen: seen.add(v); q.append(v)
```
**Dijkstra**
```python
heap = [(0, src)]
while heap:
    d, u = heappop(heap)
    if d > dist[u]: continue
    for v, w in adj[u]:
        if d + w < dist[v]: dist[v] = d + w; heappush(heap, (dist[v], v))
```
**Kahn**: in-degrees → queue of zeros → reduce neighbours.
**DSU**: `find` with compression, `union` by size.

## 4. Common mistakes

- Missing `visited` → infinite loops.
- BFS marking on pop, not on push.
- BFS on weighted graphs, Dijkstra with negative edges.
- Missing disconnected components (loop over all nodes).
- Directed vs undirected edges (adding only one direction).
- 0-indexed vs 1-indexed nodes.
- Recursion depth on large graphs/grids.
- Reversed edge direction in prerequisite problems.
- Building a matrix for V = 10⁵.

## 5. Interview script

1. Clarify: directed? weighted? cycles? connected? size?
2. "Nodes are ..., edges are ... This is a (shortest path / ordering / connectivity) problem."
3. Pick the algorithm and justify it.
4. State complexity in V and E.
5. Test: single node, disconnected graph, cycle, duplicate edges, unreachable target.

## Quick Revision

- Name nodes and edges first. Pick BFS / DFS / DSU / topo / Dijkstra / MST from the table.
- Visited always. Check direction and indexing.
- Weighted + non-negative = Dijkstra. K edges = Bellman-Ford.
- Reverse thinking from the boundary or target.
