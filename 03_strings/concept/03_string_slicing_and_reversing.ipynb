# 03. String Slicing and Reversing

**Goal:** Learn how to select a part of a string and reverse it.

## Core pattern

Slicing syntax: `s[start:stop:step]`

- `start`: where to begin (included).
- `stop`: where to stop (excluded).
- `step`: how far to move each time.

Think of slicing as selecting a range of characters.

## 1. Understand the indexes

For `s = "python"`:

| Character | p | y | t | h | o | n |
|---|---:|---:|---:|---:|---:|---:|
| Index | 0 | 1 | 2 | 3 | 4 | 5 |

The stop index is **not included**. Therefore, `s[1:4]` takes indexes `1`, `2`, and `3`.

```python
s = "python"

print(s[1:4])  # yth
print(s[:2])   # py
print(s[4:])   # on
print(s[:])    # python
```

## 2. The three slicing parts

- `s[a:b]`: characters from index `a` up to, but not including, `b`.
- `s[:b]`: from the beginning up to `b`.
- `s[a:]`: from `a` to the end.
- `s[a:b:step]`: move by the given step.

If the start or stop is outside the string, slicing normally returns the available characters instead of raising `IndexError`.

```python
s = "abcdef"

print(s[1:5])   # bcde
print(s[::2])   # ace: indexes 0, 2, 4
print(s[1::2])  # bdf: indexes 1, 3, 5
print(s[2:2])   # '' (empty string)
```

## 3. Reverse a string

A negative step moves backwards. The common Python shortcut is `s[::-1]`.

**Interview clue:** If the task only asks for a reversed string, slicing is a simple solution. If the interviewer asks you to explain the algorithm or solve it without slicing, use a loop or two pointers.

```python
s = "python"
print(s[::-1])  # nohtyp

s = "level"
print(s[::-1])  # level
```

## 4. Indexing vs slicing

- `s[i]` returns one character and raises `IndexError` if `i` is invalid.
- `s[a:b]` returns a portion and safely handles boundaries beyond the string.

Check for an empty string before using `s[0]` or `s[-1]`.

```python
s = "cat"

print(s[0])       # c
print(s[0:100])   # cat
# print(s[100])   # IndexError
```

## 5. Interview thinking

Before coding a slicing problem, ask:
1. Do I need the whole string or only a portion?
2. Is the end index included? (In Python slicing, no.)
3. Is a negative step needed to move backwards?
4. Does the question require a general algorithm rather than a Python shortcut?

Do not use slicing automatically for every problem. Sometimes you must compare characters or build the answer using a specific pattern.

## Quick revision

- Syntax: `s[start:stop:step]`.
- `start` is included; `stop` is excluded.
- `s[:k]` takes the beginning up to `k`.
- `s[k:]` takes from `k` to the end.
- `s[::2]` takes indexes `0, 2, 4, ...`.
- `s[::-1]` reverses the string.
- Slicing usually handles out-of-range boundaries; indexing does not.
