## 1. What is a string?

A **string** is a sequence of characters. Characters can be letters, digits, spaces, or symbols.

```python
name = "Manish"
message = "Hello, world!"
code = "A1B2"
```

Think of a string as a row of character boxes. Each character has a position called an **index**.

In Python, strings are written inside single quotes (`'...'`) or double quotes (`"..."`).

````python
name = "Manish"
message = "Hello, world!"
code = "A1B2"

print("Name:", name)
print("Message:", message)
print("Code:", code)
print("Type:", type(name))
````

## 2. Length of a string

Use `len(s)` to find how many characters are in a string.

Important:
- Spaces count as characters.
- Repeated characters count separately.
- An empty string `""` has length `0`.

**Interview clue:** If a question asks how many characters are present, think of `len()`.

````python
s = "hello"
print(len(s))  # 5

s = "hi there"
print(len(s))  # 8: the space also counts

s = ""
print(len(s))  # 0
````

## 3. Positive and negative indexing

**Indexing** means accessing a character by its position.

For `s = "hello"`:

| Character | h | e | l | l | o |
|---|---:|---:|---:|---:|---:|
| Positive index | 0 | 1 | 2 | 3 | 4 |
| Negative index | -5 | -4 | -3 | -2 | -1 |

Examples:
- `s[0]` → first character, `"h"`
- `s[1]` → second character, `"e"`
- `s[-1]` → last character, `"o"`
- `s[-2]` → second character from the end, `"l"`

**Remember:** Positive indexes start at `0`. Negative indexes start at `-1` from the end.

A position that does not exist raises `IndexError`. For example, `"hello"[5]` is invalid because the valid positive indexes are `0` through `4`.

````python
s = "hello"

print(s[0])   # h
print(s[1])   # e
print(s[-1])  # o
print(s[-2])  # l

# Uncomment to see the error:
# print(s[5])
````

### First and last character

For a **non-empty** string:
- Last index = `len(s) - 1`
- Last character = `s[len(s) - 1]` or simply `s[-1]`

Be careful: indexing an empty string with `s[0]` or `s[-1]` raises `IndexError`. Check whether the string is empty when a question allows empty input.

````python
s = "python"

print("Length:", len(s))
print("Last index:", len(s) - 1)
print("Last character:", s[len(s) - 1])
print("Last character (short way):", s[-1])
````

## 4. String traversal

**Traversal** means visiting characters one by one.

### Method A: Visit each character

Use `for ch in s` when you need the character but do not need its position.

````python
s = "cat"

for ch in s:
    print(ch)
````

### Method B: Visit each index

Use `range(len(s))` when you need the index as well as the character.

- `range(len(s))` produces indexes from `0` to `len(s) - 1`.
- `s[i]` accesses the character at index `i`.

**Interview rule:** Choose the simplest loop that gives you the information you need.

````python
s = "cat"

for i in range(len(s)):
    print(i, s[i])
````

## 5. Count a character: traversal + counter

**Real-life analogy:** Count how many times a particular letter appears in a word. Look at each letter; whenever it matches the target, increase your count by one.

Steps:
1. Choose the string and target character.
2. Start `count = 0`.
3. Visit every character.
4. If the character equals the target, do `count += 1`.
5. After the loop, `count` is the answer.

This is the **traversal + counter** pattern.

````python
s = "banana"
target = "a"
count = 0

for ch in s:
    if ch == target:
        count += 1

print("Frequency:", count)  # 3
````

### Dry run: `"banana"` and target `"a"`

| Character | Is it `"a"`? | Count after checking |
|---|---|---:|
| b | No | 0 |
| a | Yes | 1 |
| n | No | 1 |
| a | Yes | 2 |
| n | No | 2 |
| a | Yes | 3 |

Final answer: `3`.

**Complexity:** Time `O(n)` because every character is checked. Extra space `O(1)` because only a fixed number of variables are used.

Python also has `s.count(target)`, but writing the loop teaches the pattern used in many interview problems.

````python
# Built-in shortcut (learn it, but understand the loop above too)
s = "banana"
print(s.count("a"))  # 3
print(s.count("z"))  # 0
````

## 6. Strings are immutable

**Immutable** means a string cannot be changed in place after it is created.

```python
s = "cat"
# s[0] = "b"  # TypeError
```

To get `"bat"`, create a **new** string:

```python
s = "b" + s[1:]
```

This does not modify the old string character by character. It creates a new string and assigns it to `s`.

**Interview relevance:** When you need to build or change strings, remember that string operations often create new strings. For many repeated edits, a list of characters and `''.join(...)` can be useful; we will learn that when needed.

````python
s = "cat"

# This would raise TypeError:
# s[0] = "b"

s = "b" + s[1:]
print(s)  # bat
````

## 7. Recognize the basic pattern

When you read a string question, identify what the question asks before choosing code.

| Question clue | First pattern to consider |
|---|---|
| Visit or inspect every character | Traversal |
| Read a character at a given position | Indexing |
| Count matching characters | Traversal + counter |
| Compare characters from both ends | Two pointers |
| Compare how often characters occur | Frequency map / hashing |

These are starting clues, not fixed rules. Some questions need more than one pattern. We will study two pointers and hashing in their own lessons.

## 8. Common mistakes to avoid

- Forgetting that the first index is `0`, not `1`.
- Using `s[len(s)]` for the last character; use `s[len(s) - 1]` or `s[-1]`.
- Accessing an index in an empty string.
- Trying to assign to `s[i]`; strings are immutable.
- Forgetting to initialize a counter before the loop.
- Incrementing the counter for every character instead of only when the condition matches.
