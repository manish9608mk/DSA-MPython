# 05. String Frequency Counting and Hashing

**Goal:** Count characters and compare how often they appear.

## Core idea

A frequency map stores each character and its count.

Example for `"banana"`:
- `b: 1`
- `a: 3`
- `n: 2`

In Python, a dictionary (`dict`) is a common frequency map.

**Question clues:** count characters, compare character counts, find duplicates, or find the first character with a special frequency.

## 1. Count with a dictionary

For each character, increase its count by one.

`freq[ch] = freq.get(ch, 0) + 1`

`get(ch, 0)` means: use the current count if the key exists; otherwise start at `0`.

```python
s = "banana"
freq = {}

for ch in s:
    freq[ch] = freq.get(ch, 0) + 1

print(freq)  # {'b': 1, 'a': 3, 'n': 2}
```

## 2. Find duplicates

Once counts are stored, characters with a count greater than `1` are duplicates.

```python
s = "programming"
freq = {}

for ch in s:
    freq[ch] = freq.get(ch, 0) + 1

duplicates = []
for ch, count in freq.items():
    if count > 1:
        duplicates.append(ch)

print(duplicates)  # ['r', 'g', 'm']
```

## 3. Compare two strings by frequency

**Example:** Anagrams contain the same characters with the same frequencies, ignoring order.

`"listen"` and `"silent"` are anagrams because their character counts match.

A simple approach is to count each string and compare the dictionaries. For this example, case and spaces are treated as significant.

```python
def are_anagrams(a, b):
    if len(a) != len(b):
        return False

    freq = {}
    for ch in a:
        freq[ch] = freq.get(ch, 0) + 1

    for ch in b:
        if ch not in freq:
            return False
        freq[ch] -= 1
        if freq[ch] < 0:
            return False

    return True


print(are_anagrams("listen", "silent"))  # True
print(are_anagrams("hello", "world"))    # False
```

## 4. First unique character: use counts, then scan

A useful pattern is **two passes**:
1. Count all characters.
2. Scan the original string in order and find the first character with count `1`.

The second pass preserves the original string order.

```python
def first_unique_char(s):
    freq = {}

    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1

    for ch in s:
        if freq[ch] == 1:
            return ch

    return None


print(first_unique_char("swiss"))  # w
print(first_unique_char("aabb"))   # None
```

## 5. Complexity and recognition

For a string of length `n`:
- Building a dictionary is typically `O(n)` time.
- Dictionary operations are typically `O(1)` on average.
- The frequency map uses `O(k)` space, where `k` is the number of distinct characters.

**Use a frequency map when:** counts matter more than positions or order.

**Do not use it automatically:** if the task only asks to reverse a string, a frequency map is unnecessary.

3. Check whether `"race"` and `"care"` are anagrams.

```python
# Answer check — run after trying yourself.

def frequency_map(s):
    freq = {}
    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1
    return freq

print("Q1:", frequency_map("mississippi"))
# {'m': 1, 'i': 4, 's': 4, 'p': 2}

freq = frequency_map("aabbcdeee")
print("Q2:", [ch for ch, count in freq.items() if count > 1])
# ['a', 'b', 'e']

print("Q3:", are_anagrams("race", "care"))  # True
print("Q4:", first_unique_char("aabbcddee"))  # c
```

## Quick revision

- Frequency map: `freq[ch] = freq.get(ch, 0) + 1`.
- Counts help identify duplicates and compare character composition.
- Anagrams have the same character frequencies.
- To find the first unique character, count first, then scan the original string.
- Typical time: `O(n)`; extra space: `O(k)`.
