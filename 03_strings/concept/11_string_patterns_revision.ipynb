## 2. Complexity cheat sheet

| Operation | Time |
|---|---|
| index, `len` | O(1) |
| slice, `+`, `join`, `split`, `replace`, `reverse` | O(n) |
| `in`, `find` | O(n·m) worst (KMP gives O(n + m)) |
| build by `+=` in a loop | O(n²) worst, avoid |
| `Counter(s)` | O(n) |
| sort characters | O(n log n), or O(n) with a 26-array |

## 3. Common mistakes

- Treating strings as mutable (`s[0] = "x"` raises `TypeError`).
- `s += ch` in a long loop; use a list and `join`.
- Slicing in recursion or loops (copies).
- Case and whitespace assumptions; non-alphanumeric characters.
- Off-by-one in substring bounds (`s[i:j]` excludes `j`).
- Ignoring empty string and single-character input.
- Mixing **subsequence** and **substring**.
- Using `lower()` for case-insensitive compare in Unicode text (use `casefold`).

## 4. Interview script

1. Clarify: alphabet (lowercase / ASCII / Unicode), case, length, empty input.
2. Brute force, then the pattern: window, hashing, pointers, DP, stack.
3. State time/space, remembering the cost of slicing and concatenation.
4. Test: empty, one char, all same, all different, mixed case, repeated blocks.

## Quick Revision

- Pattern table first. Immutable strings: use `join`.
- Substring ≠ subsequence. Slicing costs O(n).
- KMP/Z for guaranteed O(n + m), Manacher for O(n) palindromes.
- Unicode: characters vs bytes, normalize before comparing.
