# 04. String Building and Immutability

**Goal:** Understand why strings cannot be changed in place and how to create a result string.

## 1. Strings are immutable

Immutable means a string cannot be changed character by character after creation.

```python
s = "cat"
s[0] = "b"  # TypeError
```

To get `"bat"`, create a new string instead.

````python
s = "cat"

# s[0] = "b"  # Uncomment to see TypeError

s = "b" + s[1:]
print(s)  # bat
````

## 2. Build a result with a loop

**Question clue:** Keep certain characters, remove others, or transform characters.

Start with an empty string and add the characters you want to keep. This is easy to understand for small examples.

````python
s = "a1b2c3"
result = ""

for ch in s:
    if ch.isalpha():
        result += ch

print(result)  # abc
````

## 3. Build a list, then join

For many repeated additions, a common Python approach is to collect pieces in a list and join them at the end. This avoids repeatedly creating a growing string.

Pattern:
1. Start with `chars = []`.
2. Append each character you want.
3. Use `"".join(chars)` to make the final string.

````python
s = "a1b2c3"
chars = []

for ch in s:
    if ch.isalpha():
        chars.append(ch)

result = "".join(chars)
print(result)  # abc
````

## 4. Transform characters

Strings are immutable, so methods such as `upper()` return a new string rather than changing the original string.

````python
s = "hello"
new_s = s.upper()

print(s)      # hello
print(new_s)  # HELLO
````

## 5. Interview recognition

| Requirement | First approach to consider |
|---|---|
| Replace or transform characters | Build a new string |
| Keep only characters matching a rule | Traversal + condition + result |
| Build output from many pieces | List + `join()` |
| Change a character at a specific position | Build a new string or use a character list |

Use the simplest correct approach first. Optimize when input size or constraints make it necessary.
