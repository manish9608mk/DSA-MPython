# 06. Two Pointers and Sliding Window Basics

**Goal:** Recognize two important patterns used in string interviews.

These patterns are not interchangeable:
- **Two pointers:** Track two positions, often moving toward each other or in the same direction.
- **Sliding window:** Maintain a continuous section of the string and move its boundaries.

We will learn the basic recognition clues. Most of the skill comes from solving problems.

## 1. Two pointers: compare both ends

**Question clue:** Check whether a string reads the same forward and backward, or compare characters from opposite ends.

Start one pointer at the beginning and another at the end. Compare, then move inward.

```python
def is_palindrome(s):
    left = 0
    right = len(s) - 1

    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1

    return True


print(is_palindrome("level"))  # True
print(is_palindrome("hello"))  # False
```

## 2. Two pointers: move in the same direction

Sometimes both pointers move forward. This is common when removing or skipping characters, or comparing two sequences. The exact movement depends on the problem; we will learn it through questions.

```python
# Example: check whether s is a subsequence of t.
def is_subsequence(s, t):
    i = 0

    for ch in t:
        if i < len(s) and s[i] == ch:
            i += 1

    return i == len(s)


print(is_subsequence("ace", "abcde"))  # True
print(is_subsequence("aec", "abcde"))  # False
```

## 3. Sliding window: a continuous substring

A **substring** is a continuous part of a string.

**Question clues:** longest substring, shortest substring, or a substring satisfying a condition.

A window is a continuous range `[left, right]`. Move `right` to expand it. When the window violates a condition, move `left` to shrink it. The exact rule depends on the problem.

```python
# Example: maximum number of vowels in any window of size k.
def max_vowels(s, k):
    vowels = set("aeiou")
    count = 0

    for i in range(k):
        if s[i] in vowels:
            count += 1

    best = count

    for right in range(k, len(s)):
        if s[right - k] in vowels:
            count -= 1
        if s[right] in vowels:
            count += 1
        best = max(best, count)

    return best


print(max_vowels("abciiidef", 3))  # 3
```

## 4. How to choose the pattern

| Clue in the question | Consider |
|---|---|
| Compare first and last characters | Two pointers |
| Check a palindrome | Two pointers |
| Match a sequence in order | Two pointers |
| Longest/shortest continuous substring | Sliding window |
| Fixed-size substring/window | Fixed-size sliding window |
| Character counts or anagrams | Frequency map; sometimes sliding window |

**Important:** These clues guide your first attempt; they do not replace understanding the constraints.

## 5. Complexity

- Two-pointer palindrome check: `O(n)` time, `O(1)` extra space.
- Subsequence check: `O(len(t))` time, `O(1)` extra space.
- Fixed-size sliding window: typically `O(n)` time; extra space depends on what the window tracks.

A good window solution updates its state as boundaries move instead of recalculating the entire window every time.

## Quick revision

- **Two pointers:** track two positions; useful for comparing ends or matching sequences.
- **Sliding window:** track a continuous substring and update it as boundaries move.
- **Fixed-size window:** remove the outgoing character and add the incoming one.
- **Variable-size window:** expand and shrink according to the problem's condition.
