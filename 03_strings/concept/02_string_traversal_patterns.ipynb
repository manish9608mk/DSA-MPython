# 02. String Traversal Patterns

**Goal:** Learn to turn a string question into a simple loop.

## The core idea

Traversal means checking characters one by one.

Use:
- `for ch in s` when you need each character.
- `for i in range(len(s))` when you need each index.
- A condition (`if`) to decide what to do.
- A counter or result variable to store the answer.

**Question-solving steps:** Understand the output → choose what to track → traverse → update when the condition is true → return the result.

## 1. Pattern: inspect every character

**Clue:** The question asks you to print, inspect, or process every character.

Real-life example: Read every letter in a word from left to right.

````python
s = "code"

for ch in s:
    print(ch)
````

## 2. Pattern: traversal + condition

**Clue:** Do something only when a character meets a rule.

Example: Count vowels. First decide what counts as a vowel, then check each character.

````python
s = "education"
vowels = "aeiou"
count = 0

for ch in s:
    if ch in vowels:
        count += 1

print(count)  # 5
````

## 3. Pattern: traversal + counter

**Clue:** The question asks “how many?”

Example: Count uppercase letters. `isupper()` checks whether a character is uppercase.

````python
s = "PyTHon3"
count = 0

for ch in s:
    if ch.isupper():
        count += 1

print(count)  # 3
````

## 4. Pattern: traversal + build a result

**Clue:** Keep only certain characters or create a transformed string.

Strings are immutable, so build a new result instead of changing `s` in place.

````python
s = "a1b2c3"
result = ""

for ch in s:
    if ch.isalpha():
        result += ch

print(result)  # abc
````

## 5. Pattern: traversal with indexes

**Clue:** You need the character's position, such as checking characters at even indexes.

````python
s = "python"

for i in range(len(s)):
    if i % 2 == 0:
        print(i, s[i])  # indexes 0, 2, 4
````

## 6. A reusable interview template

```python
def solve(s):
    result = 0  # choose a suitable starting value

    for ch in s:
        if condition:
            result += 1

    return result
```

This is a **template**, not code to copy unchanged. Replace `condition` and `result` according to the question. For building a string, use `result = ""` and append matching characters.

````python
def count_digits(s):
    count = 0

    for ch in s:
        if ch.isdigit():
            count += 1

    return count


print(count_digits("a1b23"))  # 3
````

## 7. Complexity

For a single loop over a string of length `n`:
- **Time:** usually `O(n)` because each character is checked once.
- **Extra space:** `O(1)` for a counter; `O(k)` if a result of up to `k` characters is built.

Note: Repeated string concatenation can cost extra time for large inputs. We will learn efficient string building when needed.

## Quick revision

- **Inspect each character:** `for ch in s`
- **Need positions:** `for i in range(len(s))`
- **Question asks “how many?”:** use a counter.
- **Apply a rule:** use `if`.
- **Create filtered output:** build a new result.
- **One full traversal:** usually `O(n)` time.
