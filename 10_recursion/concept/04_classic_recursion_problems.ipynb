# 04. Classic Recursion Problems

These problems train you to **think in recursion**. The last two lead straight into **Backtracking**.

**Learning order:** TOWER OF HANOI → BINARY STRINGS → SUBSETS (TAKE OR SKIP) → CLIMBING STAIRS → INTERVIEW THINKING

## 1. Tower of Hanoi

Move `n` disks from `A` to `C` using `B`. A bigger disk can never sit on a smaller one.

**Leap of faith:** to move `n` disks:
1. Move the top `n-1` disks from `A` to `B` (trust recursion).
2. Move the biggest disk from `A` to `C`.
3. Move the `n-1` disks from `B` to `C` (trust recursion).

`T(n) = 2T(n-1) + 1` → **2ⁿ − 1 moves**.

```python
def hanoi(n, src, dst, aux, moves):
    if n == 0:
        return
    hanoi(n - 1, src, aux, dst, moves)
    moves.append((src, dst))
    hanoi(n - 1, aux, dst, src, moves)

moves = []
hanoi(2, "A", "C", "B", moves)
print(moves)                  # [('A', 'B'), ('A', 'C'), ('B', 'C')]

moves = []
hanoi(3, "A", "C", "B", moves)
print(len(moves))             # 7  (= 2^3 - 1)
```

## 2. All binary strings of length n

At every position we have **2 choices**: `0` or `1`. Build the string one character at a time. When its length is `n`, we have one answer.

The recursion tree has 2ⁿ leaves. Real life: all possible ON/OFF settings of n switches.

```python
def binary_strings(n):
    result = []

    def build(current):
        if len(current) == n:
            result.append(current)
            return
        build(current + "0")
        build(current + "1")

    build("")
    return result

print(binary_strings(3))   # ['000', '001', '010', '011', '100', '101', '110', '111']
```

## 3. Subsets: take it or skip it

For each element there are two choices: **skip** it or **take** it. After deciding all elements, we have one subset. Total subsets = 2ⁿ.

Real life: you have n pizza toppings; each pizza is "take or skip" for every topping.

**Undo step:** after trying "take", we remove the element (`pop()`) so the next branch starts clean. This **choose → explore → undo** pattern is the core of **Backtracking**.

```python
def subsets(nums):
    result = []
    current = []

    def go(i):
        if i == len(nums):
            result.append(current[:])    # copy, not the same list
            return
        go(i + 1)                        # skip nums[i]
        current.append(nums[i])          # take nums[i]
        go(i + 1)
        current.pop()                    # undo

    go(0)
    return result

out = subsets([1, 2, 3])
print(len(out))                                  # 8
print(sorted(out))
```

## 4. Climbing Stairs

You climb 1 or 2 steps at a time. Ways to reach step `n`? The last move came from step `n-1` or `n-2`, so `ways(n) = ways(n-1) + ways(n-2)`. Same shape as Fibonacci, so memoize it.

```python
from functools import lru_cache

@lru_cache(maxsize=None)
def climb(n):
    if n <= 2:
        return n
    return climb(n - 1) + climb(n - 2)

print(climb(5), climb(10), climb(40))   # 8 89 165580141
```

## 5. Interview thinking

**Signals:**
- "Move / transform n things" with smaller copies inside (Hanoi) → recursion.
- "Generate all ..." with choices per position → recursion tree with `choices ^ depth` leaves.
- "Number of ways" with `ways(n)` built from `ways(n-1)`, `ways(n-2)` → recursion + memo.

**Common mistakes:** appending `current` instead of `current[:]` (all answers become the same list), forgetting the undo step, not stating that the answer size is exponential (2ⁿ), forgetting memoization in count-the-ways problems.

## Quick Revision

- Hanoi: move n-1, move the big one, move n-1. 2ⁿ − 1 moves.
- "Generate all": one choice per position → tree of `choices^n` leaves.
- Subsets: skip or take, `append` then `pop` (undo), store `current[:]`.
- Count ways with `f(n-1) + f(n-2)` → memoize.
