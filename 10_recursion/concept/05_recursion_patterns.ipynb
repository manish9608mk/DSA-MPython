# 05. Recursion Patterns (Revision Sheet)

**Learning order:** FRAMEWORK → SHAPES TABLE → RECURSION VS LOOP → MISTAKES → INTERVIEW SCRIPT → PRACTICE MAP

## 1. Framework (5 steps)

1. Write what `f(args)` **means** in one sentence.
2. **Base case:** the smallest input with an obvious answer.
3. **Recursive case:** make a smaller input, trust `f`, combine the result.
4. Check: does every path reach the base case? (progress)
5. Complexity: draw the tree. Time = nodes × work. Space = depth.

## 2. Shapes

| Shape | Recurrence | Time | Examples |
|---|---|---|---|
| shrink by 1 | `f(n-1)` | O(n) | factorial, sum, reverse |
| shrink by half | `f(n/2)` | O(log n) | fast power, binary search, gcd |
| two halves | `f(left) + f(right)` + combine | O(n) or O(n log n) | max, merge sort |
| two overlapping calls | `f(n-1) + f(n-2)` | O(2ⁿ) → O(n) with memo | fib, climbing stairs |
| take or skip | `f(i+1)` twice | O(2ⁿ) | subsets |
| k choices per step | `k` calls per level | O(kⁿ) | binary strings, permutations (next) |

## 3. Recursion vs loop

| Use recursion when | Use a loop when |
|---|---|
| structure is nested or tree-like | simple counting / scanning |
| you need to explore choices | depth can be huge (10⁵+) in Python |
| divide and conquer | memory is tight |

Any recursion can be rewritten as a loop (sometimes with an explicit stack).

## 4. Common mistakes

- Missing or wrong **base case**.
- Input does **not shrink** → infinite recursion.
- Forgetting `return` on the recursive call (function returns `None`).
- Slicing inside recursion (hidden O(n) per call).
- Same subproblem repeated → no memoization.
- Appending the same list object instead of a copy.
- Mutable default arguments (`memo={}`) kept between calls.
- Claiming O(1) space (it is O(depth)).

## 5. Interview script

1. "I will solve this recursively. `f(...)` returns ..."
2. "Base case: ..."
3. "Otherwise I reduce to ... and combine ..."
4. "The tree has ... nodes, so time is ...; the depth is ..., so space is ..."
5. "Subproblems repeat, so I can memoize." (if true)
6. Test: smallest input, one element, negative or zero, large input (recursion limit).

## Quick Revision

- Define `f`, base case, trust the smaller call, combine.
- Time = nodes × work; space = depth.
- Shapes: n-1, n/2, two halves, overlapping (memo), take/skip, k choices.
- Repeated subproblems → memoization → DP.
- Always copy the list when saving an answer; undo after exploring.
