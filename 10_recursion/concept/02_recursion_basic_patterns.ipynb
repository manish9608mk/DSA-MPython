# 02. Basic Recursion Patterns

Most recursion problems fit one of a few shapes. Learn the shape and the code writes itself.

**Learning order:** LINEAR (n-1) → HALVING (n/2) → DIVIDE AND CONQUER → INDEX PARAMETERS → INTERVIEW THINKING

## 1. Shape A: shrink by one (linear recursion)

`f(n)` uses `f(n - 1)` once. Depth = n.

Examples: factorial, sum of list, sum of digits, palindrome check with two indexes.

```python
def sum_digits(n):
    if n < 10:
        return n
    return n % 10 + sum_digits(n // 10)

print(sum_digits(493))   # 16

def is_palindrome(s, l=0, r=None):
    if r is None:
        r = len(s) - 1
    if l >= r:
        return True
    if s[l] != s[r]:
        return False
    return is_palindrome(s, l + 1, r - 1)

print(is_palindrome("racecar"), is_palindrome("hello"))   # True False
```

## 2. Shape B: shrink by half (O(log n) depth)

`f(n)` uses `f(n / 2)` once. Examples: fast power, binary search, Euclid's GCD.

**Fast power:** `x^n = (x^(n/2))²` (times `x` if `n` is odd). Instead of n multiplications we do about log n. Compute `x^(n/2)` **once** and reuse it. Calling it twice kills the speedup.

```python
def power(x, n):
    if n < 0:
        return 1 / power(x, -n)
    if n == 0:
        return 1
    half = power(x, n // 2)          # compute once
    if n % 2 == 0:
        return half * half
    return half * half * x

print(power(2, 10), power(2, -2), power(2, 0))   # 1024 0.25 1

def gcd(a, b):
    if b == 0:
        return a
    return gcd(b, a % b)

print(gcd(48, 18))   # 6
```

```python
def binary_search(nums, target, lo, hi):
    if lo > hi:
        return -1
    mid = (lo + hi) // 2
    if nums[mid] == target:
        return mid
    if nums[mid] < target:
        return binary_search(nums, target, mid + 1, hi)
    return binary_search(nums, target, lo, mid - 1)

nums = [1, 3, 5, 7, 9, 11]
print(binary_search(nums, 7, 0, len(nums) - 1))   # 3

def is_power_of_two(n):
    if n <= 0:
        return False
    if n == 1:
        return True
    if n % 2 == 1:
        return False
    return is_power_of_two(n // 2)

print(is_power_of_two(16), is_power_of_two(18))   # True False
```

## 3. Shape C: divide and conquer (split into two halves)

Split the problem into two halves, solve each recursively, combine the answers. You already saw this in **merge sort**.

Example: maximum of a list. `max(whole) = max(max(left half), max(right half))`.

```python
def max_dc(nums, lo, hi):
    if lo == hi:                       # one element
        return nums[lo]
    mid = (lo + hi) // 2
    return max(max_dc(nums, lo, mid), max_dc(nums, mid + 1, hi))

nums = [4, 9, 2, 7, 5]
print(max_dc(nums, 0, len(nums) - 1))   # 9
```

## 4. Passing information down: index and extra parameters

- Use an **index** (`i`, or `l` and `r`) instead of slicing. `nums[1:]` copies the list every call → O(n²) overall.
- Pass an **accumulator** (running total, current path) as a parameter when the answer is built on the way down.
- A recursive function can **return** a value (build the answer on the way up) or **fill a shared list** (used in backtracking).

## 5. Interview thinking

1. Say the shape: "this shrinks by one / by half / splits in two".
2. State the base case first.
3. Say the depth: linear = n, halving = log n, splitting in two halves = log n.
4. If asked for an iterative version: halving and linear recursions convert easily to loops.

**Common mistakes:** calling the half-power twice, wrong base case for `n = 0` or negatives, off-by-one in `lo/hi`, slicing in every call.

## Quick Revision

- Shrink by 1: depth n (factorial, sum, digits, palindrome).
- Shrink by half: depth log n (fast power, binary search, gcd). Compute the half **once**.
- Divide and conquer: solve both halves, combine (max, merge sort).
- Use indexes, not slices.
