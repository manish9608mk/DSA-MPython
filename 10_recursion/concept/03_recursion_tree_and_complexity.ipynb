# 03. Recursion Tree, Complexity and Memoization

To find the time of a recursive function, **draw its recursion tree**.

**Learning order:** TREE IDEA → FIBONACCI TREE → COUNT CALLS → MEMOIZATION → RECURRENCE TABLE → LOOP VERSION → INTERVIEW THINKING

## 1. The recursion tree

Every call is a node. Its recursive calls are its children.

- **Time** = (number of nodes) × (work done in each node outside the recursive calls).
- **Space** = height of the tree (the stack holds one root-to-leaf path at a time).

Fibonacci tree for `fib(4)`:

```
              fib(4)
            /        \
      fib(3)          fib(2)
      /    \          /    \
  fib(2)  fib(1)   fib(1) fib(0)
  /    \
fib(1) fib(0)
```

`fib(2)` is computed **twice**, `fib(1)` three times. The same work repeats.

````python
calls = 0

def fib_naive(n):
    global calls
    calls += 1
    if n <= 1:
        return n
    return fib_naive(n - 1) + fib_naive(n - 2)

for n in (5, 10, 20):
    calls = 0
    fib_naive(n)
    print(n, "calls:", calls)    # 15, 177, 21891  -> roughly doubles with each n
````

## 2. Memoization (remember answers)

Store the result of each `fib(n)` the first time. Next time, return the stored value. Every `n` is computed once → **O(n) time**, O(n) space.

This idea is the doorway to **Dynamic Programming**.

````python
def fib_memo(n, memo=None):
    if memo is None:
        memo = {}
    if n <= 1:
        return n
    if n in memo:
        return memo[n]
    memo[n] = fib_memo(n - 1, memo) + fib_memo(n - 2, memo)
    return memo[n]

print(fib_memo(50))    # 12586269025  (naive version would take forever)

from functools import lru_cache

@lru_cache(maxsize=None)
def fib_cache(n):
    return n if n <= 1 else fib_cache(n - 1) + fib_cache(n - 2)

print(fib_cache(50))   # 12586269025
````

## 3. Recurrence relation table

| Recurrence | Time | Example |
|---|---|---|
| `T(n) = T(n-1) + O(1)` | O(n) | factorial, sum of list |
| `T(n) = T(n/2) + O(1)` | O(log n) | binary search, fast power |
| `T(n) = 2T(n/2) + O(1)` | O(n) | max by divide and conquer |
| `T(n) = 2T(n/2) + O(n)` | O(n log n) | merge sort |
| `T(n) = T(n-1) + T(n-2)` | about O(1.6ⁿ), say O(2ⁿ) | naive Fibonacci |
| `T(n) = 2T(n-1) + O(1)` | O(2ⁿ) | subsets, Tower of Hanoi |

**Space** in all cases = recursion depth: n, log n, n, log n (+ O(n) for merge), n, n, n.

## 4. Recursion vs loop

Any recursion can be written as a loop (with your own stack if needed). Loops avoid the stack limit and the call overhead. Recursion is shorter and clearer when the problem is naturally nested (trees, subsets, divide and conquer).

````python
def fib_loop(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a

print(fib_loop(50))   # 12586269025   (O(n) time, O(1) space)
````

## 5. Interview thinking

**Say:** "The recursion tree has about ... nodes, each doing O(...) work, so time is ... Depth is ..., so space is ..."

**Red flag to mention:** "The same subproblem appears many times, so I can add memoization and get O(n)."

**Common mistakes:** saying space is O(1) for recursion (it is O(depth)), forgetting memo dictionary as a mutable default argument, ignoring the Python recursion limit for n ≈ 10⁵.

## Quick Revision

- Time = nodes in the tree × work per node. Space = tree height.
- Naive `fib` = O(2ⁿ) because of repeated subproblems.
- Memoization stores answers → O(n). This is where DP starts.
- Memorize: `T(n)=T(n-1)+O(1)` is O(n), `T(n/2)+O(1)` is O(log n), `2T(n/2)+O(n)` is O(n log n), `2T(n-1)` is O(2ⁿ).
- Any recursion can be a loop.
