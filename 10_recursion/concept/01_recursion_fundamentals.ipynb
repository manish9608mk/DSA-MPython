# 01. Recursion: Fundamentals

Recursion means a function **calls itself** on a **smaller version of the same problem**, until the problem is so small that the answer is obvious.

**Learning order:** REAL LIFE → TWO PARTS → CALL STACK → TRACE → 3-QUESTION FRAMEWORK → EXAMPLES → LIMITS → INTERVIEW THINKING

## 1. Real-life examples

- **Russian dolls:** to find the smallest doll, open the doll. If there is another doll inside, "find the smallest doll" in that one. If the doll is empty inside, you found it.
- **Folders in folders:** to count all files, count files in this folder, then do the same for every sub-folder.
- **Queue in front of you:** "how many people are ahead of me?" Ask the person in front. They ask the person in front of them. The first person says 0. Answers come back one by one, each adds 1.

## 2. Every recursive function has two parts

1. **Base case:** the smallest problem, answered directly. It stops the recursion.
2. **Recursive case:** reduce the problem to a **smaller** one, call yourself, and use the result.

If the base case is missing, or the problem does not get smaller, the function calls itself forever → `RecursionError` (stack overflow).

````python
def factorial(n):
    if n == 0:                      # base case
        return 1
    return n * factorial(n - 1)     # recursive case: smaller problem n-1

print(factorial(5))   # 120
````

## 3. The call stack (what really happens)

Each call gets its own **frame** (its own `n`, waiting for the answer of the next call). Frames pile up like plates:

```
factorial(4)  waits for factorial(3)
  factorial(3)  waits for factorial(2)
    factorial(2)  waits for factorial(1)
      factorial(1)  waits for factorial(0)
        factorial(0) -> 1          # base case, starts returning
      factorial(1) -> 1 * 1 = 1
    factorial(2) -> 2 * 1 = 2
  factorial(3) -> 3 * 2 = 6
factorial(4) -> 4 * 6 = 24
```

Calls go **down** (stack grows), then answers come **back up** (stack unwinds).

````python
def factorial_trace(n, depth=0):
    pad = "  " * depth
    print(f"{pad}factorial({n}) called")
    if n == 0:
        print(f"{pad}base case -> 1")
        return 1
    result = n * factorial_trace(n - 1, depth + 1)
    print(f"{pad}factorial({n}) returns {result}")
    return result

factorial_trace(4)
````

## 4. The 3-question framework (use it for every recursion problem)

1. **What does `f(input)` mean?** Write it in one sentence. (`factorial(n)` = n × (n-1) × ... × 1)
2. **What is the smallest input** I can answer directly? → base case.
3. **If I trust `f(smaller)` to be correct, how do I build `f(input)` from it?** → recursive case.

This trust is called the **leap of faith**. Don't trace every call in your head. Trust the smaller call, and only check the base case and the combining step.

````python
# Sum of a list, without slicing: sum of nums[i:]
def sum_list(nums, i=0):
    if i == len(nums):                 # nothing left
        return 0
    return nums[i] + sum_list(nums, i + 1)

print(sum_list([3, 1, 4, 1, 5]))   # 14

# Reverse a list in place with two indexes
def reverse_list(arr, l=0, r=None):
    if r is None:
        r = len(arr) - 1
    if l >= r:                         # 0 or 1 element: already reversed
        return arr
    arr[l], arr[r] = arr[r], arr[l]
    return reverse_list(arr, l + 1, r - 1)

print(reverse_list([1, 2, 3, 4, 5]))   # [5, 4, 3, 2, 1]
````

````python
# Two recursive calls: Fibonacci
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)

print(fib(10))   # 55
````

## 5. Limits you must know

- **Space:** each active call uses stack memory. Depth `d` → **O(d) space**, even if you store nothing yourself.
- **Python limit:** about 1000 nested calls by default. Deeper → `RecursionError`.
- **Python does not optimize tail calls**, so a loop is safer for very deep problems.

````python
import sys
print(sys.getrecursionlimit())     # 1000 (default)

def forever(n):                    # no base case
    return forever(n + 1)

try:
    forever(0)
except RecursionError:
    print("RecursionError: no base case / too deep")
````

## 6. Interview thinking

**Signals:** the problem contains a **smaller copy of itself** (tree, nested structure, "n depends on n-1"), choices at each step, divide and conquer.

**Say:** "The base case is ... The recursive case reduces the input by ... and combines ... Recursion depth is ..., so space is O(depth)."

**Common mistakes:** missing or wrong base case, input does not shrink, forgetting to `return` the recursive call, changing a shared list/global without undoing it, using slicing inside recursion (each slice costs O(n)).

## Quick Revision

- Recursion = function calls itself on a **smaller** input.
- Base case stops it. Recursive case shrinks the problem.
- Call stack: calls go down, answers come back up.
- Framework: define `f`, base case, trust the smaller call and combine.
- Space = recursion depth. Python limit ~1000.
