# 02. Fixed Size Window

The window length is given (`k`, or the length of a pattern). Every step: **add one element on the right, remove one on the left**.

**Learning order:** TEMPLATE → AVERAGE → DUPLICATE IN WINDOW → ANAGRAMS → PERMUTATION → INTERVIEW THINKING

## 1. Template

```python
# build first window
for i in range(k):
    add(arr[i])
answer = current_value()

# slide
for right in range(k, len(arr)):
    add(arr[right])          # enters
    remove(arr[right - k])   # leaves
    answer = update(answer)
```

Always handle `len(arr) < k` first if the problem allows it.

## 2. Maximum Average Subarray

Same as max sum. The maximum average is `max sum / k` because `k` is constant.

````python
def max_average(nums, k):
    window_sum = sum(nums[:k])
    best = window_sum
    for right in range(k, len(nums)):
        window_sum += nums[right] - nums[right - k]
        best = max(best, window_sum)
    return best / k

print(max_average([1, 12, -5, -6, 50, 3], 4))   # 12.75
````

## 3. Contains Nearby Duplicate

Is there a value that appears twice with distance at most `k`? Keep a **set of the last `k` elements** as the window. If the new element is already in the set, we found it.

Real life: "has the same customer ordered twice within the last `k` orders?"

````python
def contains_nearby_duplicate(nums, k):
    window = set()
    for i, x in enumerate(nums):
        if x in window:
            return True
        window.add(x)
        if len(window) > k:
            window.remove(nums[i - k])   # oldest element leaves
    return False

print(contains_nearby_duplicate([1, 2, 3, 1], 3))         # True
print(contains_nearby_duplicate([1, 2, 3, 1, 2, 3], 2))   # False
````

## 4. Find All Anagrams in a String

**Problem:** find all start indexes of substrings of `s` that are anagrams of `p`.

An anagram has the **same letter counts**. Window size = `len(p)`. Slide it, update counts for the entering and leaving character, and compare with `p`'s counts.

Comparing two counters costs O(26) which is constant for lowercase letters.

````python
from collections import Counter

def find_anagrams(s, p):
    n, m = len(s), len(p)
    if m > n:
        return []
    need = Counter(p)
    window = Counter(s[:m])
    result = []
    if window == need:
        result.append(0)
    for right in range(m, n):
        window[s[right]] += 1                 # enters
        out = s[right - m]
        window[out] -= 1                      # leaves
        if window[out] == 0:
            del window[out]                   # keep counters comparable
        if window == need:
            result.append(right - m + 1)
    return result

print(find_anagrams("cbaebabacd", "abc"))   # [0, 6]
````

## 5. Permutation in String

Does `s2` contain a permutation of `s1`? Same as above, return `True` at the first match. A permutation = an anagram.

````python
def check_inclusion(s1, s2):
    m = len(s1)
    if m > len(s2):
        return False
    need = Counter(s1)
    window = Counter(s2[:m])
    if window == need:
        return True
    for right in range(m, len(s2)):
        window[s2[right]] += 1
        out = s2[right - m]
        window[out] -= 1
        if window[out] == 0:
            del window[out]
        if window == need:
            return True
    return False

print(check_inclusion("ab", "eidbaooo"))   # True
print(check_inclusion("ab", "eidboaoo"))   # False
````

## 6. Interview thinking

**Signals:** "subarray / substring of size k", "every window of length k", "anagram / permutation of a fixed pattern".

**Say:** "The window size is fixed, so I slide it one step at a time: one element enters, one leaves, and I update the state in O(1). Total O(n)."

**Common mistakes:** wrong first window, off-by-one in `right - k` (leaving index), forgetting to delete zero counts before comparing counters, not handling `len < k`.

**Preview:** *Sliding Window Maximum* (max of every window) needs a deque. We will do it in **Queue + Deque**.

## Quick Revision

- Build first window, then: add `arr[right]`, remove `arr[right - k]`.
- Max sum / average: keep running sum.
- Duplicate within distance k: set of last k elements.
- Anagram / permutation: counters, delete zero counts, compare with pattern counter.
- Leaving index = `right - k` (window `[right-k+1 .. right]`).
