## 1. The four patterns

| Pattern | Question looks like | Shrink rule | Record |
|---|---|---|---|
| **Fixed** | window of size `k` | slide by one | every window |
| **Longest** | longest with condition | `while invalid` | after the loop: `right-left+1` |
| **Shortest** | shortest with condition | `while valid` | inside the loop |
| **Counting** | number of subarrays | `while invalid` | `count += right-left+1` |

## 2. When sliding window fails: negative numbers

"Subarray with sum equal to `k`" with negative numbers: adding an element can **decrease** the sum and removing one can **increase** it. We can't decide whether to expand or shrink. The window idea breaks.

Use **prefix sum + hash map** instead: `prefix[j] - prefix[i] = k` → look for `prefix[j] - k` seen before.

```python
from collections import defaultdict

def subarray_sum_equals_k(nums, k):
    seen = defaultdict(int)
    seen[0] = 1                  # empty prefix
    prefix = 0
    count = 0
    for x in nums:
        prefix += x
        count += seen[prefix - k]
        seen[prefix] += 1
    return count

print(subarray_sum_equals_k([1, 1, 1], 2))    # 2
print(subarray_sum_equals_k([1, -1, 0], 0))   # 3  (works with negatives)
```

## 3. Sliding window vs similar tools

| Tool | Use when |
|---|---|
| **Sliding window** | contiguous subarray/substring, a rule that changes predictably when you add or remove at the ends |
| **Two pointers (opposite ends)** | sorted array, pair or triplet, compare two ends |
| **Prefix sum + hash** | subarray sum with negatives, "equals k" |
| **Hashing alone** | non-contiguous pairs, frequencies, indexes |
| **Deque** | max or min of each window (Sliding Window Maximum, next in Queue + Deque) |

## 4. Quick decision flow

1. Is it a **contiguous** subarray or substring? If no → not a window.
2. Is the window **size given**? → fixed window.
3. Is it **longest**? → expand, `while invalid` shrink, record after.
4. Is it **shortest**? → `while valid` shrink, record inside.
5. Is it **count of subarrays**? → `right - left + 1`; for exactly k use at-most trick.
6. Does the rule depend on a **sum with negatives**? → prefix sum + hash.

## 5. Common mistakes

- `if` instead of `while` when shrinking.
- Recording `best` at the wrong place (before the window is valid).
- Not deleting a key whose count becomes 0.
- Off-by-one: window length is `right - left + 1`.
- Using shortest template on negative numbers.
- Not defining "valid" before coding.
- Forgetting to handle empty input or `k > len`.

## 6. Interview script

1. "Brute force checks all subarrays, O(n²) or worse."
2. "The subarray is contiguous and the condition changes predictably, so I can use a sliding window."
3. "I define the window as valid when ... (say the exact rule)."
4. "I expand `right`. When the window becomes invalid, I shrink `left`."
5. "Each element enters and leaves the window once → O(n) time, O(1) or O(alphabet) space."
6. Test: empty, size 1, all same, all different, `k` larger than the array.

## Quick Revision

- Fixed: add `right`, remove `right-k`.
- Longest: `while invalid` shrink, record after.
- Shortest: `while valid` shrink, record inside.
- Count: add `right-left+1`; exactly k = atMost(k) − atMost(k−1).
- Negatives with sums → prefix sum + hash, not window.
- Always say what **valid** means.
