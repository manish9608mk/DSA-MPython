# 01. Sliding Window: Fundamentals

A **window** is a contiguous part of an array or string, `arr[left..right]`. Sliding window means we **move this window** forward, and update the answer by adding the new element and removing the old one. We never recompute the whole window from scratch.

**Learning order:** WHAT → REAL LIFE → BRUTE FORCE PROBLEM → THE TRICK → WHY O(n) → 2 TYPES → CLUES → INTERVIEW THINKING

## 1. Real-life examples

- **Moving average on a stock app:** "average of the last 7 days". Tomorrow, you add the new day and drop the oldest day. You do not recompute all 7.
- **CCTV storage:** "keep the last 24 hours of footage". Each minute, new footage enters, old footage leaves.
- **Internet speed meter:** speed over the last 5 seconds, updated every second.
- **Longest streak:** "longest run of days where I studied with at most 1 break".

## 2. The problem

Given an array and `k`, find the maximum sum of **any `k` consecutive elements**.

`[2, 1, 5, 1, 3, 2]`, `k = 3` → windows: `[2,1,5]=8`, `[1,5,1]=7`, `[5,1,3]=9`, `[1,3,2]=6` → answer **9**.

```python
# Brute force: recompute every window from scratch -> O(n * k)
def max_sum_brute(nums, k):
    best = float("-inf")
    for i in range(len(nums) - k + 1):
        best = max(best, sum(nums[i:i + k]))
    return best

print(max_sum_brute([2, 1, 5, 1, 3, 2], 3))   # 9
```

## 3. The trick

Consecutive windows share almost all elements. Moving from `[2,1,5]` to `[1,5,1]`:

`new sum = old sum + (element that enters) - (element that leaves)`

So each step is **O(1)** instead of O(k).

```python
def max_sum_window(nums, k):
    window_sum = sum(nums[:k])          # first window
    best = window_sum
    for right in range(k, len(nums)):
        window_sum += nums[right]       # element enters
        window_sum -= nums[right - k]   # element leaves
        best = max(best, window_sum)
    return best

print(max_sum_window([2, 1, 5, 1, 3, 2], 3))   # 9
```

```python
# Trace: see the window slide
nums, k = [2, 1, 5, 1, 3, 2], 3
window_sum = sum(nums[:k])
print(f"window {nums[0:k]} sum = {window_sum}")
for right in range(k, len(nums)):
    window_sum += nums[right] - nums[right - k]
    left = right - k + 1
    print(f"window {nums[left:right + 1]} sum = {window_sum}")
```

## 4. Why is it O(n)?

Every element **enters the window once** and **leaves once**. Total work = about 2n = **O(n)**. Brute force does O(n × k).

**Space:** O(1) for sums. If you keep counts of characters, O(size of alphabet).

## 5. Two types of windows

| Type | Window size | Example question |
|---|---|---|
| **Fixed size** | given `k` | max sum of `k` elements, anagram of length `m` |
| **Variable size** | grows and shrinks | longest substring without repeats, shortest subarray with sum ≥ target |

Fixed window moves like a train of constant length. Variable window uses two pointers: `right` **expands**, `left` **shrinks** when the window becomes invalid (or is already valid and can be smaller).

## 6. Clues that tell you to use sliding window

- "**subarray**" or "**substring**" (contiguous!)
- "longest / shortest / maximum / minimum ... with a condition"
- "at most `k` distinct", "at most `k` zeros", "sum ≥ target"
- "every window of size `k`"
- Brute force checks all `(i, j)` pairs → O(n²)

**Not a window problem:** subsequences (not contiguous), or arrays with **negative numbers** when the rule is based on a sum (see notebook 05).

## 7. Interview thinking

1. "Brute force tries every subarray, O(n²) or O(n·k)."
2. "Consecutive windows overlap, so I can update the answer by adding one element and removing one."
3. Decide: is the size **fixed** or **variable**?
4. For variable: define clearly **what makes the window valid**.
5. "Each element enters and leaves once, so O(n)."

## Quick Revision

- Window = contiguous `arr[left..right]`, slide it, don't recompute.
- Update = add entering element, subtract leaving element.
- Fixed size (given `k`) vs variable size (two pointers, expand/shrink).
- O(n) because each element enters once and leaves once.
- Clue words: subarray, substring, longest, shortest, at most k.
