# 01. Binary Tree: Fundamentals

A tree is a **hierarchy**: one **root**, and every node has **children**. In a **binary** tree, each node has at most **2** children: `left` and `right`.

**Learning order:** REAL LIFE → TERMS → NODE → BUILD → HEIGHT AND SIZE → TYPES → ARRAY FORM → INTERVIEW THINKING

## 1. Real-life examples

- **Folders on your laptop:** a folder contains folders and files.
- **Company org chart**, **family tree**.
- **HTML page (DOM):** `<html>` contains `<body>` contains `<div>`...
- **Decision tree:** each question splits into yes/no.
- **Expression tree:** `(2 + 3) * 4` is a tree with `*` at the root.

Trees are used when data is **hierarchical** or when we want **fast search with ordering** (BST, heaps, tries, databases).

## 2. Terms (know these by heart)

| Term | Meaning |
|---|---|
| root | top node (no parent) |
| leaf | node with no children |
| parent / child / sibling | direct relations |
| depth of a node | edges from the root to it (root = 0) |
| height of a node | edges on the longest path down to a leaf (leaf = 0) |
| height of the tree | height of the root |
| level | all nodes with the same depth |
| subtree | a node and all its descendants |

Many problems define height in **nodes** (empty tree = 0, single node = 1). Always read the definition in the question.

````python
from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val, self.left, self.right = val, left, right

def tree_from_list(vals):             # level order, None = missing child
    if not vals or vals[0] is None:
        return None
    root = TreeNode(vals[0])
    q, i = deque([root]), 1
    while q and i < len(vals):
        node = q.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i]); q.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i]); q.append(node.right)
        i += 1
    return root

def level_order(root):
    if not root:
        return []
    q, res = deque([root]), []
    while q:
        res.append([])
        for _ in range(len(q)):
            node = q.popleft()
            res[-1].append(node.val)
            if node.left: q.append(node.left)
            if node.right: q.append(node.right)
    return res

def inorder(root):
    return inorder(root.left) + [root.val] + inorder(root.right) if root else []

root = TreeNode(1, TreeNode(2, TreeNode(4), TreeNode(5)), TreeNode(3))
print(level_order(root))   # [[1], [2, 3], [4, 5]]

same = tree_from_list([1, 2, 3, 4, 5])
print(level_order(same))   # [[1], [2, 3], [4, 5]]
````

## 3. Height, size, leaves (first recursive thinking)

Trees are **recursive**: a tree = root + left tree + right tree. So most functions look like:

```
f(None) = base answer
f(node) = combine(node.val, f(node.left), f(node.right))
```

````python
def height(root):                              # in nodes: empty = 0
    if not root:
        return 0
    return 1 + max(height(root.left), height(root.right))

def count_nodes(root):
    if not root:
        return 0
    return 1 + count_nodes(root.left) + count_nodes(root.right)

def count_leaves(root):
    if not root:
        return 0
    if not root.left and not root.right:
        return 1
    return count_leaves(root.left) + count_leaves(root.right)

print(height(root), count_nodes(root), count_leaves(root))   # 3 5 3
````

## 4. Types of binary trees

| Type | Meaning |
|---|---|
| **Full** | every node has 0 or 2 children |
| **Complete** | all levels filled except possibly the last, which is filled left to right (heaps) |
| **Perfect** | all leaves at the same level, `2^h - 1` nodes for height `h` (in nodes) |
| **Balanced** | height of left and right subtrees differ by at most 1 at every node |
| **Skewed** | every node has one child (like a linked list) |

**Why balance matters:** a balanced tree has height **O(log n)**; a skewed one has height **O(n)**. Operations on a BST cost O(height).

## 5. Array representation (complete trees)

For node at index `i` (0-based): left child `2i + 1`, right child `2i + 2`, parent `(i - 1) // 2`. This is how **heaps** are stored (no pointers needed).

````python
arr = [1, 2, 3, 4, 5]
def children(i, n):
    return [c for c in (2 * i + 1, 2 * i + 2) if c < n]

print(children(0, 5), children(1, 5), children(2, 5))   # [1, 2] [3, 4] []
print((4 - 1) // 2)                                      # 1  (parent of index 4)
````

## 6. Useful facts

- A tree with `n` nodes has exactly `n - 1` edges.
- A binary tree of height `h` (in edges) has at most `2^(h+1) - 1` nodes.
- Level `d` has at most `2^d` nodes.
- Time for tree traversal is **O(n)**; recursion space is **O(h)** (O(log n) balanced, O(n) skewed).

## 7. Interview thinking

**First questions to ask:** Is it a binary tree or a BST? Can it be empty? Can values repeat? Is height counted in nodes or edges?

**Say:** "A tree is recursive, so I'll solve for the left and right subtrees and combine at the root. Time O(n), space O(h)."

**Common mistakes:** forgetting the `None` base case, confusing depth vs height, assuming the tree is balanced, mixing node and edge counts.

## Quick Revision

- Tree = root + left subtree + right subtree. Think recursively.
- Height (longest path down), depth (distance from root), leaf, level.
- Balanced = O(log n) height, skewed = O(n).
- Complete tree in an array: children `2i+1`, `2i+2`.
- Traversal O(n) time, O(h) stack.
