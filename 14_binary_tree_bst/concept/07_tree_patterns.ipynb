## 1. Decision table

| Question looks like | Pattern |
|---|---|
| height, depth, count, sum | postorder / bottom-up return |
| root-to-leaf path, path with a target | top-down parameter, check at a leaf |
| diameter, max path sum | bottom-up + global `best` |
| balanced? | return height with `-1` sentinel |
| symmetric / same / subtree | recursion on two nodes at once |
| LCA | both sides non-null → this node |
| level, view, width, min depth | BFS |
| sorted order, k-th, range in BST | inorder, BST pruning |
| build from traversals | root from preorder/postorder, split with inorder |
| serialize | preorder with null markers |
| "choose or skip" on a tree (no adjacent) | tree DP returning a tuple |

## 2. Tree DP: return a tuple

**House Robber III:** a thief cannot rob two **directly linked** houses (parent and child). For every node return two numbers: `(best if we rob this node, best if we skip it)`.

- Rob this node: `val + skip(left) + skip(right)`.
- Skip this node: `max(rob, skip)` of left + `max(rob, skip)` of right.

````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val, self.left, self.right = val, left, right

def build(vals):
    from collections import deque
    if not vals: return None
    root = TreeNode(vals[0]); q, i = deque([root]), 1
    while q and i < len(vals):
        node = q.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i]); q.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i]); q.append(node.right)
        i += 1
    return root

def rob(root):
    def dfs(node):
        if not node:
            return (0, 0)                         # (rob this, skip this)
        l, r = dfs(node.left), dfs(node.right)
        rob_this = node.val + l[1] + r[1]
        skip_this = max(l) + max(r)
        return (rob_this, skip_this)
    return max(dfs(root))

print(rob(build([3, 2, 3, None, 3, None, 1])), rob(build([3, 4, 5, 1, 3, None, 1])))   # 7 9
````


## 4. Common mistakes

- Missing `None` base case.
- Treating a node with one child as a leaf.
- Returning the answer when the parent needs a different value (diameter/max path).
- Recomputing height inside recursion (O(n²)).
- Validating a BST with only local checks.
- Using slicing in tree construction.
- Forgetting recursion depth (O(n) skewed trees; Python limit ~1000).

## 5. Interview script

1. Clarify: binary tree or BST? empty tree? duplicates? definition of height/leaf?
2. "I'll solve it recursively. Base case `None`. For each node I need ... from the children."
3. State what the function **returns** and what the **answer** is (if different).
4. "Time O(n), space O(h)."
5. Test: empty, single node, skewed, perfect tree, negative values.
