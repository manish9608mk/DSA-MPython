# 11. From Constraints to the Required Complexity

Before writing code, look at the **input limits**. They tell you which complexity the problem expects, and therefore which algorithm family to use.

**Learning order:** OPERATIONS BUDGET → CONSTRAINT TABLE → EXAMPLES → ESTIMATE IN CODE → AMORTIZED THINKING → INTERVIEW THINKING

## 1. The operations budget

A typical judge allows about **10⁸ simple operations** per second in C++/Java. **Python is 20 to 50 times slower**, so assume about **10⁶ to 10⁷ simple operations** per second in a tight Python loop (built-ins like `sort`, `sum`, `Counter` run in C and are faster).

In interviews you are not timed by a machine, but the interviewer expects a solution that **would pass** for the stated limits.

## 2. Constraint → target complexity

| Max `n` | Expected complexity | Typical algorithms |
|---|---|---|
| ≤ 10 | O(n!) , O(n⁶) | permutations, brute-force backtracking |
| ≤ 20 to 25 | O(2ⁿ · n) | subsets, bitmask DP, meet in the middle |
| ≤ 100 | O(n⁴) | DP with 3 to 4 dimensions |
| ≤ 400 to 500 | O(n³) | Floyd-Warshall, interval DP |
| ≤ 2000 to 5000 | O(n²) | 2D DP (LCS, edit distance), pairs |
| ≤ 10⁵ | **O(n log n)** | sorting, heap, binary search, segment tree, divide and conquer |
| ≤ 10⁶ | **O(n)** (or n log n with a tiny constant) | hashing, two pointers, sliding window, prefix sum |
| ≤ 10⁹ to 10¹⁸ | O(log n) or O(√n) | binary search, math, fast exponentiation, matrix power |

**Also:** values up to 10⁹ with `n` small usually mean you must avoid loops over the value range.

```python
def target_complexity(n):
    if n <= 10:
        return "O(n!)"
    if n <= 25:
        return "O(2^n * n)"
    if n <= 100:
        return "O(n^4)"
    if n <= 500:
        return "O(n^3)"
    if n <= 5000:
        return "O(n^2)"
    if n <= 10**5:
        return "O(n log n)"
    if n <= 10**6:
        return "O(n)"
    return "O(log n) or O(sqrt n)"

for n in (8, 20, 300, 4000, 10**5, 10**6, 10**9):
    print(n, target_complexity(n))
# 8 O(n!) / 20 O(2^n * n) / 300 O(n^3) / 4000 O(n^2) / 100000 O(n log n) / 1000000 O(n) / 1000000000 O(log n) or O(sqrt n)
```

## 3. Examples: reading the constraint

- "`n ≤ 10⁵`, find two numbers with sum `target`" → O(n²) = 10¹⁰ is too slow. Use a hash map (O(n)) or sort + two pointers (O(n log n)).
- "`n ≤ 20`, split into two groups with the smallest difference" → `2²⁰ ≈ 10⁶`: try all subsets (bitmask).
- "`n ≤ 1000`, longest common subsequence of two strings" → O(n²) DP = 10⁶: fine.
- "`n ≤ 10⁹`, count primes below `n`" → impossible with a plain sieve in memory; think math/segmented sieve, or the real limit is smaller.
- "`k ≤ n ≤ 10⁵`, top k frequent" → O(n log k) heap or O(n) bucket sort, not O(n²).
- "`m, n ≤ 10³` grid" → O(m·n) = 10⁶ BFS/DP is fine. O((m·n)²) is not.

```python
# A quick estimator: how many basic operations does a complexity produce for a given n?
import math

def ops(n, complexity):
    table = {
        "n": n,
        "n log n": n * math.log2(n),
        "n^2": n ** 2,
        "n^3": n ** 3,
        "2^n": 2 ** n,
    }
    return table[complexity]

print(f"{ops(10**5, 'n^2'):.0e}", f"{ops(10**5, 'n log n'):.0e}")   # 1e+10 2e+06
print(ops(20, "2^n"), ops(1000, "n^3") > 10**8)   # 1048576 True
```

## 4. Amortized analysis in one page

An operation that is **rarely expensive but usually cheap** can have a cheap **average over a sequence**.

- **`list.append`:** when the array is full, we copy it (O(n)) and double the capacity. Over `n` appends the total copy work is `1 + 2 + 4 + ... + n ≤ 2n`, so each append is **O(1) amortized**.
- **Monotonic stack / deque:** each element is pushed once and popped once → O(n) total even though a single step may pop many.
- **Two pointers / sliding window:** each pointer moves at most `n` times.
- **Union-Find with path compression:** nearly O(1) amortized.

**Not the same as average-case:** amortized is a **guarantee** over any sequence of operations; average-case depends on the input distribution.

## 5. Interview thinking

1. Read the **constraints first** and name the target complexity out loud: "n is 10⁵ so I need O(n log n) or better."
2. Reject brute force with numbers: "O(n²) is 10¹⁰ operations, too slow."
3. State **time and space** for the final solution, plus the recursion depth if any.
4. If an interviewer asks for better, ask which resource matters: time, space, or both.

**Common mistakes:** coding before reading limits, ignoring the constant factor in Python (nested loops at 10⁷ are risky), forgetting recursion depth, saying "O(1)" for operations that copy.

## Quick Revision

- Python ≈ 10⁶ to 10⁷ simple ops/s. Plan for ~10⁷ total.
- `n ≤ 20` exponential, `≤ 5000` n², `≤ 10⁵` n log n, `≤ 10⁶` n, `≤ 10⁹` log n.
- Amortized = cheap on average over a sequence (guaranteed), not average-case.
- Say complexity **and** the reason out loud.
