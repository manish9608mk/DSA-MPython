# 03. Counting Bits and Bitmasks

A **bitmask** is an integer used as a **set of yes/no flags**. With `n` items there are `2ⁿ` masks, one per subset. This is perfect when `n` ≤ about 20.

**Learning order:** COUNTING BITS → HAMMING → REVERSE BITS → SUBSETS WITH MASKS → GRAY CODE → SINGLE NUMBER II → MAX PRODUCT OF WORD LENGTHS → INTERVIEW THINKING

## 1. Counting Bits for 0..n

`bits[i] = bits[i >> 1] + (i & 1)`: the number of set bits of `i` equals that of `i` without its last bit, plus that last bit. O(n) total.

```python
def count_bits(n):
    bits = [0] * (n + 1)
    for i in range(1, n + 1):
        bits[i] = bits[i >> 1] + (i & 1)
    return bits

print(count_bits(5))   # [0, 1, 1, 2, 1, 2]
```

## 2. Hamming Distance and Reverse Bits

- **Hamming distance** = number of positions where bits differ = set bits of `x ^ y`.
- **Reverse bits** of a 32-bit number: build the result by shifting it left and taking the last bit of the input each time.

```python
def hamming_distance(x, y):
    return bin(x ^ y).count("1")

def reverse_bits(n):
    result = 0
    for _ in range(32):
        result = (result << 1) | (n & 1)
        n >>= 1
    return result

print(hamming_distance(1, 4), reverse_bits(43261596))   # 2 964176192
```

## 3. Subsets with bitmasks

For `n` items, loop `mask` from `0` to `2ⁿ - 1`. Bit `i` of the mask says "include item `i`". Real life: all combinations of 3 toppings = 8 masks `000..111`.

No recursion needed. Time O(n · 2ⁿ).

```python
def subsets_bitmask(nums):
    n = len(nums)
    return [[nums[i] for i in range(n) if mask >> i & 1] for mask in range(1 << n)]

print(sorted(subsets_bitmask([1, 2, 3])))   # [[], [1], [1, 2], [1, 2, 3], [1, 3], [2], [2, 3], [3]]
```

## 4. Gray Code

A sequence where consecutive numbers differ in **exactly one bit**. Formula: `i ^ (i >> 1)`. Used in rotary encoders and error-free counters.

```python
def gray_code(n):
    return [i ^ (i >> 1) for i in range(1 << n)]

print(gray_code(2), gray_code(3))   # [0, 1, 3, 2] [0, 1, 3, 2, 6, 7, 5, 4]
```

## 5. Single Number II (others appear 3 times)

Count each bit position modulo 3, or use two variables:
- `ones` = bits seen once (mod 3), `twos` = bits seen twice.
- When a bit appears the third time it is removed from both.

```python
def single_number_ii(nums):
    ones = twos = 0
    for x in nums:
        ones = (ones ^ x) & ~twos
        twos = (twos ^ x) & ~ones
    return ones

print(single_number_ii([2, 2, 3, 2]), single_number_ii([0, 1, 0, 1, 0, 1, 99]))   # 3 99
```

## 6. Maximum Product of Word Lengths

Find two words with **no common letters** maximizing `len(a) * len(b)`. Represent each word as a **26-bit mask** (bit `c` = letter present). Two words share no letter iff `mask_a & mask_b == 0`. A pair check becomes O(1).

```python
def max_product(words):
    masks = []
    for w in words:
        m = 0
        for ch in w:
            m |= 1 << (ord(ch) - ord("a"))
        masks.append(m)
    best = 0
    for i in range(len(words)):
        for j in range(i + 1, len(words)):
            if masks[i] & masks[j] == 0:
                best = max(best, len(words[i]) * len(words[j]))
    return best

print(max_product(["abcw", "baz", "foo", "bar", "xtfn", "abcdef"]))   # 16
```

## 7. Bitmask DP preview

When `n ≤ 20`, a **mask can be the state** of DP: "which items are already used". Examples: Traveling Salesman, assigning tasks to people, partition into groups. This is covered in the Dynamic Programming topic.

## 8. Interview thinking

**Signals:** "n ≤ 20", "all subsets", "set of letters / flags", "differ in one bit", "number of 1 bits".

**Say:** "I represent each set as an integer mask so union is `|`, intersection `&`, membership `(mask >> i) & 1`, and I can enumerate all 2ⁿ subsets with a simple loop."

**Common mistakes:** `1 << n` for `n` up to 30 is fine in Python but 32+ overflows in other languages, precedence bugs (`mask >> i & 1` is OK, but `mask & 1 << i` needs care), using masks when `n` is too large (> 25).

## Quick Revision

- Count bits DP: `bits[i] = bits[i>>1] + (i&1)`.
- Hamming = set bits of XOR.
- Subsets: `for mask in range(1 << n)`, bit `i` = include item `i`.
- Gray code: `i ^ (i >> 1)`.
- Letter-set masks: `a & b == 0` means no common letters.
