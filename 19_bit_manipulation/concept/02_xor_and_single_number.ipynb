# 02. XOR and the "Single Number" Family

XOR has magical properties that make many "find the odd one out" problems O(n) time and O(1) space.

**Learning order:** XOR RULES → SINGLE NUMBER → MISSING NUMBER → TWO SINGLE NUMBERS → FIND THE DIFFERENCE → ADD WITHOUT + → INTERVIEW THINKING

## 1. XOR rules

- `a ^ a = 0` (equal bits cancel)
- `a ^ 0 = a`
- **Commutative and associative:** order does not matter.

So XOR of a list where every number appears **twice except one** = that one number. All pairs cancel. Real life: a pile of socks where all have a pair except one.

```python
def single_number(nums):
    result = 0
    for x in nums:
        result ^= x
    return result

print(single_number([4, 1, 2, 1, 2]))   # 4
```

## 2. Missing Number

Numbers `0..n` with one missing. XOR all indexes `0..n` with all values. Pairs cancel; the missing one remains. (Or use the sum formula `n(n+1)/2 - sum(nums)`. Mention overflow is not a problem in Python but is in other languages.)

```python
def missing_number(nums):
    result = len(nums)
    for i, x in enumerate(nums):
        result ^= i ^ x
    return result

print(missing_number([3, 0, 1]), missing_number([9, 6, 4, 2, 3, 5, 7, 0, 1]))   # 2 8
```

## 3. Single Number III (two numbers appear once)

XOR of everything = `a ^ b` (the two singles). They differ in at least one bit. Find any set bit of `a ^ b` (`diff & -diff`). Split all numbers into two groups by that bit. Each group contains **one** single number (and pairs), so XOR within each group gives `a` and `b`.

```python
def single_number_iii(nums):
    xor_all = 0
    for x in nums:
        xor_all ^= x
    low_bit = xor_all & -xor_all
    a = b = 0
    for x in nums:
        if x & low_bit:
            a ^= x
        else:
            b ^= x
    return sorted([a, b])

print(single_number_iii([1, 2, 1, 3, 2, 5]))   # [3, 5]
```

## 4. Find the Difference

String `t` is `s` shuffled with one extra letter. XOR the character codes of both strings; pairs cancel.

```python
def find_the_difference(s, t):
    result = 0
    for ch in s + t:
        result ^= ord(ch)
    return chr(result)

print(find_the_difference("abcd", "abcde"))   # e
```

## 5. Add Two Integers without `+`

- `a ^ b` = sum **without carries**.
- `(a & b) << 1` = the **carries**.
- Repeat until no carry remains.

In Python we must **mask to 32 bits** (negative numbers never run out of leading 1s) and convert back to a signed value at the end.

```python
def get_sum(a, b):
    mask = 0xFFFFFFFF
    while b:
        a, b = (a ^ b) & mask, ((a & b) << 1) & mask
    return a if a <= 0x7FFFFFFF else ~(a ^ mask)

print(get_sum(1, 2), get_sum(-1, 1), get_sum(-2, 3))   # 3 0 1
```

## 6. Interview thinking

**Signals:** "every element appears twice except ...", "find the missing/extra", "without extra space", "swap without a temp", "sum without +".

**Say:** "XOR cancels equal values, so XOR of all elements leaves the unique one. O(n) time, O(1) space."

**Common mistakes:** applying XOR when elements appear **three** times (use bit counting), forgetting masks in Python for the add problem, mixing up `&` carry shifting.

## Quick Revision

- XOR: `a^a=0`, `a^0=a`, order does not matter.
- Single number: XOR everything. Missing number: XOR with indexes.
- Two singles: split by a differing bit (`x & -x`).
- Add without `+`: XOR + shifted AND carries, mask 32 bits.
