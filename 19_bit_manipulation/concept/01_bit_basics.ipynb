# 01. Bit Manipulation: Basics

Computers store numbers as **bits** (0 and 1). Bit tricks work directly on those bits: very fast (O(1)) and memory-light.

**Learning order:** REAL LIFE → BINARY → OPERATORS → PYTHON DETAILS → ESSENTIAL TRICKS → INTERVIEW THINKING

## 1. Real-life examples

- **Unix file permissions** `chmod 755`: each of read/write/execute is one bit.
- **Feature flags / settings:** 32 on/off switches stored in one integer.
- **IP address masks**, **graphics colors** (`0xFF0000`), **compression**, **hash functions**, **bitsets** (a set of 0..63 inside one number).

**Recognition clue:** "without extra space", "single number appearing once", "powers of two", "subsets of up to 20 items", "XOR".

## 2. Binary and the operators

`13` in binary is `1101` (8 + 4 + 0 + 1).

| Operator | Meaning | Example |
|---|---|---|
| `a & b` | AND: 1 if both bits are 1 | `1100 & 1010 = 1000` |
| `a \| b` | OR: 1 if either is 1 | `1100 \| 1010 = 1110` |
| `a ^ b` | XOR: 1 if bits differ | `1100 ^ 1010 = 0110` |
| `~a` | NOT: flip all bits (`-a - 1` in Python) | `~5 = -6` |
| `a << k` | shift left = multiply by `2^k` | `5 << 1 = 10` |
| `a >> k` | shift right = floor divide by `2^k` | `5 >> 1 = 2` |

```python
print(bin(13), 13 & 11, 13 | 11, 13 ^ 11)   # 0b1101 9 15 6
print(~5, 5 << 1, 5 >> 1, -5 >> 1)          # -6 10 2 -3
print(int("1101", 2), format(13, "08b"))    # 13 00001101
```

## 3. Python-specific details

- Python integers have **unlimited size** and **negative numbers act like an infinite string of leading 1s** (two's complement). `~x == -x - 1`.
- To simulate a **32-bit** integer, mask with `0xFFFFFFFF` (`-5 & 0xFFFFFFFF = 4294967291`).
- `x >> 1` on negatives **rounds down** (`-5 >> 1 = -3`), unlike C's truncation in some cases.
- `bin(-5)` gives `'-0b101'` (sign and magnitude), not the 32-bit pattern.

```python
print(-5 & 0xFFFFFFFF)    # 4294967291
print(bin(-5), bin(5 & 0xFF))   # -0b101 0b101
```

## 4. Essential tricks (memorize these)

| Goal | Expression |
|---|---|
| is `n` odd? | `n & 1` |
| get bit `i` | `(n >> i) & 1` |
| set bit `i` | `n \| (1 << i)` |
| clear bit `i` | `n & ~(1 << i)` |
| toggle bit `i` | `n ^ (1 << i)` |
| is power of two | `n > 0 and n & (n - 1) == 0` |
| clear lowest set bit | `n & (n - 1)` |
| isolate lowest set bit | `n & -n` |
| count set bits (Kernighan) | `while n: n &= n - 1; count += 1` |
| multiply / divide by 2 | `n << 1` / `n >> 1` |

**Why `n & (n-1)` clears the lowest set bit:** subtracting 1 flips the lowest set bit to 0 and all the zeros below it to 1. ANDing cancels exactly that part.

```python
n = 12                                  # 1100
print(n & 1, (n >> 2) & 1, n | (1 << 0), n & ~(1 << 2), n ^ (1 << 1))   # 0 1 13 8 14
print(n & (n - 1), n & -n)              # 8 4

def is_power_of_two(n):
    return n > 0 and n & (n - 1) == 0

def count_set_bits(n):
    count = 0
    while n:
        n &= n - 1                      # each loop removes one set bit
        count += 1
    return count

print(is_power_of_two(16), is_power_of_two(18), count_set_bits(13))   # True False 3
```

## 5. Interview thinking

**Say:** "I can solve this with bit operations in O(1) extra space, using the fact that ..."

**Common mistakes:** operator precedence (`n & 1 == 0` means `n & (1 == 0)`; **use parentheses** `(n & 1) == 0`), forgetting masks for 32-bit behaviour, assuming `>>` on negatives truncates, off-by-one in bit positions (bit 0 is the **least** significant).

## Quick Revision

- AND `&`, OR `|`, XOR `^`, NOT `~`, shifts `<<` `>>`.
- `n & (n-1)` clears the lowest set bit; `n & -n` isolates it.
- Power of two: `n > 0 and n & (n-1) == 0`.
- Parenthesize bit expressions in comparisons.
- Mask with `0xFFFFFFFF` for 32-bit behaviour in Python.
