# 04. Bit Manipulation Patterns (Revision Sheet)

**Learning order:** DECISION TABLE → CHEAT SHEET → MISTAKES → INTERVIEW SCRIPT → PRACTICE MAP

## 1. Decision table

| Question looks like | Trick |
|---|---|
| every number twice except one | XOR all |
| two numbers appear once | XOR all, split by a differing bit |
| others appear 3 times | `ones` / `twos` or count bits mod 3 |
| missing / extra element | XOR with indexes or letters |
| power of two / four | `n & (n-1) == 0` (+ mask for four) |
| number of set bits | Kernighan loop / `bin(n).count("1")` |
| bits differ between x and y | `x ^ y` |
| all subsets, `n ≤ 20` | masks `0..2ⁿ-1` |
| sets of letters | 26-bit masks, `&` to test intersection |
| sum without `+` | XOR + carry loop (mask in Python) |
| check / set / clear a flag | `&`, `\|`, `~`, `^` with `1 << i` |
| maximum XOR | binary trie (Trie topic) |

## 2. Cheat sheet

```
odd:            n & 1
bit i:          (n >> i) & 1
set / clear:    n | (1 << i)     n & ~(1 << i)
toggle:         n ^ (1 << i)
lowest set bit: n & -n
clear lowest:   n & (n - 1)
all ones (k):   (1 << k) - 1
32-bit view:    n & 0xFFFFFFFF
```

## 3. Common mistakes

- Precedence: always write `(n & 1) == 0`.
- Forgetting `0xFFFFFFFF` for fixed-width behaviour.
- Using XOR when elements repeat 3 times.
- Starting bit positions from 1 instead of 0.
- Too many masks (`n > 25`) → memory/time blow up.
- Using bit hacks that make the code unreadable when a simple approach is accepted. Explain the trick in a comment.

## 4. Interview script

1. "A hash set would work in O(n) space. With XOR I can use O(1) space."
2. Explain the property: "equal values cancel, so the unique value remains."
3. "Complexity O(n) time, O(1) space."
4. Test: single element, negatives, zero, large values.

## Quick Revision

- XOR for odd-one-out, `n & (n-1)` for power of two and bit counting.
- Masks for subsets (`n ≤ 20`) and letter sets.
- Parenthesize, mask 32 bits, bit 0 is the least significant.
