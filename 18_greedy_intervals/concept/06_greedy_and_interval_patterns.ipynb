# 06. Greedy and Interval Patterns (Revision Sheet)

**Learning order:** DECISION TABLE → HOW TO PROVE → SORT KEY CHEAT SHEET → MISTAKES → INTERVIEW SCRIPT → PRACTICE MAP

## 1. Decision table

| Question looks like | Pattern |
|---|---|
| reach the end, minimum jumps | running farthest reach |
| circular route / tank | total check + reset start |
| relative rankings, two-sided constraints | two passes |
| cut into parts, last occurrence | extend end by `last[c]` |
| maximum subarray | Kadane |
| merge / insert intervals | sort by **start** |
| max non-overlapping, min removals, min arrows | sort by **end** |
| max overlap, rooms, platforms | sweep line / heap |
| pair heaviest with lightest | sort + two pointers |
| split between two groups with a cost difference | sort by difference |
| "always pick the largest/smallest available" | heap (see Heap topic) |

## 2. How to justify a greedy

1. State the greedy choice.
2. **Exchange argument:** assume an optimal solution that differs; swap to the greedy choice; show it is still optimal.
3. After the choice, the remaining problem has the same form (optimal substructure).
4. Mention a counterexample you checked for the wrong alternatives.

## 3. Sort key cheat sheet

| Sort by | Used for |
|---|---|
| start | merge intervals, calendar conflicts |
| end | activity selection, min removals, arrows |
| value / weight | fractional knapsack |
| frequency | reduce array size, task counts |
| `a - b` | two cities, assign with differences |
| height desc, `k` asc | queue reconstruction |
| custom comparator | largest number (Sorting topic) |

## 4. Common mistakes

- Assuming greedy works without a proof or counterexample.
- Sorting by start when end is required (and vice versa).
- `<` vs `<=` for touching intervals.
- Not copying input lists when merging.
- Two passes needed but only one used (candy).
- Not initializing Kadane with `nums[0]`.
- Using greedy for 0/1 knapsack or coin change with arbitrary coins.

## 5. Interview script

1. "Brute force tries all choices: exponential."
2. "A greedy choice works here because (exchange argument)."
3. "I sort by ..., then one pass. O(n log n) time, O(1) extra space."
4. "If the denominations/constraints were arbitrary, greedy could fail and I would use DP."
5. Test: empty input, one item, all equal, touching intervals, already sorted.

## Quick Revision

- Greedy = right sort key + one pass + a justification.
- Start for merging, end for selecting.
- Sweep line for overlaps.
- Always try to break your greedy with a small counterexample.
