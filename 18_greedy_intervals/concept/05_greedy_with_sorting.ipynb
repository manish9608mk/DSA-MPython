# 05. Greedy with Sorting

Many greedy solutions are one line: **find the right sort key**. Then walk through once.

**Learning order:** TWO CITIES → BOATS → TRUCK UNITS → REDUCE ARRAY → QUEUE RECONSTRUCTION → FRACTIONAL KNAPSACK → INTERVIEW THINKING

## 1. Two City Scheduling

`2n` people, each has a cost to fly to city A and to city B. Send exactly `n` to each city with minimum total cost. Look at the **difference** `costA - costB`: people with the smallest difference are the cheapest to send to A relative to B. Sort by `a - b`, send the first half to A, the rest to B.

```python
def two_city_sched_cost(costs):
    costs.sort(key=lambda c: c[0] - c[1])
    n = len(costs) // 2
    return sum(c[0] for c in costs[:n]) + sum(c[1] for c in costs[n:])

print(two_city_sched_cost([[10, 20], [30, 200], [400, 50], [30, 20]]))   # 110
```

## 2. Boats to Save People

Each boat carries at most 2 people with a weight `limit`. Sort. Pair the **heaviest** with the **lightest** if they fit; otherwise the heaviest goes alone. Two pointers.

```python
def num_rescue_boats(people, limit):
    people.sort()
    lo, hi, boats = 0, len(people) - 1, 0
    while lo <= hi:
        if people[lo] + people[hi] <= limit:
            lo += 1
        hi -= 1
        boats += 1
    return boats

print(num_rescue_boats([3, 2, 2, 1], 3), num_rescue_boats([3, 5, 3, 4], 5))   # 3 4
```

## 3. Maximum Units on a Truck

Boxes of types `[count, units per box]`. Take boxes with the **most units per box** first.

```python
def maximum_units(box_types, truck_size):
    total = 0
    for count, units in sorted(box_types, key=lambda b: -b[1]):
        take = min(count, truck_size)
        total += take * units
        truck_size -= take
        if truck_size == 0:
            break
    return total

print(maximum_units([[1, 3], [2, 2], [3, 1]], 4), maximum_units([[5, 10], [2, 5], [4, 7], [3, 9]], 10))   # 8 91
```

## 4. Reduce Array Size to The Half

Remove all occurrences of chosen numbers so that at least half of the array is removed. Remove the **most frequent** numbers first (each removal kills the most elements).

```python
from collections import Counter

def min_set_size(arr):
    removed = chosen = 0
    for freq in sorted(Counter(arr).values(), reverse=True):
        removed += freq
        chosen += 1
        if removed >= len(arr) / 2:
            return chosen

print(min_set_size([3, 3, 3, 3, 5, 5, 5, 2, 2, 7]), min_set_size([7, 7, 7, 7, 7, 7]))   # 2 1
```

## 5. Queue Reconstruction by Height

`[h, k]`: a person of height `h` with exactly `k` people **in front who are at least as tall**. Rebuild the queue.

Sort by height **descending**, and for equal height by `k` **ascending**. Then **insert each person at index `k`**. Taller people are already placed and do not care about shorter ones, so inserting shorter people never breaks the taller people's counts.

```python
def reconstruct_queue(people):
    people.sort(key=lambda p: (-p[0], p[1]))
    queue = []
    for h, k in people:
        queue.insert(k, [h, k])
    return queue

print(reconstruct_queue([[7, 0], [4, 4], [7, 1], [5, 0], [6, 1], [5, 2]]))
# [[5, 0], [7, 0], [5, 2], [6, 1], [4, 4], [7, 1]]
```

## 6. Fractional Knapsack (the greedy one)

You can take **fractions** of items. Sort by **value per weight** descending. Take whole items while they fit, then a fraction of the next one. (0/1 knapsack, where you cannot split items, needs **DP**.)

```python
def fractional_knapsack(items, capacity):          # items: (value, weight)
    total = 0.0
    for value, weight in sorted(items, key=lambda x: x[0] / x[1], reverse=True):
        take = min(weight, capacity)
        total += value * take / weight
        capacity -= take
        if capacity == 0:
            break
    return total

print(fractional_knapsack([(60, 10), (100, 20), (120, 30)], 50))   # 240.0
```

## 7. Interview thinking

**Say:** "The right order is ..., because swapping any pair that is out of this order cannot improve the answer. So I sort and scan. O(n log n)."

**Common mistakes:** sorting by the wrong key (value vs value/weight), applying fractional knapsack logic to 0/1 knapsack (wrong), not stating **why** the sort order is correct, forgetting ties in the sort key (queue reconstruction).

## Quick Revision

- Find the sort key: `a - b` (two cities), value/weight (fractional knapsack), frequency (reduce array), height desc + k asc (queue).
- Two pointers on sorted data (boats).
- 0/1 knapsack ≠ greedy.
