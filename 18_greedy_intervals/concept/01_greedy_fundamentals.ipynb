# 01. Greedy Algorithms: Fundamentals

A greedy algorithm makes the **best-looking choice at each step** and **never looks back**. It is fast (usually O(n log n) or O(n)) but works **only when local best choices add up to a global best**.

**Learning order:** REAL LIFE → WHEN IT WORKS → WHEN IT FAILS → PROOF IDEAS → EASY EXAMPLES → GREEDY VS DP → INTERVIEW THINKING

## 1. Real-life examples

- **Giving change** with ₹10, ₹5, ₹2, ₹1 coins: take the biggest coin that fits, repeat.
- **Packing for a trip:** put the most valuable-per-kg items first.
- **Catching trains/meetings:** always take the meeting that **ends earliest**, so you keep the most time for the next ones.

**Recognition clue:** "minimum/maximum number of ...", "choose in some order", and sorting by some key makes the problem look easy.

## 2. When does greedy work?

Two properties must hold:
1. **Greedy-choice property:** some optimal solution contains the greedy choice.
2. **Optimal substructure:** after the choice, the rest is a smaller problem of the same kind.

**Proof tools (what to say in interviews):**
- **Exchange argument:** take any optimal solution; if it does not use the greedy choice, swap something to use it without making the answer worse.
- **"Greedy stays ahead":** after each step, the greedy solution is at least as good as any other.

## 3. When greedy FAILS: coins `[1, 3, 4]`, amount 6

Greedy takes `4`, then `1`, `1` = **3 coins**. But `3 + 3` = **2 coins**. With normal currency denominations (1, 2, 5, 10, ...) greedy happens to work. With arbitrary coins it does not. Then you need **dynamic programming**.

```python
def greedy_coins(coins, amount):
    count = 0
    for c in sorted(coins, reverse=True):
        count += amount // c
        amount %= c
    return count if amount == 0 else -1

def dp_coins(coins, amount):                     # preview of DP
    dp = [0] + [float("inf")] * amount
    for a in range(1, amount + 1):
        for c in coins:
            if c <= a:
                dp[a] = min(dp[a], dp[a - c] + 1)
    return dp[amount] if dp[amount] != float("inf") else -1

print(greedy_coins([10, 5, 2, 1], 28))   # 5
print(greedy_coins([1, 3, 4], 6), dp_coins([1, 3, 4], 6))   # 3 2
```

## 4. Easy greedy examples

- **Assign Cookies:** sort children by greed and cookies by size. Give each child the **smallest cookie** that satisfies them.
- **Lemonade Change:** always give back change using the **largest bills first**.
- **Best Time to Buy and Sell Stock II:** buy and sell on every **upward step**: the sum of positive daily differences.

```python
def find_content_children(g, s):
    g.sort(); s.sort()
    child = 0
    for cookie in s:
        if child < len(g) and cookie >= g[child]:
            child += 1
    return child

def lemonade_change(bills):
    five = ten = 0
    for b in bills:
        if b == 5:
            five += 1
        elif b == 10:
            if not five: return False
            five -= 1; ten += 1
        else:
            if ten and five:
                ten -= 1; five -= 1
            elif five >= 3:
                five -= 3
            else:
                return False
    return True

def max_profit_ii(prices):
    return sum(max(0, prices[i] - prices[i - 1]) for i in range(1, len(prices)))

print(find_content_children([1, 2, 3], [1, 1]), find_content_children([1, 2], [1, 2, 3]))   # 1 2
print(lemonade_change([5, 5, 5, 10, 20]), lemonade_change([5, 5, 10, 10, 20]))   # True False
print(max_profit_ii([7, 1, 5, 3, 6, 4]))   # 7
```

## 5. Greedy vs DP vs backtracking

| | Greedy | DP | Backtracking |
|---|---|---|---|
| Decides | one choice, never revisits | considers all choices via stored subresults | tries all choices |
| Speed | fastest | polynomial | exponential |
| Correctness | needs proof | always correct for the model | always correct |

**Test:** try to find a **counterexample** with small inputs. If you cannot break it and you can give an exchange argument, use greedy. Otherwise use DP.

## 6. Interview thinking

**Say:** "I'll sort by ... and always pick ... because swapping any other choice for this one never makes the answer worse (exchange argument). Time O(n log n) for the sort."

**Common mistakes:** assuming greedy without checking a counterexample, sorting by the wrong key (start vs end time), not handling ties, forgetting to prove or explain correctness.

## Quick Revision

- Greedy = best local choice, no going back. Needs greedy-choice + optimal substructure.
- Prove with an exchange argument or "stays ahead".
- Coin change with arbitrary coins breaks greedy → DP.
- Find the right **sort key**.
- Try to break your greedy with a small counterexample.
