# 02. Classic Greedy Problems

Problems where one clever **running quantity** (farthest reach, tank, last position) gives the answer in one pass.

**Learning order:** JUMP GAME → JUMP GAME II → GAS STATION → CANDY → PARTITION LABELS → MAX SUBARRAY → INTERVIEW THINKING

## 1. Jump Game

`nums[i]` = maximum jump length from `i`. Can you reach the last index? Track the **farthest index reachable**. If you stand at an index beyond `reach`, you are stuck.

```python
def can_jump(nums):
    reach = 0
    for i, x in enumerate(nums):
        if i > reach:
            return False
        reach = max(reach, i + x)
    return True

print(can_jump([2, 3, 1, 1, 4]), can_jump([3, 2, 1, 0, 4]))   # True False
```

## 2. Jump Game II (minimum jumps)

Think in **BFS levels without a queue**: `cur_end` = the farthest index reachable with the current number of jumps, `farthest` = the farthest index reachable with one more jump. When we reach `cur_end`, we must jump: `jumps += 1`, `cur_end = farthest`.

```python
def jump(nums):
    jumps = cur_end = farthest = 0
    for i in range(len(nums) - 1):
        farthest = max(farthest, i + nums[i])
        if i == cur_end:
            jumps += 1
            cur_end = farthest
    return jumps

print(jump([2, 3, 1, 1, 4]), jump([2, 3, 0, 1, 4]))   # 2 2
```

## 3. Gas Station

Circular route. `gas[i]` fuel at station `i`, `cost[i]` to reach the next. Find the start index to complete the circle (or `-1`).

- If `sum(gas) < sum(cost)` it is impossible.
- Otherwise, drive and track the tank. If the tank goes **negative** at `i`, no station from the current start up to `i` can be the start. The next candidate is `i + 1`. The answer is unique, so the last candidate wins.

```python
def can_complete_circuit(gas, cost):
    if sum(gas) < sum(cost):
        return -1
    tank = start = 0
    for i in range(len(gas)):
        tank += gas[i] - cost[i]
        if tank < 0:
            start = i + 1
            tank = 0
    return start

print(can_complete_circuit([1, 2, 3, 4, 5], [3, 4, 5, 1, 2]), can_complete_circuit([2, 3, 4], [3, 4, 3]))   # 3 -1
```

## 4. Candy

Each child gets at least 1 candy, a child with a **higher rating than a neighbour** gets more candy than that neighbour. Minimum candies?

Two passes: left to right (compare with the left neighbour), right to left (compare with the right neighbour, take the **max** with the first pass).

```python
def candy(ratings):
    n = len(ratings)
    candies = [1] * n
    for i in range(1, n):
        if ratings[i] > ratings[i - 1]:
            candies[i] = candies[i - 1] + 1
    for i in range(n - 2, -1, -1):
        if ratings[i] > ratings[i + 1]:
            candies[i] = max(candies[i], candies[i + 1] + 1)
    return sum(candies)

print(candy([1, 0, 2]), candy([1, 2, 2]))   # 5 4
```

## 5. Partition Labels

Cut the string so **each letter appears in at most one part**. Minimise cuts → maximise parts. Compute the **last index** of each letter. Extend the current part's end to the farthest `last[c]` seen. When `i` reaches the end, cut.

```python
def partition_labels(s):
    last = {c: i for i, c in enumerate(s)}
    res, start, end = [], 0, 0
    for i, c in enumerate(s):
        end = max(end, last[c])
        if i == end:
            res.append(end - start + 1)
            start = i + 1
    return res

print(partition_labels("ababcbacadefegdehijhklij"))   # [9, 7, 8]
```

## 6. Maximum Subarray (Kadane's algorithm)

At every element decide: **extend** the previous subarray or **start fresh**. If the running sum becomes negative, it only hurts the future, so drop it.

```python
def max_sub_array(nums):
    best = cur = nums[0]
    for x in nums[1:]:
        cur = max(x, cur + x)
        best = max(best, cur)
    return best

print(max_sub_array([-2, 1, -3, 4, -1, 2, 1, -5, 4]))   # 6
```

## 7. Interview thinking

**Signals:** "reach", "minimum number of steps/jumps", "circular route", "relative ranking", "partition so that ...".

**Say:** "I keep one running value (farthest reach / tank / current sum) and update it in a single pass. O(n) time, O(1) space."

**Common mistakes:** jump game II loop to `n` instead of `n - 1`, forgetting the total sum check in gas station, one-pass candy (needs two passes), resetting `cur` wrongly in Kadane with all negatives (initialize with `nums[0]`, not 0).

## Quick Revision

- Jump game: farthest reach, stuck if `i > reach`.
- Jump II: `cur_end` / `farthest`, jump at `cur_end`.
- Gas station: total check + reset start when the tank < 0.
- Candy: left pass, right pass, take max.
- Partition labels: last index per letter.
- Kadane: `cur = max(x, cur + x)`.
