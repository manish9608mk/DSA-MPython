# 506. Relative Ranks

Link:
https://leetcode.com/problems/relative-ranks/

PATTERN:
Sorting + Original Index Mapping

PROBLEM:
We need to rank athletes by score.

1st → Gold Medal
2nd → Silver Medal
3rd → Bronze Medal
4th onward → "4", "5", ...

The answer must be in the ORIGINAL order.


HOW TO THINK:

Ranking depends on score order
        ↓
Need descending order
        ↓
SORTING

But after sorting, original positions are lost
        ↓
Store original index with every score

(score, original_index)
        ↓
Sort by score descending
        ↓
Sorted position = rank
        ↓
Put rank back at original index


APPROACH 1 — BRUTE FORCE:

For every athlete, count how many scores are greater.

rank = greater scores + 1

Time  : O(n²)
Space : O(n)


APPROACH 2 — SORT + HASH MAP:

Store:

score → original index

Sort scores in descending order.
Use the map to place each rank at its original position.

Time  : O(n log n)
Space : O(n)


APPROACH 3 — STANDARD / OPTIMAL:

Store:

(score, original_index)

Example:

[(10,0), (3,1), (8,2), (9,3), (4,4)]

Sort descending:

[(10,0), (9,3), (8,2), (4,4), (3,1)]

Now:

position 0 → rank 1
position 1 → rank 2
position 2 → rank 3
position 3 → rank 4
position 4 → rank 5

Use the stored original index to restore the answer.

Time  : O(n log n)
Space : O(n)


WHY THIS WORKS:

Sorting tells us the ranking order.
Original index tells us where the answer belongs.

So:

SORT
  +
PRESERVE INDEX
  +
RESTORE ORIGINAL ORDER


INTERVIEW THINKING:

Need ranking
    ↓
Ranking depends on score
    ↓
Sort descending
    ↓
Need original positions
    ↓
Store index
    ↓
(score, index)
    ↓
Sort
    ↓
Position = rank
    ↓
Put answer at original index


INTERVIEW EXPLANATION:

"I'll sort the athletes by score in descending order because
the ranking depends on their score. Since the output must
remain in the original order, I'll store each score with its
original index. After sorting, the position gives the rank,
and I use the original index to place the result correctly."


RECOGNITION TRICKS:

1. Ranking depends on ordering
2. Need sorting
3. Original positions matter
4. Need to restore the answer

→ Think:

SORT + ORIGINAL INDEX


COMPLEXITY:

Time  : O(n log n)
Space : O(n)


FAMILY:

Relative Ranks
Sort + Restore Index
Ranking problems
Top K problems
Sorting with original positions


```python
# 506. Relative Ranks

class Solution:

    def findRelativeRanks(self, score: list[int]) -> list[str]:

        athletes = [
            (value, index)
            for index, value in enumerate(score)
        ]

        athletes.sort(reverse=True)

        answer = [""] * len(score)

        for rank, (value, index) in enumerate(athletes, start=1):

            if rank == 1:
                answer[index] = "Gold Medal"

            elif rank == 2:
                answer[index] = "Silver Medal"

            elif rank == 3:
                answer[index] = "Bronze Medal"

            else:
                answer[index] = str(rank)

        return answer


# Test case
score = [10, 3, 8, 9, 4]

solution = Solution()

result = solution.findRelativeRanks(score)

print("Result =", result)

# Output:
# ['Gold Medal', '5', 'Bronze Medal', 'Silver Medal', '4']
```
