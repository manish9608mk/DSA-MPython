## 2. Complexity of the common array solutions

| Pattern | Time | Space |
|---|---|---|
| scan once | O(n) | O(1) |
| hash map | O(n) | O(n) |
| sort first | O(n log n) | O(1) or O(n) |
| two pointers (sorted) | O(n) | O(1) |
| Kadane | O(n) | O(1) |
| prefix sum | O(n) build, O(1) query | O(n) |
| binary search | O(log n) | O(1) |
| matrix traversal | O(m·n) | O(1) to O(m·n) |

## 3. Common mistakes

- Off-by-one in loops (`range(n - 1)` vs `range(n)`) and slices (`a[l:r]` excludes `r`).
- Modifying a list while iterating over it.
- Forgetting empty array, size 1, all equal, all negative, duplicates.
- Aliasing (`[[0]*n]*m`, `b = a`).
- Using `in list` inside loops (O(n²)).
- Assuming the array is sorted.
- Using nested loops when the constraints need O(n log n).
- Returning indexes after sorting (they changed).

## 4. Interview script

1. Clarify: sorted? duplicates? negatives? size? can I modify the input?
2. Brute force and its complexity.
3. Name the pattern (table above) and why it applies.
4. Code with clear variable names, handle edge cases.
5. Dry-run on a small example. State time and space.

## Quick Revision

- Pattern first, code second: sorted → pointers, contiguous → window / prefix, range `1..n` → index as hash.
- Edge cases: empty, 1, all equal, negatives, duplicates.
- State time and space. Ask if the input may be modified.
