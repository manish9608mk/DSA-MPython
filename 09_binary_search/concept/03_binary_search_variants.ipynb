# 03. Binary Search Variants: Rotated, Peak, Matrix, Square Root

Binary search works whenever you can answer: "**which half can I throw away?**" Even if the array is not fully sorted.

**Learning order:** ROTATED ARRAY → MINIMUM IN ROTATED → PEAK → 2D MATRIX → SQUARE ROOT → FIRST BAD VERSION → INTERVIEW THINKING

## 1. Search in Rotated Sorted Array

`[4, 5, 6, 7, 0, 1, 2]` is a sorted array rotated at some point.

**Key fact:** after taking `mid`, **at least one half is sorted**. Find which, and check if the target lies inside that sorted half. If yes, go there. If no, go to the other half.

```python
def search_rotated(nums, target):
    lo, hi = 0, len(nums) - 1
    while lo <= hi:
        mid = lo + (hi - lo) // 2
        if nums[mid] == target:
            return mid
        if nums[lo] <= nums[mid]:                      # left half sorted
            if nums[lo] <= target < nums[mid]:
                hi = mid - 1
            else:
                lo = mid + 1
        else:                                          # right half sorted
            if nums[mid] < target <= nums[hi]:
                lo = mid + 1
            else:
                hi = mid - 1
    return -1

print(search_rotated([4, 5, 6, 7, 0, 1, 2], 0))   # 4
print(search_rotated([4, 5, 6, 7, 0, 1, 2], 3))   # -1
```

## 2. Minimum in Rotated Sorted Array

Compare `nums[mid]` with `nums[hi]`:
- `nums[mid] > nums[hi]` → the drop (minimum) is on the **right** of mid → `lo = mid + 1`.
- otherwise mid could be the minimum → `hi = mid`.

```python
def find_min_rotated(nums):
    lo, hi = 0, len(nums) - 1
    while lo < hi:
        mid = (lo + hi) // 2
        if nums[mid] > nums[hi]:
            lo = mid + 1
        else:
            hi = mid
    return nums[lo]

print(find_min_rotated([3, 4, 5, 1, 2]))          # 1
print(find_min_rotated([4, 5, 6, 7, 0, 1, 2]))    # 0
print(find_min_rotated([11, 13, 15, 17]))         # 11
```

## 3. Find Peak Element

A peak is bigger than its neighbours. The array is **not sorted**, but compare `mid` with `mid + 1`:
- `nums[mid] < nums[mid + 1]` → we are going uphill, a peak exists on the right → `lo = mid + 1`.
- else a peak exists at `mid` or on the left → `hi = mid`.

Real life: climbing a hill in fog. If the ground ahead goes up, keep going that way; you will reach a top.

```python
def find_peak(nums):
    lo, hi = 0, len(nums) - 1
    while lo < hi:
        mid = (lo + hi) // 2
        if nums[mid] < nums[mid + 1]:
            lo = mid + 1
        else:
            hi = mid
    return lo

print(find_peak([1, 2, 3, 1]))               # 2
print(find_peak([1, 2, 1, 3, 5, 6, 4]))      # 5  (index 1 is also a valid peak)
```

## 4. Search a 2D Matrix

Rows sorted, and each row starts after the previous row ends. Treat it as a **flattened sorted array** of `m × n` elements. Index `mid` maps to `row = mid // n`, `col = mid % n`.

```python
def search_matrix(matrix, target):
    m, n = len(matrix), len(matrix[0])
    lo, hi = 0, m * n - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        val = matrix[mid // n][mid % n]
        if val == target:
            return True
        if val < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return False

grid = [[1, 3, 5, 7], [10, 11, 16, 20], [23, 30, 34, 60]]
print(search_matrix(grid, 3), search_matrix(grid, 13))   # True False
```

## 5. Square Root (integer)

We want the largest `r` with `r * r <= x`. The rule `r*r <= x` is `True True True False False`. Find the **last True**: when `mid*mid <= x`, store it as a candidate and search right.

```python
def my_sqrt(x):
    lo, hi = 0, x
    ans = 0
    while lo <= hi:
        mid = (lo + hi) // 2
        if mid * mid <= x:
            ans = mid
            lo = mid + 1
        else:
            hi = mid - 1
    return ans

print(my_sqrt(8), my_sqrt(4), my_sqrt(0))   # 2 2 0
```

## 6. First Bad Version

Versions are `good good good BAD BAD BAD`. Find the first bad one. This is exactly the **"first True"** template with `pred = is_bad`. (Real life: `git bisect`.)

```python
def first_bad_version(n, is_bad):
    lo, hi = 1, n
    while lo < hi:
        mid = (lo + hi) // 2
        if is_bad(mid):
            hi = mid
        else:
            lo = mid + 1
    return lo

print(first_bad_version(5, lambda v: v >= 4))   # 4
```

## 7. Interview thinking

**Ask yourself:** "After looking at `mid`, can I **prove** that one side cannot contain the answer?" If yes, binary search works, even without a fully sorted array.

**Common mistakes:** rotated search with `nums[lo] < nums[mid]` instead of `<=` (breaks with 2 elements), using `lo <= hi` with `hi = mid`, comparing `mid` with the wrong neighbour in peak search, forgetting duplicates change the rotated-array logic (worst case becomes O(n)).

## Quick Revision

- Rotated: one half is always sorted, check if target is in it.
- Min in rotated: `nums[mid] > nums[hi]` → `lo = mid + 1` else `hi = mid`.
- Peak: compare `mid` and `mid + 1`, go uphill.
- 2D matrix: flatten, `row = mid // n`, `col = mid % n`.
- sqrt / first bad: "last True" or "first True" on a monotonic rule.
