# 01. Binary Search: Fundamentals

Binary search finds something by **cutting the search space in half every step**. It needs the data to be **ordered in a way that lets you throw away half** after one comparison.

**Learning order:** REAL LIFE → IDEA → CODE → TRACE → INVARIANT → WHY O(log n) → BISECT MODULE → PITFALLS → INTERVIEW THINKING

## 1. Real-life examples

- **Dictionary:** to find "mango", you open the middle. You see "kite", so "mango" is after it. You never look at the first half again.
- **Guess the number (1 to 100):** you say 50. "Higher." You say 75. "Lower." At most **7 guesses** are enough.
- **git bisect:** 1000 commits, one broke the app. Test the middle commit. Broken → bug is in the first half. Fine → second half. About 10 tests find it.
- **Phone contacts list** sorted by name.

## 2. The idea

1. Look at the **middle** element.
2. If it is the target → done.
3. If the target is bigger → the answer can only be on the **right**. Throw away the left half.
4. If smaller → throw away the right half.
5. Repeat until nothing is left.

**Requirement:** a sorted array (or any "yes/no" rule that flips only once, see notebook 04).

```python
def binary_search(nums, target):
    lo, hi = 0, len(nums) - 1          # search space: nums[lo..hi] (inclusive)
    while lo <= hi:
        mid = lo + (hi - lo) // 2      # safe way to take the middle
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            lo = mid + 1               # discard mid and everything left of it
        else:
            hi = mid - 1               # discard mid and everything right of it
    return -1

print(binary_search([-1, 0, 3, 5, 9, 12], 9))   # 4
print(binary_search([-1, 0, 3, 5, 9, 12], 2))   # -1
```

```python
# Trace: watch the search space shrink
nums, target = [1, 3, 5, 7, 9, 11, 13], 11
lo, hi = 0, len(nums) - 1
while lo <= hi:
    mid = lo + (hi - lo) // 2
    print(f"lo={lo} hi={hi} mid={mid} nums[mid]={nums[mid]}")
    if nums[mid] == target:
        print("found at", mid)
        break
    elif nums[mid] < target:
        lo = mid + 1
    else:
        hi = mid - 1
```

## 3. The invariant (the thing that is always true)

**"If the target exists, it is inside `nums[lo..hi]`."**

Every step keeps this true: when we move `lo = mid + 1` we only discard elements we have proved are too small. The loop ends when `lo > hi`, i.e. the search space is empty, so the target is not there.

Why `lo <= hi` (not `<`): with `lo == hi` there is still **one element** left to check.

## 4. Why O(log n)?

Each step halves the space: n → n/2 → n/4 → … → 1. Number of steps = log₂ n.

- 1,000 elements → about 10 steps
- 1,000,000 → about 20 steps
- 1,000,000,000 → about **30 steps**

Linear search on 10⁹ elements takes 10⁹ steps. Space: **O(1)** for the loop version.

```python
import math
print(math.ceil(math.log2(10**9)))   # 30
```

## 5. Python's `bisect` module

`bisect_left(a, x)` = first index where `x` could be inserted (first element ≥ x).
`bisect_right(a, x)` = index after the last element equal to `x` (first element > x).

In interviews, write the loop yourself unless the interviewer allows the library.

```python
import bisect
a = [1, 2, 4, 4, 4, 7]
print(bisect.bisect_left(a, 4))    # 2
print(bisect.bisect_right(a, 4))   # 5
print(bisect.bisect_left(a, 5))    # 5  (where 5 would go)
```

## 6. Pitfalls

- `mid = (lo + hi) // 2` is fine in Python (no overflow), but in Java/C++ use `lo + (hi - lo) // 2`. Make it a habit.
- Writing `lo = mid` or `hi = mid` in the `lo <= hi` version → **infinite loop**.
- Using binary search on an **unsorted** array.
- Off-by-one: `hi = len(nums)` vs `len(nums) - 1`. Pick one style and keep the loop condition consistent with it.

## 7. Interview thinking

**Signals:** "sorted array", "find in O(log n)", "minimum such that ...", "first / last position".

**Say:** "The array is sorted, so after comparing with the middle I can discard half the elements. That gives O(log n) time and O(1) space."

Always **state the invariant** and test: empty array, one element, target smaller than all, bigger than all, duplicates.

## Quick Revision

- Compare with middle, discard half. Needs sorted / monotonic data.
- `lo, hi = 0, n-1`, `while lo <= hi`, `mid = lo + (hi-lo)//2`, move `mid ± 1`.
- Invariant: target (if present) is in `[lo, hi]`.
- O(log n) time, O(1) space.
- `bisect_left` = first ≥ x, `bisect_right` = first > x.
