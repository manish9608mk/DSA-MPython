## 1. The templates (copy these)

**A. Exact search** (`lo <= hi`): return `mid` on match, else `lo = mid + 1` or `hi = mid - 1`.

**B. First True** (`lo < hi`, `hi` can be `n`): `mid = (lo+hi)//2`, True → `hi = mid`, False → `lo = mid + 1`.

**C. Last True** (`lo < hi`): `mid = (lo+hi+1)//2`, True → `lo = mid`, False → `hi = mid - 1`.

**D. Search on answer**: define `feasible(x)`, then use B (minimize) or C (maximize).

```python
def first_true(lo, hi, pred):
    while lo < hi:
        mid = (lo + hi) // 2
        if pred(mid):
            hi = mid
        else:
            lo = mid + 1
    return lo

def last_true(lo, hi, pred):
    while lo < hi:
        mid = (lo + hi + 1) // 2
        if pred(mid):
            lo = mid
        else:
            hi = mid - 1
    return lo

print(first_true(0, 100, lambda x: x * x >= 50))   # 8
print(last_true(0, 100, lambda x: x * x <= 50))    # 7
```

## 2. Decision table

| Question | Template |
|---|---|
| does x exist in a sorted array | A |
| first / last occurrence, insert position, count of x | B (lower / upper bound) |
| smallest element ≥ x, largest ≤ x | B, and B minus 1 |
| first bad version, first index where rule holds | B |
| integer sqrt, last value where rule holds | C |
| rotated array, peak, matrix | A or B with a smart comparison |
| minimum speed / capacity / time such that ... | D with B |
| maximum distance / size such that ... | D with C |

## 3. How to spot a monotonic rule

Write the candidate answers in a row and mark each with feasible / not feasible. If you can see **one single switch** (`F F F T T T`), binary search works.

Quick test: "if `x` works, does `x + 1` (or `x - 1`) also work?" If yes, the rule is monotonic.

**Not binary-searchable:** rules that flip many times, or unsorted data with no way to discard half.

## 4. Common mistakes

- `lo = mid` or `hi = mid` inside `while lo <= hi` → infinite loop.
- `hi = mid - 1` in the first-True template → skips the answer.
- Lower middle in the last-True template → infinite loop. Use `(lo + hi + 1) // 2`.
- Wrong answer range (`lo`, `hi`), especially the smallest/largest possible answer.
- Reading `nums[lo]` when `lo == len(nums)`.
- `feasible` not truly monotonic.
- Rotated array with duplicates: worst case becomes O(n).

## 5. Interview script

1. "Brute force tries every possible answer, O(range × n)."
2. "If a value works, a larger one also works, so the feasibility is monotonic."
3. "I binary search the answer between `lo` and `hi`. Each check is O(n)."
4. "Total O(n log range), O(1) extra space."
5. Test: smallest possible, largest possible, single element, answer at the boundaries.

## Quick Revision

- Exact: `lo <= hi`, move `mid ± 1`.
- First True: `lo < hi`, `hi = mid` / `lo = mid + 1`.
- Last True: upper middle, `lo = mid` / `hi = mid - 1`.
- Search on answer: write `feasible(x)`, check monotonic, set range, pick template.
- Always test the boundaries and the empty / single case.
