# 01. Prefix Sum: Fundamentals

A **prefix sum** array stores the running total. After one O(n) pass, the sum of **any range** can be answered in **O(1)**.

**Learning order:** WHAT → REAL LIFE → THE PROBLEM → BUILD → RANGE FORMULA → WHY n+1 → CLASS VERSION → INTERVIEW THINKING

## 1. Real-life examples

- **Cricket scorecard:** the board shows total score after every over. Runs in overs 6 to 10 = score after over 10 − score after over 5. You never add each over again.
- **Bank passbook:** it shows the running balance. Money spent between two dates = balance difference.
- **Odometer:** distance of a trip = reading at the end − reading at the start.

## 2. The problem

Array `nums` and many queries `(l, r)`: give `sum(nums[l..r])`.

Brute force adds the elements for every query → **O(n) per query**. With `q` queries that is O(n × q).
Prefix sum: **O(n) once**, then **O(1) per query**.

## 3. Build and formula

`prefix[i]` = sum of the **first `i` elements** (so `prefix[0] = 0`, the sum of nothing).

```
nums   :  3  1  4  1  5  9  2  6
prefix : 0  3  4  8  9 14 23 25 31
```

**Range sum:** `sum(nums[l..r]) = prefix[r + 1] - prefix[l]`

Picture: `prefix[r+1]` covers everything up to `r`. `prefix[l]` covers everything before `l`. Subtract and only `l..r` is left.

````python
def build_prefix(nums):
    prefix = [0] * (len(nums) + 1)
    for i, x in enumerate(nums):
        prefix[i + 1] = prefix[i] + x
    return prefix

def range_sum(prefix, l, r):
    return prefix[r + 1] - prefix[l]

nums = [3, 1, 4, 1, 5, 9, 2, 6]
prefix = build_prefix(nums)
print(prefix)                      # [0, 3, 4, 8, 9, 14, 23, 25, 31]
print(range_sum(prefix, 1, 4))     # 11  (1 + 4 + 1 + 5)
print(range_sum(prefix, 0, 7))     # 31  (whole array)
````

## 4. Why the array has size n + 1

The extra `prefix[0] = 0` removes the special case when `l = 0`. Without it you would need `if l == 0: return prefix[r]`. With it, the same formula works for every range, including the whole array.

Python shortcut: `list(itertools.accumulate(nums, initial=0))` builds the same array.

````python
from itertools import accumulate

print(list(accumulate([3, 1, 4], initial=0)))   # [0, 3, 4, 8]

class NumArray:                      # Range Sum Query - Immutable
    def __init__(self, nums):
        self.prefix = [0] + list(accumulate(nums))

    def sum_range(self, left, right):
        return self.prefix[right + 1] - self.prefix[left]

obj = NumArray([-2, 0, 3, -5, 2, -1])
print(obj.sum_range(0, 2), obj.sum_range(2, 5), obj.sum_range(0, 5))   # 1 -1 -3
````

## 5. Complexity

- Build: **O(n) time, O(n) space**
- Each query: **O(1)**
- Works with **negative numbers** too (unlike sliding window).

**Limit:** if the array **changes** between queries, the prefix array becomes wrong. For updates + queries you need a Fenwick tree or segment tree (Advanced DSA).

## 6. Interview thinking

**Signals:** "many range-sum queries", "sum of subarray", "sum between indices i and j", "sum of subarray equals k" (with hashing, next notebook).

**Say:** "Brute force recomputes each range in O(n). I precompute prefix sums once, then every range is `prefix[r+1] - prefix[l]` in O(1)."

**Common mistakes:** using `prefix[r] - prefix[l]` (off-by-one), forgetting the leading 0, using it when the array is updated between queries.

## Quick Revision

- `prefix[0] = 0`, `prefix[i+1] = prefix[i] + nums[i]`.
- Range `l..r` = `prefix[r+1] - prefix[l]`.
- Build O(n), query O(1), space O(n).
- Works with negatives. Breaks if the array is updated (need Fenwick tree).
