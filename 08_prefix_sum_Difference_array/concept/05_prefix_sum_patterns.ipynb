# 05. Prefix Sum and Difference Array Patterns (Revision Sheet)

**Learning order:** DECISION TABLE → TEMPLATES → VS OTHER TOOLS → MISTAKES → INTERVIEW SCRIPT → PRACTICE MAP

## 1. Which pattern?

| Question looks like | Pattern |
|---|---|
| many `sum(l, r)` queries on a static array | prefix sum, `prefix[r+1] - prefix[l]` |
| count subarrays with sum `k` | prefix sum + hash map of **counts** `{0: 1}` |
| longest subarray with sum `k` | prefix sum + hash map of **first index** `{0: -1}` |
| divisible by `k` | prefix **remainders** + hash map |
| equal 0s and 1s | map 0 → −1, then longest sum 0 |
| left side equals right side | running `left`, `total - left - x` |
| everything except self | left pass × right pass |
| rectangle sum queries | 2D prefix sum |
| many range updates, then final array | difference array |
| overlapping bookings, capacity | difference array (`+` at start, `-` at end) |

## 2. Templates

**Prefix sum**
```python
prefix = [0] * (n + 1)
for i in range(n):
    prefix[i + 1] = prefix[i] + nums[i]
# sum(l..r) = prefix[r + 1] - prefix[l]
```

**Prefix + hash (count)**
```python
seen = {0: 1}; prefix = 0; count = 0
for x in nums:
    prefix += x
    count += seen.get(prefix - k, 0)
    seen[prefix] = seen.get(prefix, 0) + 1
```

**Difference array**
```python
diff = [0] * (n + 1)
diff[l] += x; diff[r + 1] -= x
# final = running sum of diff
```

## 3. Prefix sum vs other tools

| Tool | When |
|---|---|
| **Sliding window** | contiguous subarray, all numbers positive (or the rule changes predictably) |
| **Prefix sum + hash** | negatives allowed, "sum equals k", count of subarrays |
| **Difference array** | many range **updates**, read at the end |
| **Fenwick / segment tree** | updates and queries are mixed |
| **Binary search on prefix** | positive numbers, find first position where the prefix reaches a value |

## 4. Common mistakes

- `prefix[r] - prefix[l]` instead of `prefix[r+1] - prefix[l]`.
- Missing `{0: 1}` or `{0: -1}` at the start of the map.
- Storing the current prefix **before** looking up (breaks `k = 0`).
- Overwriting the first index in "longest" problems.
- Difference array of size `n` (crash on `r + 1`).
- Using sliding window when there are negative numbers.
- Forgetting the final prefix sum over the difference array.

## 5. Interview script

1. "Brute force recomputes the sum of every range, O(n) each."
2. "I precompute prefix sums in O(n). A range sum is a subtraction."
3. For counting: "A subarray sum is `prefix[j+1] - prefix[i]`, so I look for `current - k` in a hash map of earlier prefixes."
4. For updates: "Instead of updating every element, I mark where the change starts and stops, then one prefix sum builds the final array."
5. Test: empty array, single element, all negatives, `k = 0`, all zeros.

## Quick Revision

- Prefix: `prefix[r+1] - prefix[l]`, leading 0.
- Count subarrays: map of counts, `{0: 1}`, look up first.
- Longest: map of first index, `{0: -1}`.
- Divisible: equal remainders. Equal 0/1: treat 0 as −1.
- Range updates: `diff[l] += x`, `diff[r+1] -= x`, prefix at the end.
- Negatives → prefix + hash, not sliding window.
