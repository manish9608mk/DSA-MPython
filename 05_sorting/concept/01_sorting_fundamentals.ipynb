# 01. Sorting Fundamentals

Sorting puts elements in a fixed order (usually ascending). In interviews, sorting is mostly a **tool**: sort first, and many problems become easy.

**Learning order:** WHAT → WHY → PROPERTIES → N LOG N LIMIT → PYTHON SORT → KEY → TRAPS → INTERVIEW THINKING

## 1. What is sorting and why do we need it?

After sorting:
- equal elements sit next to each other (duplicates become easy)
- binary search becomes possible
- two pointers becomes possible (3Sum, pair sum)
- greedy decisions become easy (intervals, scheduling)

**Recognition clue:** brute force is O(n²) because you compare pairs → ask "will sorting make neighbours enough?"

```python
nums = [5, 2, 9, 1]
print(sorted(nums))   # [1, 2, 5, 9]
print(nums)           # original unchanged
```

## 2. Four properties of every sorting algorithm

| Property | Question it answers |
|---|---|
| Time | best / average / worst case? |
| Extra space | in-place (O(1)) or needs extra memory? |
| **Stable** | do equal elements keep their original order? |
| Comparison-based | does it only compare elements, or use their values (counting)? |

**Stable example:** students sorted by marks. If Ravi and Neha both have 80, a stable sort keeps Ravi before Neha if he was before in the input.

```python
students = [("Ravi", 80), ("Amit", 90), ("Neha", 80), ("Sita", 90)]
by_marks = sorted(students, key=lambda s: s[1])
print(by_marks)
# Ravi stays before Neha, Amit stays before Sita -> stable
```

## 3. Why O(n log n) is the limit for comparison sorts

n elements can be arranged in n! ways. Each comparison has 2 outcomes, so it cuts the possibilities at most in half. To find the one correct order you need at least log₂(n!) ≈ n log n comparisons.

So no comparison-based sort can beat O(n log n) in the worst case.
Counting / bucket / radix sort *can* be O(n) because they use the values themselves instead of comparing (notebook 05).

## 4. All algorithms at a glance

| Algorithm | Best | Average | Worst | Extra space | Stable |
|---|---|---|---|---|---|
| Bubble | O(n) | O(n²) | O(n²) | O(1) | Yes |
| Selection | O(n²) | O(n²) | O(n²) | O(1) | No |
| Insertion | O(n) | O(n²) | O(n²) | O(1) | Yes |
| Merge | O(n log n) | O(n log n) | O(n log n) | O(n) | Yes |
| Quick | O(n log n) | O(n log n) | O(n²) | O(log n) stack | No |
| Counting | O(n + k) | O(n + k) | O(n + k) | O(k) | Can be |
| Python `sort` (Timsort) | O(n) | O(n log n) | O(n log n) | O(n) | Yes |

`k` = range of values.

## 5. Python sorting: `sorted()` vs `.sort()`

- `sorted(x)` returns a **new list** and works on any iterable.
- `list.sort()` sorts **in place** and returns `None`.
- Python uses Timsort: stable, O(n log n) worst case, O(n) on already sorted data.

**Common bug:** `a = a.sort()` makes `a` equal to `None`.

```python
a = [3, 1, 2]
r = a.sort()
print(r)   # None
print(a)   # [1, 2, 3]

print(sorted("cab"))            # ['a', 'b', 'c']  (a list, not a string)
print("".join(sorted("cab")))   # abc
```

## 6. Custom order with `key`

`key` is a function that gives each element a **sort value**. Python sorts by that value.

- Sort by length: `key=len`
- Sort by several rules: return a tuple
- Descending number but ascending text: use `-number` in the tuple
- `reverse=True` still keeps stability

```python
words = ["pear", "fig", "banana", "kiwi"]
print(sorted(words, key=len))        # ['fig', 'pear', 'kiwi', 'banana']

people = [("Amit", 90), ("Ravi", 80), ("Neha", 90)]
print(sorted(people, key=lambda p: (-p[1], p[0])))
# marks high -> low, then name A -> Z
```

## 7. Two traps to remember

**Trap 1: sorting loses original indexes.** For problems that return indexes (Two Sum), sort pairs `(value, index)`.

**Trap 2: sorting is not always the best idea.** Duplicate check: sort = O(n log n) time, O(1)–O(n) space; hash set = O(n) time, O(n) space. Say the tradeoff out loud.

```python
nums = [3, 2, 4]
pairs = sorted((v, i) for i, v in enumerate(nums))
print(pairs)   # [(2, 1), (3, 0), (4, 2)] -> value + original index
```

## 8. Interview thinking

1. Brute force compares many pairs? → can sorting reduce it to neighbours?
2. Does the order of the original array matter in the answer (indexes, subsequence)? → careful with sort.
3. Say: "Sorting costs O(n log n), and that dominates the rest."
4. If the question is *about* sorting, implement it yourself; otherwise use built-in.
5. Ask: is stability needed? Are values in a small range? (counting sort)
