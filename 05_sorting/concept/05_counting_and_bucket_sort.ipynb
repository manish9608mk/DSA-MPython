# 05. Counting Sort and Bucket Sort

These sorts do **not compare** elements. They use the values themselves, so they can beat O(n log n), but only when the value range is small.

**Learning order:** COUNTING SORT → LIMITS → LOWERCASE LETTERS TRICK → BUCKET SORT → RADIX (IDEA) → INTERVIEW THINKING

## 1. Counting sort

**Idea:** count how many times each value appears, then write values back in order.

For values in range `[lo, hi]`, make an array of size `hi - lo + 1`.
**Time:** O(n + k), **space:** O(k), where `k = hi - lo + 1`.

Subtracting `lo` also handles negative numbers.

```python
def counting_sort(arr):
    if not arr:
        return []
    lo, hi = min(arr), max(arr)
    count = [0] * (hi - lo + 1)
    for x in arr:
        count[x - lo] += 1
    result = []
    for i, c in enumerate(count):
        result.extend([i + lo] * c)
    return result

print(counting_sort([4, 2, 2, 8, 3, 3, 1]))   # [1, 2, 2, 3, 3, 4, 8]
print(counting_sort([-2, 3, -1, 3, 0]))       # [-2, -1, 0, 3, 3]
```

## 2. When NOT to use it

- Range `k` is huge (e.g. values up to 10⁹) → memory and time blow up.
- Values are not integers (floats, objects) without a mapping.
- This simple version rebuilds numbers, so it does not keep the original records. A stable version uses prefix sums of the counts.

**Rule:** use it only when `k` is about O(n) or smaller.

## 3. Useful trick: sorting lowercase letters in O(n)

Only 26 possible values → counting sort with a fixed array of size 26. Works for anagram signatures too.

```python
def sort_lowercase(s):
    count = [0] * 26
    for ch in s:
        count[ord(ch) - ord("a")] += 1
    return "".join(chr(ord("a") + i) * c for i, c in enumerate(count))

print(sort_lowercase("banana"))   # aaabnn
```

## 4. Bucket sort idea

Make buckets, drop each element into its bucket, then read the buckets in order.

Best known interview use: **Top K Frequent Elements**. Frequencies are between 1 and n, so make `n + 1` buckets where **index = frequency**. Read from the highest frequency down. This avoids sorting by frequency (O(n log n)) and gives O(n).

```python
from collections import Counter

def top_k_frequent(nums, k):
    freq = Counter(nums)
    buckets = [[] for _ in range(len(nums) + 1)]   # index = frequency
    for num, f in freq.items():
        buckets[f].append(num)
    result = []
    for f in range(len(nums), 0, -1):
        for num in buckets[f]:
            result.append(num)
            if len(result) == k:
                return result
    return result

print(top_k_frequent([1, 1, 1, 2, 2, 3], 2))   # [1, 2]
```

## 5. Radix sort (idea only)

Sort numbers digit by digit (least significant first) using a **stable** counting sort for each digit. Time O(d · (n + b)) with `d` digits and base `b`. It is why stability matters. Used in *Maximum Gap* (Hard).

## 6. Interview thinking

**Signals:** "values are in a small range", "only lowercase letters", "0/1/2", "frequency up to n", "sort in O(n)".

**Say:** "Comparison sort is O(n log n), but values lie in `[0, k]`, so counting sort gives O(n + k) time and O(k) space."

**Common mistakes:** forgetting to offset negatives, using it when `k` is huge, assuming it is always faster, ignoring the extra O(k) memory.

## Quick Revision

- Counting sort: count values, rebuild. O(n + k) time, O(k) space.
- Offset by `min` for negatives. Do not use when range is huge.
- 26-letter array = O(n) sort for lowercase strings.
- Bucket by frequency = Top K Frequent in O(n).
- Radix = counting sort per digit, needs stability.
