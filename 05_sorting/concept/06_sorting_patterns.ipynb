## 1. Pattern: sort, then compare neighbours

**Clue:** duplicates, closest pair, minimum difference, "are any two elements …".

After sorting, the best pair is always adjacent. Time O(n log n).

```python
def has_duplicate(nums):
    s = sorted(nums)
    return any(s[i] == s[i + 1] for i in range(len(s) - 1))

def min_abs_difference(nums):
    s = sorted(nums)
    return min(s[i + 1] - s[i] for i in range(len(s) - 1))

print(has_duplicate([3, 1, 4, 1]))     # True
print(min_abs_difference([4, 2, 1, 3]))   # 1
```

## 2. Pattern: sort + two pointers

**Clue:** pair / triplet with a target sum, "return values" (not indexes).

Sort, then `left` at start, `right` at end. Sum too small → `left += 1`. Too big → `right -= 1`. This is the base of 3Sum and 4Sum.

```python
def has_pair_with_sum(nums, target):
    s = sorted(nums)
    left, right = 0, len(s) - 1
    while left < right:
        total = s[left] + s[right]
        if total == target:
            return True
        if total < target:
            left += 1
        else:
            right -= 1
    return False

print(has_pair_with_sum([8, 1, 5, 3], 9))   # True (8 + 1)
```

## 3. Pattern: sort by start time (intervals)

**Clue:** meetings, overlapping ranges, merge, schedule.

Sort by **start**. Then only the previous interval matters:
- overlap if `current.start < previous.end` (touching counts as overlap only if the problem says so)
- merge: extend previous end with `max(previous.end, current.end)`

```python
def can_attend_all(intervals):
    intervals = sorted(intervals)
    for i in range(1, len(intervals)):
        if intervals[i][0] < intervals[i - 1][1]:
            return False
    return True

def merge_intervals(intervals):
    merged = []
    for start, end in sorted(intervals):
        if merged and start <= merged[-1][1]:
            merged[-1][1] = max(merged[-1][1], end)
        else:
            merged.append([start, end])
    return merged

print(can_attend_all([[0, 30], [5, 10], [15, 20]]))      # False
print(merge_intervals([[1, 3], [2, 6], [8, 10], [15, 18]]))   # [[1, 6], [8, 10], [15, 18]]
```

## 4. Pattern: custom comparison order

**Clue:** "arrange to form the largest number", "order by a rule that involves two items together".

When one `key` value is not enough, use `functools.cmp_to_key` with a comparator: return `-1` if `a` should come first, `1` if `b` should, `0` if equal.

*Largest Number:* `a` before `b` if `a + b > b + a` (as strings).

```python
from functools import cmp_to_key

def largest_number(nums):
    strs = list(map(str, nums))
    def compare(a, b):
        if a + b > b + a:
            return -1
        if a + b < b + a:
            return 1
        return 0
    strs.sort(key=cmp_to_key(compare))
    result = "".join(strs)
    return "0" if result[0] == "0" else result

print(largest_number([3, 30, 34, 5, 9]))   # 9534330
```

## 5. Pattern: count first, then sort by count

**Clue:** "sort by frequency", "top k", "group by".

Hashing gives counts; sorting orders the distinct values. Cost O(n + k log k) where `k` = distinct values. For top-k only, a heap or bucket sort is better.

```python
from collections import Counter

def sort_by_frequency(s):
    freq = Counter(s)
    ordered = sorted(freq.items(), key=lambda item: -item[1])
    return "".join(ch * f for ch, f in ordered)

print(sort_by_frequency("tree"))   # eetr
```

## 6. Which technique to pick?

| Signal in the question | Technique |
|---|---|
| duplicates / closest pair | sort + neighbours |
| pair or triplet sum, return values | sort + two pointers |
| intervals, meetings, merge | sort by start |
| Kth largest / median | quickselect or heap |
| 0/1/2 or few distinct values | 3-way partition / counting |
| values in small range, lowercase letters | counting sort |
| count pairs `i < j` with `a[i] > a[j]` | merge sort |
| order depends on a pair rule | `cmp_to_key` |
| must return original indexes | sort `(value, index)` or use hashing |

## 7. Interview thinking

1. State brute force first (usually O(n²)).
2. Ask: "does sorting remove the need to compare all pairs?"
3. Say the cost: "sort is O(n log n), then one pass O(n)."
4. Check: does the answer need original order or indexes?
5. Mention a faster alternative (hash set, heap, bucket) when one exists.

**Common mistakes:** forgetting `.sort()` returns `None`, sorting when indexes are needed, sorting intervals by end when the template needs start, mutating the caller's input without saying so.
