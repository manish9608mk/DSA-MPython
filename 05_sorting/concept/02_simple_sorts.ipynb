# 02. Simple Sorts: Bubble, Selection, Insertion

These are O(n²) sorts. You will rarely write them in an interview, but they teach the idea of an **invariant** (something that stays true after every step), and **insertion sort** is genuinely useful.

**Learning order:** BUBBLE → SELECTION → INSERTION → COMPARE → INTERVIEW THINKING

## 1. Bubble sort

**Idea:** compare neighbours and swap if they are in the wrong order. After each pass the largest remaining element "bubbles" to the end.

**Invariant:** after pass `p`, the last `p` elements are in their final place.

**Early exit:** if a full pass makes no swap, the array is sorted → best case O(n).

```python
def bubble_sort(arr):
    n = len(arr)
    for end in range(n - 1, 0, -1):
        swapped = False
        for i in range(end):
            if arr[i] > arr[i + 1]:
                arr[i], arr[i + 1] = arr[i + 1], arr[i]
                swapped = True
        if not swapped:      # already sorted
            break
    return arr

print(bubble_sort([5, 1, 4, 2]))   # [1, 2, 4, 5]
```

## 2. Selection sort

**Idea:** find the minimum of the unsorted part and put it at the front.

**Invariant:** `arr[0..i-1]` holds the `i` smallest elements in final order.

Only O(n) swaps, but always O(n²) comparisons. **Not stable**, because the long-distance swap can jump over an equal element.

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n - 1):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

print(selection_sort([64, 25, 12, 22, 11]))   # [11, 12, 22, 25, 64]

# Why it is not stable: sort records by first value only
recs = [(2, "a"), (2, "b"), (1, "c")]
for i in range(len(recs) - 1):
    m = i
    for j in range(i + 1, len(recs)):
        if recs[j][0] < recs[m][0]:
            m = j
    recs[i], recs[m] = recs[m], recs[i]
print(recs)   # [(1, 'c'), (2, 'b'), (2, 'a')]  -> 2a and 2b swapped order
```

## 3. Insertion sort

**Idea:** like sorting playing cards in your hand. Take the next element and insert it into the sorted part on its left.

**Invariant:** `arr[0..i-1]` is sorted (not final, just sorted among themselves).

Use `>` (not `>=`) in the shifting condition, otherwise stability breaks.

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:   # shift bigger elements right
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key                 # drop key into the gap
    return arr

print(insertion_sort([5, 2, 4, 1]))   # [1, 2, 4, 5]
```

```python
# Trace: watch the sorted part grow
arr = [5, 2, 4, 1]
for i in range(1, len(arr)):
    key, j = arr[i], i - 1
    while j >= 0 and arr[j] > key:
        arr[j + 1] = arr[j]
        j -= 1
    arr[j + 1] = key
    print(f"after inserting {key}: {arr}")
```

## 4. Connection: shifts = inversions

An **inversion** is a pair `i < j` with `arr[i] > arr[j]`. Insertion sort does exactly one shift per inversion. This is why it is O(n) on nearly sorted data (few inversions) and O(n²) on reversed data (n(n-1)/2 inversions).

You will use this idea again in merge sort (Count Inversions).

```python
arr = [2, 4, 1, 3, 5]
shifts = 0
for i in range(1, len(arr)):
    key, j = arr[i], i - 1
    while j >= 0 and arr[j] > key:
        arr[j + 1] = arr[j]
        j -= 1
        shifts += 1
    arr[j + 1] = key
print(shifts)   # 3  -> inversions (2,1) (4,1) (4,3)
```

## 5. Compare the three

| | Bubble | Selection | Insertion |
|---|---|---|---|
| Best | O(n) with early exit | O(n²) | O(n) |
| Worst | O(n²) | O(n²) | O(n²) |
| Swaps / writes | many | O(n) | one per inversion |
| Stable | Yes | No | Yes |
| Good for | teaching | few writes | small or nearly sorted data |

## 6. Interview thinking

- If asked "why not use bubble/selection?" → O(n²) is too slow for large n.
- Insertion sort is used inside real libraries for tiny arrays (Timsort uses it).
- Fast answer for any simple sort: "n² comparisons, O(1) space, nested loops."
- Always state the invariant when you explain a sorting loop. Interviewers like it.

## Quick Revision

- Bubble: swap neighbours, largest goes to end, early exit → O(n) best.
- Selection: pick minimum, O(n) swaps, **not stable**.
- Insertion: build sorted prefix, shifts = inversions, best for nearly sorted.
- All three: O(n²) worst, O(1) space.
- Use `>` not `>=` in insertion to stay stable.
