# 01. Dynamic Programming: Fundamentals

DP = solve a big problem by solving **smaller overlapping subproblems once** and **reusing their answers**.

**Learning order:** REAL LIFE → WHEN DP APPLIES → FIBONACCI 4 WAYS → TOP-DOWN VS BOTTOM-UP → 5-STEP FRAMEWORK → CLIMBING STAIRS → MIN COST STAIRS → COMPLEXITY → INTERVIEW THINKING

## 1. Real life

A teacher writes `1 + 1 + 1 + 1 + 1 + 1 + 1 + 1`. "What is it?" You say **8**. She adds `+ 1` at the end. "Now?" You say **9** instantly, **without recounting**. You **remembered** the earlier answer. That is DP.

Other examples: shortest route that reuses already-computed shortest routes, making change, editing one word into another with fewest edits.

## 2. When does DP apply?

Two properties:
1. **Overlapping subproblems:** the same smaller problem appears again and again (the recursion tree repeats work).
2. **Optimal substructure:** the best answer for the big problem is built from best answers of smaller ones.

**Recognition clues:**
- "**number of ways** to ...", "**minimum/maximum** cost/length/profit", "**is it possible** to ..." (true/false)
- Choices at each step + the same states reappear
- Brute force is exponential recursion

Not DP: you need to **list** all solutions (backtracking), or a local choice is provably best (greedy).

## 3. Fibonacci in four ways (the whole DP journey)

1. **Naive recursion:** O(2ⁿ), repeats work.
2. **Memoization (top-down):** same recursion + a cache. O(n).
3. **Tabulation (bottom-up):** fill a table from the base cases up. O(n).
4. **Space-optimized:** keep only the last two values. O(n) time, **O(1)** space.

```python
from functools import lru_cache

def fib_naive(n):
    return n if n <= 1 else fib_naive(n - 1) + fib_naive(n - 2)

@lru_cache(maxsize=None)
def fib_memo(n):
    return n if n <= 1 else fib_memo(n - 1) + fib_memo(n - 2)

def fib_table(n):
    if n <= 1:
        return n
    dp = [0] * (n + 1)
    dp[1] = 1
    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
    return dp[n]

def fib_optimized(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a

print(fib_naive(20), fib_memo(20), fib_table(20), fib_optimized(20))   # 6765 6765 6765 6765
print(fib_optimized(90))   # 2880067194370816120
```

## 4. Top-down vs Bottom-up

| | Top-down (memoization) | Bottom-up (tabulation) |
|---|---|---|
| How | recursion + cache | loops filling a table |
| Easier to write | **yes**, follows the recurrence directly | needs the right order |
| Computes | only needed states | all states |
| Recursion limit | can hit Python's limit (~1000) | no |
| Space optimization | harder | easy (rolling array) |

**Interview tip:** start with the top-down recurrence (clear), then mention the bottom-up and the space optimization.

## 5. The 5-step framework (use it for every DP problem)

1. **State:** what do I need to know to describe a subproblem? `dp[i]`, `dp[i][j]`, ... **Write its meaning in words.**
2. **Transition:** how does a state depend on smaller states? (the recurrence)
3. **Base cases:** the smallest states with obvious answers.
4. **Order:** compute smaller states first (loop direction).
5. **Answer:** which state holds the final answer?

**Complexity = (number of states) × (work per transition).**

## 6. Climbing Stairs

You climb 1 or 2 steps. Ways to reach step `n`?

1. State: `dp[i]` = number of ways to reach step `i`.
2. Transition: last move was 1 or 2 steps: `dp[i] = dp[i-1] + dp[i-2]`.
3. Base: `dp[0] = 1`, `dp[1] = 1`.
4. Order: increasing `i`. 5. Answer: `dp[n]`.

```python
def climb_stairs(n):
    a, b = 1, 1
    for _ in range(n - 1):
        a, b = b, a + b
    return b

print(climb_stairs(2), climb_stairs(3), climb_stairs(10))   # 2 3 89
```

## 7. Min Cost Climbing Stairs

`cost[i]` is paid when you **leave** step `i`; you may start at step 0 or 1 and finish above the last step.

State: `dp[i]` = min cost to **stand on** step `i`. `dp[i] = cost[i] + min(dp[i-1], dp[i-2])`. Answer = `min(dp[n-1], dp[n-2])` (you can finish from either).

```python
def min_cost_climbing_stairs(cost):
    a, b = cost[0], cost[1]
    for i in range(2, len(cost)):
        a, b = b, cost[i] + min(a, b)
    return min(a, b)

print(min_cost_climbing_stairs([10, 15, 20]), min_cost_climbing_stairs([1, 100, 1, 1, 1, 100, 1, 1, 100, 1]))   # 15 6
```

## 8. Interview thinking

**Say:** "Brute force recursion repeats subproblems. The state is `dp[...]` meaning ..., transition ..., base ... Time is states × transitions, space can be reduced to ... because only the last row/values are needed."

**Common mistakes:** unclear state definition, wrong base case, iterating in the wrong order, off-by-one in indexes (`dp[n+1]` vs `dp[n]`), forgetting the answer is not always `dp[n]`, using DP when greedy or backtracking is the right tool.

## Quick Revision

- DP = overlapping subproblems + optimal substructure.
- 5 steps: state, transition, base, order, answer. Say the state's meaning in words.
- Memo (top-down) = recursion + cache. Tabulation = loops. Rolling variables save space.
- Time = states × transition cost.
