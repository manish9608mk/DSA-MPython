## 1. Pattern families

| Family | State | Examples |
|---|---|---|
| **Linear (1D)** | `dp[i]` | climbing stairs, house robber, decode ways, max subarray |
| **Coins / unbounded knapsack** | `dp[amount]` | coin change I/II, perfect squares, combination sum IV |
| **0/1 knapsack** | `dp[capacity]` | subset sum, partition, target sum, last stone II |
| **LIS** | `dp[i]` ending at `i` | LIS, Russian dolls, number of LIS |
| **Two strings** | `dp[i][j]` | LCS, edit distance, distinct subsequences, wildcard |
| **Grid** | `dp[r][c]` | unique paths, min path sum, maximal square, dungeon |
| **Interval** | `dp[l][r]` | burst balloons, matrix chain, palindrome partition |
| **State machine** | `dp[i][state]` | stocks, paint house |
| **Bitmask** | `dp[mask]` | TSP, partition k subsets, visit all nodes |
| **Tree DP** | return tuple | house robber III, diameter, max path sum |
| **Digit DP** | `(pos, tight, ...)` | count numbers with a property |

## 2. How to find the state

1. Write the brute-force recursion. List its **parameters**.
2. Remove parameters that are not needed to decide the future (what you must remember: index, remaining capacity, last choice).
3. Those parameters are the state. Memoize on them.
4. Convert to a table if you want (order = dependencies).

**Question to ask:** "What is the *minimum* information I need to continue from here?"

## 4. Decision flow

1. Is it "list all solutions"? → backtracking.
2. Is a local greedy choice provably best? → greedy.
3. "Count ways / min / max / possible" with repeating subproblems? → DP.
4. 2 sequences → `dp[i][j]`. 1 sequence + choices → `dp[i]`. Capacity → knapsack. Range → interval. n ≤ 20 sets → bitmask.

## 5. Common mistakes

- Vague state meaning ("dp[i] is the answer for i").
- Wrong base cases or off-by-one indexes.
- Wrong loop order or direction (0/1 vs unbounded).
- Combinations vs permutations loop order.
- Modifying the same table you read from in a way that reuses items.
- Forgetting that the answer is not always the last cell.
- Not saying complexity: states × transitions.
- Using memoization with mutable arguments (lists).

## 6. Interview script

1. "Let me first write the brute-force recursion." (parameters = candidate state)
