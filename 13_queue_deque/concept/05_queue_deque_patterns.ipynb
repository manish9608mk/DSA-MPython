# 05. Queue and Deque Patterns (Revision Sheet)

**Learning order:** DECISION TABLE → COMPARE STACK / QUEUE / DEQUE → MISTAKES → INTERVIEW SCRIPT → PRACTICE MAP

## 1. Decision table

| Question looks like | Pattern |
|---|---|
| process in arrival order, simulate a line | queue (`deque`) |
| level order, minimum steps, shortest unweighted path | BFS |
| several starting points spread together | multi-source BFS |
| max/min of every window of size k | monotonic deque |
| window where `max - min <= limit` | two monotonic deques |
| shortest subarray with sum ≥ k, negatives allowed | prefix sums + deque |
| last N seconds / events | deque of timestamps |
| implement queue with stacks (or reverse) | two stacks / rotation |
| fixed-size buffer | circular queue |

## 2. Stack vs Queue vs Deque

| | Stack | Queue | Deque |
|---|---|---|---|
| Order | LIFO | FIFO | both ends |
| Add / remove | end | back / front | any end |
| Typical use | nesting, undo, DFS, monotonic stack | BFS, scheduling | sliding window max, palindromes |

## 3. Common mistakes

- `list.pop(0)` instead of `deque.popleft()`.
- Marking visited on pop in BFS.
- Storing values instead of indexes in a monotonic deque.
- Wrong window expiry condition.
- Not processing level by level when counting levels.
- Missing stop condition in simulations.
- Using BFS on weighted graphs.

## 4. Interview script

1. "Order of arrival matters, so a queue."
2. "I use `collections.deque` for O(1) operations at both ends."
3. For window max: "monotonic deque of indexes, O(n)."
4. For BFS: "O(V + E), mark visited when enqueuing."
5. Test: empty, one element, `k` equal to length, all equal values.

## Quick Revision

- Queue FIFO via `deque`. BFS = queue + visited.
- Monotonic deque = sliding window max/min, O(n).
- Time windows = deque of timestamps.
- Never `pop(0)`. Mark visited on enqueue.
