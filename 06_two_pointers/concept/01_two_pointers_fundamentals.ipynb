# 01. Two Pointers: Fundamentals

Two pointers means using **two indexes** that move through the data, instead of a nested loop. It turns many O(n²) solutions into O(n).

**Learning order:** WHAT → REAL LIFE → WHY O(n) → 3 TYPES → TEMPLATES → WHEN TO USE → INTERVIEW THINKING

## 1. What is it?

Instead of checking every pair `(i, j)` with two loops, keep two pointers and decide **which one to move** using a rule. Each pointer only moves in one direction, so we never go back.

## 2. Real-life examples

- **Gift card:** items are sorted by price. You want two items that add up to ₹1000. Start with the cheapest and the costliest. Too cheap → take a pricier first item. Too costly → drop the costliest. (opposite ends)
- **Shelf cleaning:** one hand scans the shelf, the other hand places good books at the front. (same direction, slow and fast)
- **Two queues merging at a gate:** compare the two front people, let the smaller token number in first. (two arrays)

## 3. Why is it O(n)?

Each pointer moves in one direction and never goes back. Together they make at most `n` moves. Nested loops make `n × n` checks, two pointers make about `n`.

````python
# Brute force pair sum on a sorted array: O(n^2)
def pair_sum_brute(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return (i, j)
    return None

# Two pointers: O(n)
def pair_sum_two_pointers(nums, target):
    left, right = 0, len(nums) - 1
    while left < right:
        total = nums[left] + nums[right]
        if total == target:
            return (left, right)
        if total < target:
            left += 1       # need a bigger sum
        else:
            right -= 1      # need a smaller sum
    return None

nums = [1, 3, 4, 6, 8, 11]
print(pair_sum_brute(nums, 10), pair_sum_two_pointers(nums, 10))   # (2, 3) (2, 3)
````

## 4. The three types

| Type | Pointers | Typical problems |
|---|---|---|
| **Opposite ends** | `left = 0`, `right = n-1`, move towards each other | pair sum, palindrome, container with water, 3Sum |
| **Same direction** | `slow` (write position) and `fast` (reader) | remove duplicates, move zeroes, in-place filter |
| **Two arrays / strings** | one pointer in each | merge sorted arrays, is subsequence, intersection |

## 5. Templates

**Opposite ends**
```python
left, right = 0, len(arr) - 1
while left < right:
    if condition_met:
        ...
    elif need_bigger:
        left += 1
    else:
        right -= 1
```

**Same direction (slow / fast)**
```python
slow = 0
for fast in range(len(arr)):
    if keep(arr[fast]):
        arr[slow] = arr[fast]
        slow += 1
```

**Two arrays**
```python
i = j = 0
while i < len(a) and j < len(b):
    ...
```

````python
# Two small warm-ups using opposite ends
def reverse_in_place(arr):
    left, right = 0, len(arr) - 1
    while left < right:
        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1
    return arr

def is_palindrome(s):
    left, right = 0, len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True

print(reverse_in_place([1, 2, 3, 4, 5]))   # [5, 4, 3, 2, 1]
print(is_palindrome("level"), is_palindrome("hello"))   # True False
````

## 6. When should I think of two pointers?

**Clues:**
- array or string is **sorted** (or you are allowed to sort it)
- you need a **pair / triplet** with some property
- you must work **in place** with O(1) extra space
- you compare the two **ends** (palindrome)
- you merge or compare **two sorted sequences**

**Not a good fit:** unsorted array where you need original indexes (use hashing), or a contiguous subarray with a changing window (that is *sliding window*, next topic).

## 7. Interview thinking

1. Say the brute force first: "checking all pairs is O(n²)."
2. Ask: is the data sorted, or can I sort it? Does the answer need the original indexes?
3. Explain **why a pointer can be moved safely**. This is the proof interviewers want. Example: "sum is too small, so `left` with any `right` that is smaller will be even smaller, so `left` is useless. I can drop it."
4. State complexity: O(n) after sorting, O(n log n) if sorting is needed, O(1) extra space.

## Quick Revision

- Two pointers = two indexes, no going back → O(n).
- 3 types: opposite ends, slow/fast, two arrays.
- Needs a rule to move one pointer, and a reason why that move is safe.
- Best when data is sorted or in-place work is required.
- Unsorted + need indexes → hashing instead.
