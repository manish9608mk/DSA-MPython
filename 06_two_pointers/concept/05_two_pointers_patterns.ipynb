## 1. Trapping Rain Water (two pointers version)

Water above bar `i` = `min(max height on left, max height on right) - height[i]`.

**Two pointers trick:** keep `left_max` and `right_max`. Whichever of `height[left]` and `height[right]` is **smaller**, that side is decided by its own max, because the other side already has a taller wall of at least `height[right]` (or `height[left]`). So process that side and move inwards.

Real life: rain on a skyline of buildings, water is held between taller buildings.

```python
def trap(height):
    if not height:
        return 0
    left, right = 0, len(height) - 1
    left_max = right_max = 0
    water = 0
    while left < right:
        if height[left] < height[right]:
            left_max = max(left_max, height[left])
            water += left_max - height[left]
            left += 1
        else:
            right_max = max(right_max, height[right])
            water += right_max - height[right]
            right -= 1
    return water

print(trap([0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]))   # 6
```

## 2. Decision table

| Signal | Variant |
|---|---|
| sorted array, find pair with sum | opposite ends |
| palindrome / compare from both ends | opposite ends |
| triplets / quadruplets with a sum | sort + fix + two pointers |
| max area, water between walls | opposite ends, move the shorter side |
| remove / move / partition in place | slow and fast (writer / reader) |
| merge two sorted arrays, subsequence, intersection | one pointer per array |
| 0 / 1 / 2 sorting in one pass | three pointers (Dutch flag, sorting notebook) |

## 3. Two pointers vs other patterns

| Looks similar | Difference |
|---|---|
| **Hashing** (Two Sum) | array is unsorted or you need original indexes → hash map O(n) space |
| **Sliding window** | pointers define a *contiguous window* whose content you track (sum, count, set). Two pointers just moves based on a comparison. Next topic. |
| **Binary search** | halves the search space each step, O(log n). Two pointers walks element by element, O(n). |

Quick test: "do I need to remember what is *between* the pointers?" → sliding window. "Only compare the two end values?" → two pointers.

## 4. Common mistakes

- Forgetting to sort (or sorting when indexes are needed).
- `left <= right` when the same element must not be used twice.
- Moving both pointers when only one should move.
- Skipping duplicates in the wrong place (3Sum).
- Missing the bounds check in nested `while` loops.
- Not writing the invariant, then losing track of what `slow` means.

## 5. Interview script

1. "Brute force checks all pairs, O(n²)."
2. "The array is sorted, so I can use two pointers."
3. "If the sum is too small, the left element can't pair with anything smaller than `right`, so I discard it." (proof)
4. "Each pointer moves at most n times → O(n) time, O(1) space."
5. Test with: empty array, one element, all equal, duplicates, negatives.
