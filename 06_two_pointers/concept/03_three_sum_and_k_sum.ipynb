# 03. 3Sum and K-Sum

**Idea:** fix one number, then solve a pair-sum on the rest with two pointers. 3Sum becomes `n` × O(n) = O(n²).

**Learning order:** 3SUM → DUPLICATES → 3SUM CLOSEST → K-SUM IDEA → INTERVIEW THINKING

## 1. 3Sum

**Problem:** find all unique triplets with sum 0.

**Plan:**
1. Sort the array.
2. Fix `i`. Now we need two numbers with sum `-nums[i]` in `nums[i+1:]` → two pointers.
3. Skip duplicates so the same triplet is not repeated.

Real-life: three expenses that cancel out to a net of 0.

```python
def three_sum(nums):
    nums = sorted(nums)
    result = []
    for i in range(len(nums) - 2):
        if nums[i] > 0:                         # smallest is positive -> no zero sum
            break
        if i > 0 and nums[i] == nums[i - 1]:    # skip duplicate first number
            continue
        left, right = i + 1, len(nums) - 1
        while left < right:
            total = nums[i] + nums[left] + nums[right]
            if total == 0:
                result.append([nums[i], nums[left], nums[right]])
                left += 1
                right -= 1
                while left < right and nums[left] == nums[left - 1]:
                    left += 1                   # skip duplicate second number
                while left < right and nums[right] == nums[right + 1]:
                    right -= 1                  # skip duplicate third number
            elif total < 0:
                left += 1
            else:
                right -= 1
    return result

print(three_sum([-1, 0, 1, 2, -1, -4]))   # [[-1, -1, 2], [-1, 0, 1]]
```

## 2. Why the duplicate skips are placed like this

- Skip duplicates of `nums[i]` **before** using it (compare with the previous `i`).
- Skip duplicates of `left` and `right` **only after** a valid triplet is found. Otherwise you would skip a number that is needed for a different pair (for example `[-1, -1, 2]` uses two `-1`).

**Complexity:** sort O(n log n) + n × O(n) = **O(n²)** time. Extra space O(1) apart from the sort and the output.

## 3. 3Sum Closest

Same plan, but instead of looking for an exact match, keep the sum closest to `target`.
If total < target → `left += 1`, if total > target → `right -= 1`. An exact match returns immediately.

```python
def three_sum_closest(nums, target):
    nums = sorted(nums)
    closest = nums[0] + nums[1] + nums[2]
    for i in range(len(nums) - 2):
        left, right = i + 1, len(nums) - 1
        while left < right:
            total = nums[i] + nums[left] + nums[right]
            if abs(total - target) < abs(closest - target):
                closest = total
            if total == target:
                return total
            if total < target:
                left += 1
            else:
                right -= 1
    return closest

print(three_sum_closest([-1, 2, 1, -4], 1))   # 2
```

## 4. K-Sum: the general idea

- Sort once.
- For `k > 2`: fix one number, recurse with `k - 1` on the rest.
- For `k == 2`: two pointers.

Time: **O(n^(k-1))** (3Sum n², 4Sum n³). Two pointers saves one full loop compared to brute force.

```python
def k_sum(nums, target, k):
    nums = sorted(nums)

    def helper(start, k, target):
        result = []
        if k == 2:
            left, right = start, len(nums) - 1
            while left < right:
                total = nums[left] + nums[right]
                if total == target:
                    result.append([nums[left], nums[right]])
                    left += 1
                    right -= 1
                    while left < right and nums[left] == nums[left - 1]:
                        left += 1
                elif total < target:
                    left += 1
                else:
                    right -= 1
            return result
        for i in range(start, len(nums) - k + 1):
            if i > start and nums[i] == nums[i - 1]:
                continue
            for rest in helper(i + 1, k - 1, target - nums[i]):
                result.append([nums[i]] + rest)
        return result

    return helper(0, k, target)

print(k_sum([1, 0, -1, 0, -2, 2], 0, 4))
# [[-2, -1, 1, 2], [-2, 0, 0, 2], [-1, 0, 0, 1]]
```

## 5. Interview thinking

**Signals:** "find all unique triplets / quadruplets", "sum equals target", "closest sum".

**Say:** "Brute force is O(n³). I sort, fix one element, and use two pointers for the rest, so O(n²)." Mention that 4Sum is O(n³) and the same idea generalizes.

**Common mistakes:** forgetting to sort, not skipping duplicates (repeated triplets), skipping `left/right` duplicates *before* a match, using `left <= right`, returning indexes (sorting loses them; use hashing for that).

## Quick Revision

- Sort → fix `i` → two pointers on the rest.
- Skip duplicate `i` first; skip duplicate `left/right` after a match.
- 3Sum O(n²), 4Sum O(n³), general k-sum O(n^(k-1)).
- Closest sum: track best difference, same pointer moves.
- Need original indexes → hashing, not this.
