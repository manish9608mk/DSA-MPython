# 04. Same Direction and Two-Array Pointers

Here both pointers move forward. This is the **in-place** pattern: one pointer reads, the other writes.

**Learning order:** SLOW/FAST IDEA → REMOVE DUPLICATES → MOVE ZEROES → REMOVE ELEMENT → SUBSEQUENCE → MERGE FROM THE BACK → INTERVIEW THINKING

## 1. The idea: reader and writer

- `fast` = reader, scans every element.
- `slow` = writer, the position where the next **good** element goes.

**Invariant:** `arr[0:slow]` always holds the answer built so far.

Real life: you scan a to-do list (fast). Every task that is still pending, you write at the next free line at the top (slow).

```python
def remove_duplicates(nums):
    # sorted array, keep one copy of each value, return new length
    if not nums:
        return 0
    slow = 0
    for fast in range(1, len(nums)):
        if nums[fast] != nums[slow]:
            slow += 1
            nums[slow] = nums[fast]
    return slow + 1

nums = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]
k = remove_duplicates(nums)
print(k, nums[:k])   # 5 [0, 1, 2, 3, 4]
```

## 2. Move Zeroes

Keep the order of non-zero elements and push zeroes to the end, in place.

Every time `fast` finds a non-zero, swap it with the element at `slow`, then move `slow`.

```python
def move_zeroes(nums):
    slow = 0
    for fast in range(len(nums)):
        if nums[fast] != 0:
            nums[slow], nums[fast] = nums[fast], nums[slow]
            slow += 1
    return nums

print(move_zeroes([0, 1, 0, 3, 12]))   # [1, 3, 12, 0, 0]
```

## 3. Remove Element

Remove every occurrence of `val` in place and return the new length. The template is exactly: *keep* if it is not `val`.

```python
def remove_element(nums, val):
    slow = 0
    for fast in range(len(nums)):
        if nums[fast] != val:
            nums[slow] = nums[fast]
            slow += 1
    return slow

nums = [3, 2, 2, 3]
k = remove_element(nums, 3)
print(k, nums[:k])   # 2 [2, 2]
```

## 4. Two strings: Is Subsequence

Is `s` a subsequence of `t`? Pointer `i` on `s` moves only when the current character of `t` matches. Real life: checking if your shopping list items appear in order in a store bill.

```python
def is_subsequence(s, t):
    i = 0
    for ch in t:
        if i < len(s) and s[i] == ch:
            i += 1
    return i == len(s)

print(is_subsequence("abc", "ahbgdc"))   # True
print(is_subsequence("axc", "ahbgdc"))   # False
```

## 5. Merge Sorted Array: fill from the back

`nums1` has extra space at the end. If you merge from the **front** you overwrite unread elements. If you fill from the **back** with the larger element, nothing unread gets overwritten.

```python
def merge_sorted(nums1, m, nums2, n):
    i, j, k = m - 1, n - 1, m + n - 1
    while j >= 0:
        if i >= 0 and nums1[i] > nums2[j]:
            nums1[k] = nums1[i]
            i -= 1
        else:
            nums1[k] = nums2[j]
            j -= 1
        k -= 1
    return nums1

print(merge_sorted([1, 2, 3, 0, 0, 0], 3, [2, 5, 6], 3))   # [1, 2, 2, 3, 5, 6]
```

## 6. Fast and slow at different *speeds* (preview)

There is another meaning of slow/fast: `slow` moves 1 step, `fast` moves 2 steps. This detects cycles and finds the middle of a linked list. We will learn it properly in **Linked List**.

## 7. Interview thinking

**Signals:** "in place", "O(1) extra space", "remove / move / partition", "return the new length".

**Say:** "`fast` reads every element, `slow` marks where the next valid element goes. One pass, O(n) time, O(1) space."

**Common mistakes:** returning `slow` vs `slow + 1` (depends on whether `slow` starts at 0 as *count* or as *index*), not stopping at `nums[:k]` when printing, swapping when `slow == fast` is fine but overwriting without keeping the invariant is not, merging from the front.

## Quick Revision

- `fast` reads, `slow` writes. `arr[0:slow]` is the answer so far.
- Remove duplicates: write when `nums[fast] != nums[slow]`.
- Move zeroes: swap non-zero to `slow`.
- Subsequence: move `i` only on a match.
- Merge into array with spare space: fill from the **back**.
