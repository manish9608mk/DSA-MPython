# 02. Opposite Ends Pattern

Start at both ends, move inwards. Each step throws away one element that can never be part of the answer.

**Learning order:** PAIR SUM → WHY IT WORKS → PALINDROMES → SORTED SQUARES → CONTAINER WITH MOST WATER → INTERVIEW THINKING

## 1. Pair sum in a sorted array (Two Sum II)

**Clue:** sorted array, find two numbers with a given sum.

**Why moving a pointer is safe:**
- If `nums[left] + nums[right] < target`, then `nums[left]` with any element at or before `right` gives an even smaller sum. So `nums[left]` can never work. Safe to do `left += 1`.
- If the sum is greater than target, the same argument discards `nums[right]`.

```python
def two_sum_sorted(nums, target):
    left, right = 0, len(nums) - 1
    while left < right:
        total = nums[left] + nums[right]
        if total == target:
            return [left, right]
        if total < target:
            left += 1
        else:
            right -= 1
    return []

print(two_sum_sorted([2, 7, 11, 15], 9))   # [0, 1]
```

## 2. Valid Palindrome (skip non-letters)

**Clue:** compare characters from both ends.

Skip anything that is not a letter or digit, and ignore case. Real life: checking if a sentence or a phone-number-like code reads the same both ways.

```python
def valid_palindrome(s):
    left, right = 0, len(s) - 1
    while left < right:
        while left < right and not s[left].isalnum():
            left += 1
        while left < right and not s[right].isalnum():
            right -= 1
        if s[left].lower() != s[right].lower():
            return False
        left += 1
        right -= 1
    return True

print(valid_palindrome("A man, a plan, a canal: Panama"))   # True
print(valid_palindrome("race a car"))                       # False
```

## 3. Palindrome after deleting at most one character

On the first mismatch we have only two choices: skip the left character or skip the right one. Check if either remaining part is a palindrome. Only one extra check, so still O(n).

```python
def valid_palindrome_ii(s):
    def check(left, right):
        while left < right:
            if s[left] != s[right]:
                return False
            left += 1
            right -= 1
        return True

    left, right = 0, len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return check(left + 1, right) or check(left, right - 1)
        left += 1
        right -= 1
    return True

print(valid_palindrome_ii("abca"))   # True  (delete 'b' or 'c')
print(valid_palindrome_ii("abc"))    # False
```

## 4. Squares of a sorted array

**Clue:** sorted array with negatives. Squares are no longer sorted, but the **largest square is always at one of the two ends**.

Fill the answer from the back: compare `|left|` and `|right|`, place the bigger square at the end.

```python
def sorted_squares(nums):
    n = len(nums)
    result = [0] * n
    left, right = 0, n - 1
    for pos in range(n - 1, -1, -1):
        if abs(nums[left]) > abs(nums[right]):
            result[pos] = nums[left] ** 2
            left += 1
        else:
            result[pos] = nums[right] ** 2
            right -= 1
    return result

print(sorted_squares([-4, -1, 0, 3, 10]))   # [0, 1, 9, 16, 100]
```

## 5. Container With Most Water

Heights are vertical walls. Water between walls `i` and `j` = `min(h[i], h[j]) × (j - i)`. Find the maximum.

**Greedy rule:** move the pointer at the **shorter** wall.

**Why:** water is limited by the shorter wall. If we move the taller wall inwards, the width shrinks and the height is still limited by the same shorter wall, so the area cannot improve. Moving the shorter wall is the only move that might give a taller wall.

```python
def max_area(height):
    left, right = 0, len(height) - 1
    best = 0
    while left < right:
        width = right - left
        best = max(best, min(height[left], height[right]) * width)
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1
    return best

print(max_area([1, 8, 6, 2, 5, 4, 8, 3, 7]))   # 49
```

## 6. Interview thinking

**Signals:** sorted + pair, "from both ends", palindrome, "maximize something using two positions".

**What to say:** "I start at both ends. Each step I discard one element that cannot be part of an answer, so total work is O(n), O(1) space."

**Common mistakes:** `left <= right` instead of `left < right` (pair would use the same element), forgetting to skip non-alphanumeric chars, forgetting to check bounds in the inner `while`, using this on an **unsorted** array.

## Quick Revision

- Pair sum: too small → `left += 1`, too big → `right -= 1`.
- Palindrome: compare ends; on one mismatch try skipping left or right.
- Sorted squares: biggest square is at an end, fill result from the back.
- Container: always move the shorter wall.
- Always `left < right`, O(n) time, O(1) space.
