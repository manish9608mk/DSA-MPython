# 01. Heap and Priority Queue: Fundamentals

A **heap** is a tree-shaped structure that always gives you the **smallest** (min-heap) or **largest** (max-heap) element in **O(1)**, and lets you add/remove in **O(log n)**. A **priority queue** is the abstract idea ("serve the most important first"), a heap is how we build it.

**Learning order:** REAL LIFE → RULE → ARRAY STORAGE → PUSH AND POP → IMPLEMENT → HEAPIFY → PYTHON heapq → INTERVIEW THINKING

## 1. Real-life examples

- **Hospital emergency room:** the most critical patient is treated first, not the first to arrive.
- **Operating system scheduler:** the highest-priority process runs first.
- **Dijkstra's shortest path:** always expand the closest unvisited node.
- **"Top 10 trending" lists**, **leaderboards**, **ride matching** (nearest driver).

**Recognition clue:** "repeatedly need the smallest/largest", "k-th largest", "merge k sorted", "schedule by priority".

## 2. The heap rule

A **complete** binary tree where every parent is `<=` its children (**min-heap**). So the **root is the minimum**.

- It is **not fully sorted**: siblings have no order. That is why it is cheap to maintain.
- Stored in an **array**: for index `i` children are `2i+1`, `2i+2`, parent is `(i-1)//2`.

| Operation | Time |
|---|---|
| peek min | O(1) |
| push | O(log n) |
| pop min | O(log n) |
| build from a list (heapify) | **O(n)** |
| search an arbitrary value | O(n) |

## 3. How push and pop work

- **Push:** add at the **end**, then **sift up**: while it is smaller than its parent, swap.
- **Pop:** take the root. Move the **last** element to the root, then **sift down**: swap with the smaller child until the rule holds.

The tree height is `log n`, so each takes at most `log n` swaps.

```python
class MinHeap:
    def __init__(self):
        self.a = []

    def push(self, x):
        self.a.append(x)
        self._up(len(self.a) - 1)

    def pop(self):
        top = self.a[0]
        last = self.a.pop()
        if self.a:
            self.a[0] = last
            self._down(0)
        return top

    def _up(self, i):
        while i > 0:
            p = (i - 1) // 2
            if self.a[p] <= self.a[i]:
                break
            self.a[p], self.a[i] = self.a[i], self.a[p]
            i = p

    def _down(self, i):
        n = len(self.a)
        while True:
            l, r, smallest = 2 * i + 1, 2 * i + 2, i
            if l < n and self.a[l] < self.a[smallest]: smallest = l
            if r < n and self.a[r] < self.a[smallest]: smallest = r
            if smallest == i:
                break
            self.a[i], self.a[smallest] = self.a[smallest], self.a[i]
            i = smallest

h = MinHeap()
for x in [5, 3, 8, 1, 2]:
    h.push(x)
print([h.pop() for _ in range(5)])   # [1, 2, 3, 5, 8]
```

## 4. Heapify is O(n), not O(n log n)

Building a heap by pushing `n` items costs O(n log n). But **heapify** (sift down every non-leaf, from the last parent to the root) is **O(n)**: half the nodes are leaves (cost 0), a quarter sift at most 1 level, an eighth at most 2... The sum `n/4·1 + n/8·2 + n/16·3 + ...` converges to O(n).

## 5. Python `heapq` (min-heap only)

- `heapq.heappush(h, x)`, `heapq.heappop(h)`, `h[0]` = smallest.
- `heapq.heapify(list)` in place, O(n).
- `heapq.nsmallest(k, it)`, `heapq.nlargest(k, it)`.
- **Max-heap trick:** push **negated** values and negate when popping.
- **Priority + item:** push tuples `(priority, item)`. If priorities tie, Python compares the items, so add a tie-breaker `(priority, counter, item)` when items are not comparable.
- The list is **not sorted**. Only `h[0]` is guaranteed.

```python
import heapq

nums = [5, 3, 8, 1, 2]
heapq.heapify(nums)
print(nums[0], heapq.heappop(nums), nums[0])   # 1 1 2

mh = []                                          # max-heap with negation
for x in [5, 3, 8]:
    heapq.heappush(mh, -x)
print(-heapq.heappop(mh))   # 8

tasks = []
heapq.heappush(tasks, (2, "write report"))
heapq.heappush(tasks, (1, "fix bug"))
print(heapq.heappop(tasks))   # (1, 'fix bug')

print(heapq.nlargest(2, [4, 1, 7, 3]), heapq.nsmallest(2, [4, 1, 7, 3]))   # [7, 4] [1, 3]

def heap_sort(arr):                              # O(n log n)
    h = arr[:]
    heapq.heapify(h)
    return [heapq.heappop(h) for _ in range(len(h))]

print(heap_sort([4, 10, 3, 5, 1]))   # [1, 3, 4, 5, 10]
```

## 6. Heap vs sorting vs BST

| Need | Best tool |
|---|---|
| min/max repeatedly while data changes | **heap** |
| all elements in sorted order once | sort |
| k largest of a big stream | heap of size k |
| min, max, **and** search/delete arbitrary, ordered neighbours | balanced BST / sorted container |

## 7. Interview thinking

**Say:** "I need the smallest/largest repeatedly, so a heap: O(log n) per operation instead of O(n) for scanning or O(n log n) for sorting each time."

**Common mistakes:** expecting `heapq` to be a max-heap, assuming the list is sorted, forgetting that heapq compares tuples element by element (tie → compares the next field), using a heap when only one pass min/max is needed, forgetting that arbitrary deletion is O(n) (use lazy deletion).

## Quick Revision

- Heap = complete tree, parent ≤ children. Root = min.
- Push/pop O(log n), peek O(1), heapify O(n).
- Array: children `2i+1`, `2i+2`.
- `heapq` = min-heap; max-heap by negating. Tuples for priority.
- Not sorted, only the root is guaranteed.
