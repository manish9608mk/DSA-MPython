# 05. Heap Patterns (Revision Sheet)

**Learning order:** DECISION TABLE → HEAP TYPE CHEAT → COMPLEXITIES → MISTAKES → INTERVIEW SCRIPT → PRACTICE MAP

## 1. Decision table

| Question looks like | Pattern |
|---|---|
| k largest / k-th largest | min-heap of size k |
| k smallest / k closest | max-heap of size k |
| k most frequent | Counter + heap |
| median of a stream | two heaps |
| merge k sorted lists/arrays | heap of heads |
| always process the smallest/largest available | heap simulation (stones, ropes) |
| schedule with priority / cooldown | heap (+ queue for cooldown) |
| minimum rooms / resources | heap of end times |
| shortest path with weights | heap (Dijkstra, Graphs topic) |
| next smallest from a sorted matrix | heap of `(value, row, col)` |

## 2. Which heap type?

| You want | Heap | Root holds |
|---|---|---|
| the **k largest** | min-heap of size k | the k-th largest |
| the **k smallest** | max-heap of size k | the k-th smallest |
| the **overall max** repeatedly | max-heap | the max |
| the **overall min** repeatedly | min-heap | the min |

The size-k heap is the **opposite** type of what you collect.

## 3. Complexities

| | Time | Space |
|---|---|---|
| build (heapify) | O(n) | O(1) in place |
| push / pop | O(log n) | |
| top-k with size-k heap | O(n log k) | O(k) |
| merge k lists (N total) | O(N log k) | O(k) |
| median stream | O(log n) add, O(1) median | O(n) |
| heap sort | O(n log n) | O(1) in place |

## 4. Common mistakes

- Using `heapq` as a max-heap without negating.
- Wrong heap type for top-k.
- Tuples with uncomparable second items (add a tie-breaker index).
- Assuming `heap` list is sorted.
- Arbitrary delete in O(n): use lazy deletion (mark deleted, skip on pop).
- Forgetting `heapreplace`/`heappushpop` exist for one-step push+pop.
- Using a heap when one linear pass is enough.

## 5. Interview script

1. "I repeatedly need the smallest/largest, so a heap."
2. "Size-k heap gives O(n log k) time, O(k) space. Sorting would be O(n log n)."
3. Mention alternatives: quickselect O(n) average, bucket sort when values are bounded.
4. Test: k = 1, k = n, duplicates, empty input.

## Quick Revision

- Heap = fast min/max with changes. Python = min-heap, negate for max.
- Top-k: opposite-type heap of size k.
- Two heaps = median. Merge k = heap of heads.
- Schedule/rooms = heap of end times.
- Check: sorting or quickselect may be simpler.
