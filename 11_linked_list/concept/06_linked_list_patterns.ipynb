# 06. Linked List Patterns (Revision Sheet)

**Learning order:** TECHNIQUES → DECISION TABLE → MISTAKES → INTERVIEW SCRIPT → PRACTICE MAP

## 1. The 6 techniques

| Technique | Use for |
|---|---|
| **Dummy node** | any deletion/insertion at head, building a new list |
| **Two pointers, same speed with a gap** | N-th from end |
| **Fast / slow** | middle, cycle, cycle start, palindrome |
| **Reverse** | reverse, palindrome, reorder, k-group |
| **Merge** | merge two/k lists, sort list |
| **Hash map + DLL** | LRU, random pointer copy |

## 2. Decision table

| Question looks like | Do this |
|---|---|
| delete a value / node, head may change | dummy node |
| find middle / is palindrome | fast-slow, reverse second half |
| has cycle / where cycle starts | Floyd |
| n-th node from end | gap of n, dummy |
| reverse all / part / in groups | 3-pointer reversal |
| merge sorted lists | dummy + tail |
| sort a list | merge sort with cut |
| two lists meet? | pointer switching |
| numbers as lists | carry loop |
| O(1) cache | hash map + doubly linked list |

## 3. Common mistakes

- Not saving `next` before changing a pointer.
- `cur.next.next` when `cur.next` is `None`.
- Forgetting to cut a list → cycle.
- Returning the wrong head (`head` vs `prev` vs `dummy.next`).
- Not handling empty list / one node.
- Comparing node values when identity (`is`) is required.
- Forgetting to delete from the hash map during eviction.

## 4. Interview script

1. Draw 3 to 5 nodes with arrows. Write the pointer changes in order.
2. "I'll use a dummy node so the head is not special."
3. State complexity: "O(n) time, O(1) space" for most in-place problems.
4. Test: empty, one node, two nodes, change at head, change at tail, cycle.
5. If asked about arrays vs lists: random access O(1) vs insertion O(1) with a node reference.
