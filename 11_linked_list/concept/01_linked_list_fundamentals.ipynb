# 01. Linked List: Fundamentals

A linked list is a chain of **nodes**. Each node stores a value and a **pointer to the next node**. Nodes can sit anywhere in memory; the pointers connect them.

**Learning order:** REAL LIFE → ARRAY VS LINKED LIST → NODE → TRAVERSAL → INSERT → DELETE → DUMMY NODE → PITFALLS → INTERVIEW THINKING

## 1. Real-life examples

- **Treasure hunt:** each clue tells you where the next clue is. You can't jump to clue 5 directly. You must follow the chain.
- **Train:** coaches are joined one after another. Adding a coach in the middle only needs two couplings changed, nobody moves.
- **Music playlist "next song"**, **browser history** (doubly linked: back and forward).

## 2. Array vs Linked List

| | Array | Linked List |
|---|---|---|
| Access by index | **O(1)** | O(n) (must walk) |
| Insert / delete at the front | O(n) (shift) | **O(1)** |
| Insert / delete in the middle | O(n) | O(1) **if you already have the node**, else O(n) to find it |
| Memory | contiguous, cache friendly | scattered, extra memory for pointers |
| Size | fixed or resized | grows one node at a time |

**Rule:** use a linked list when you do many inserts/deletes at known positions and never need random access.

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def build(vals):                      # list -> linked list
    dummy = ListNode()
    cur = dummy
    for v in vals:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next

def to_list(head):                    # linked list -> list (for printing)
    out = []
    while head:
        out.append(head.val)
        head = head.next
    return out

head = build([1, 2, 3, 4])
print(to_list(head))   # [1, 2, 3, 4]
```

## 3. Traversal

Move with `cur = cur.next` until `cur` is `None`. Never lose the head: keep a separate `cur`, or the list is gone (garbage collected).

```python
def length(head):
    count = 0
    cur = head
    while cur:
        count += 1
        cur = cur.next
    return count

def search(head, target):
    cur = head
    while cur:
        if cur.val == target:
            return True
        cur = cur.next
    return False

h = build([5, 6, 7])
print(length(h), search(h, 6), search(h, 9))   # 3 True False
```

## 4. Insert

- **At head:** new node points to old head, new node becomes head. O(1).
- **At tail:** walk to the last node, link. O(n) (O(1) if you keep a tail pointer).
- **After a given node:** `new.next = node.next; node.next = new`. **Order matters:** set `new.next` first, or you lose the rest of the list.

```python
def insert_head(head, val):
    return ListNode(val, head)

def insert_tail(head, val):
    node = ListNode(val)
    if not head:
        return node
    cur = head
    while cur.next:
        cur = cur.next
    cur.next = node
    return head

def insert_after(node, val):
    node.next = ListNode(val, node.next)      # new.next = node.next first, then node.next = new

h = build([2, 3])
h = insert_head(h, 1)
h = insert_tail(h, 4)
insert_after(h.next, 99)
print(to_list(h))   # [1, 2, 99, 3, 4]
```

## 5. Delete and the dummy node trick

To delete `target`, we need the node **before** it: `prev.next = prev.next.next`.

Deleting the **head** is a special case (no previous node). Avoid it with a **dummy node** placed before the head. Then every node, even the head, has a previous node. Return `dummy.next`.

```python
def delete_value(head, target):
    dummy = ListNode(0, head)
    prev = dummy
    while prev.next:
        if prev.next.val == target:
            prev.next = prev.next.next        # skip the node
        else:
            prev = prev.next
    return dummy.next

print(to_list(delete_value(build([1, 2, 6, 3, 6]), 6)))   # [1, 2, 3]
print(to_list(delete_value(build([6, 6, 1]), 6)))         # [1]
```

## 6. Pitfalls (most linked list bugs)

- `cur.next` when `cur` is `None` → `AttributeError`. Always check `while cur and cur.next`.
- Overwriting a pointer before saving it (`cur.next = prev` before `nxt = cur.next`).
- Losing the head.
- Forgetting empty list and single node cases.
- Creating accidental cycles (infinite loops when printing).

**Debug habit:** draw boxes and arrows, and update pointers in a safe order.

## 7. Interview thinking

**Say:** "I will use a dummy node so head deletion is not a special case." / "I save `next` before changing the pointer."

**Always test:** empty list, one node, two nodes, operation at head, operation at tail.

**Complexity:** traversal O(n), insert/delete at head O(1), search O(n), space O(1) for in-place operations.

## Quick Revision

- Node = value + `next`. No index access, so O(n) search.
- Insert after node: set `new.next` first, then `node.next`.
- Delete: `prev.next = prev.next.next`.
- Dummy node removes head special cases.
- Check `None` before every `.next`. Test empty, 1, 2 nodes.
