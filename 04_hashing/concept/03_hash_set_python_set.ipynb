# 03. Hash Set (Python Set)

A set stores **unique hashable elements**. Python sets use hashing internally and are especially useful for fast membership checks, duplicate detection, and set-based comparisons.

**Learning order:** WHAT → WHY → HOW → OPERATIONS → PATTERNS → PITFALLS → INTERVIEW THINKING

## 1. What is a Hash Set?

A hash set is a collection that stores each element only once.

```python
nums = {1, 2, 2, 3, 3, 3}
print(nums)  # {1, 2, 3}
```

Important properties:

- Every element is unique.
- Elements must be hashable.
- Membership checks are O(1) on average.
- A set does not provide positional indexing.
- A set is mutable; `frozenset` is its immutable counterpart.

Use a set when you need to know **whether something exists**, not when you need to associate it with a value.

````python
nums = {10, 20, 20, 30, 30}
print(nums)
print(20 in nums)  # True
print(99 in nums)  # False

nums.add(40)
print(nums)
````

## 2. Why Do We Need Sets?

Consider checking whether an array contains a duplicate.

A nested-loop solution compares pairs and takes O(n²) time. A set can remember previously seen values and detect a duplicate in O(n) average time.

The key idea is to avoid searching the entire previous part of the array repeatedly.

````python
nums = [4, 2, 7, 2, 9]

seen = set()
has_duplicate = False

for num in nums:
    if num in seen:
        has_duplicate = True
        break
    seen.add(num)

print(has_duplicate)  # True
````

## 3. Creating a Set Correctly

```python
a = {1, 2, 3}       # set with elements
b = set()           # empty set
c = set([1, 2, 2])  # convert iterable to set
```

**Important:** `{}` creates an empty dictionary, not an empty set. Use `set()` for an empty set.

A set can contain different hashable types, but all elements must be hashable. Lists and dictionaries cannot be set elements.

````python
empty_set = set()
empty_dict = {}

values = set([1, 2, 2, 3])
print(type(empty_set))
print(type(empty_dict))
print(values)
````

## 4. Core Set Operations

| Operation | Python | Average time |
|---|---|---:|
| Add | `s.add(x)` | O(1) |
| Remove if present | `s.discard(x)` | O(1) |
| Remove existing item | `s.remove(x)` | O(1) |
| Membership | `x in s` | O(1) |
| Number of elements | `len(s)` | O(1) |
| Clear | `s.clear()` | O(n) |

These are average-case expectations. Worst-case lookup, insertion, and deletion can be O(n) due to collisions.

`remove(x)` raises `KeyError` if x is absent. `discard(x)` does not raise an error.

````python
s = {10, 20, 30}

s.add(40)
s.discard(99)       # no error
s.remove(20)        # 20 exists

print(10 in s)
print(len(s))
print(s)

s.clear()
print(s)
````

## 5. Set Algebra

Sets support mathematical operations:

| Meaning | Operator | Method |
|---|---|---|
| Union | `a | b` | `a.union(b)` |
| Intersection | `a & b` | `a.intersection(b)` |
| Difference | `a - b` | `a.difference(b)` |
| Symmetric difference | `a ^ b` | `a.symmetric_difference(b)` |

- **Union:** elements in either set.
- **Intersection:** elements common to both.
- **Difference:** elements in the left set but not the right.
- **Symmetric difference:** elements in either set, but not both.

These operations are useful for comparing collections.

````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)  # {1, 2, 3, 4, 5, 6}
print(a & b)  # {3, 4}
print(a - b)  # {1, 2}
print(a ^ b)  # {1, 2, 5, 6}
````

## 6. Pattern 1 — Duplicate Detection

**Problem signal:** determine whether any value appears more than once.

Thinking:

1. Start with an empty set.
2. Scan each element.
3. If it is already in the set, a duplicate exists.
4. Otherwise, add it to the set.

Time: O(n) average. Extra space: O(n) in the worst case.

````python
def contains_duplicate(nums):
    seen = set()

    for num in nums:
        if num in seen:
            return True
        seen.add(num)

    return False


print(contains_duplicate([1, 2, 3, 1]))  # True
print(contains_duplicate([1, 2, 3]))     # False
````

## 7. Pattern 2 — Remove Duplicates

**Problem signal:** keep only distinct values.

Converting a list to a set removes duplicates. But a set does not preserve the original positional order as a contract for algorithmic output.

If order must be preserved, use a set to track seen values and build a result list.

````python
nums = [3, 1, 3, 2, 1, 4]

# Unique values; order is not the goal
unique_values = set(nums)
print(unique_values)

# Preserve first-occurrence order
seen = set()
result = []

for num in nums:
    if num not in seen:
        seen.add(num)
        result.append(num)

print(result)  # [3, 1, 2, 4]
````

## 8. Pattern 3 — Fast Membership Lookup

**Problem signal:** for many queries, check whether values exist in another collection.

Convert the lookup collection to a set once. Then each membership check takes O(1) average time.

Do not rebuild the set inside the query loop; that would waste time.

````python
available = [10, 20, 30, 40, 50]
queries = [20, 60, 40]

available_set = set(available)

for value in queries:
    print(value, value in available_set)
````

## 9. Pattern 4 — Intersection of Two Arrays

**Problem signal:** find values common to two arrays.

Convert one array to a set, then check values from the other array. If the answer must contain unique common values, store matches in a result set.

Be careful: some problems ask for unique intersection; others ask for multiplicity. A set alone loses frequency information.

````python
nums1 = [1, 2, 2, 1]
nums2 = [2, 2]

set1 = set(nums1)
answer = set()

for num in nums2:
    if num in set1:
        answer.add(num)

print(list(answer))  # [2] (order may vary)
````

## 10. Pattern 5 — Longest Consecutive Sequence

This is an important interview pattern.

**Problem signal:** find the longest run of consecutive values, regardless of their original positions.

Key idea:

1. Put all values into a set for fast membership.
2. A value is the start of a sequence only if `num - 1` is absent.
3. From each start, count forward while the next value exists.
4. Track the maximum length.

Why check for a start? It prevents scanning the same sequence from every element.

Average time: O(n), extra space: O(n).

````python
def longest_consecutive(nums):
    values = set(nums)
    longest = 0

    for num in values:
        if num - 1 not in values:
            length = 1
            current = num

            while current + 1 in values:
                current += 1
                length += 1

            longest = max(longest, length)

    return longest


print(longest_consecutive([100, 4, 200, 1, 3, 2]))  # 4
````

## 11. Set vs Dictionary vs List

| Feature | List | Set | Dictionary |
|---|---|---|---|
| Stores | Ordered elements | Unique elements | Key-value pairs |
| Duplicates | Allowed | Not allowed | Keys are unique |
| Membership | O(n) | O(1) average | O(1) average for keys |
| Index access | Yes | No | By key |
| Main use | Sequence | Existence / uniqueness | Mapping / counting |

Choose the simplest structure that stores exactly the information your algorithm needs.

## 12. Common Mistakes

1. Using `{}` when you need an empty set; it creates a dictionary.
2. Trying to access `s[0]`; sets do not support indexing.
3. Trying to add a list or dictionary to a set; these are unhashable.
4. Assuming set iteration order is sorted or meaningful.
5. Using a set when duplicates or frequencies matter.
6. Calling `remove()` when an item may be absent; use `discard()` or check membership.
7. Converting to a set repeatedly inside a loop.
8. Claiming membership is guaranteed O(1); it is O(1) on average.

## 13. Interview Thinking Checklist

When you see a problem, ask:

1. Do I only need to know whether an item exists?
2. Do I need unique values or must I preserve duplicates?
3. Do I need counts or positions? If yes, would a dictionary be better?
4. Can I convert an input to a set once to speed up repeated lookups?
5. Does the output require a particular order?
6. Can I identify sequence starts or avoid repeated work?
7. What is the average time and extra space complexity?

**Typical optimization:** replace repeated list membership checks inside a loop with a prebuilt set. This can reduce O(n × m) average work to O(n + m).

## 15. Complexity and Revision

- Set membership, add, and discard: **O(1) average**, **O(n) worst case**.
- Building a set from n values: **O(n) average**.
- Iterating over a set: **O(n)**.
- Set storage: **O(k)** for k distinct elements.
- Set operations such as union and intersection depend on the sizes of the input sets.
