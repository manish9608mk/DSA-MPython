# 07. Duplicate Detection (Hashing)

Duplicate detection can mean different things: check if any duplicate exists, return unique repeated values, return every repeated occurrence, count duplicate pairs, or find duplicate positions. First understand the exact output required.

## 1. Brute Force

Compare every pair using nested loops. Worst-case time O(n²), extra space O(1). This is a useful baseline but often too slow for large inputs.

````python
def contains_duplicate_brute(nums):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] == nums[j]:
                return True
    return False

print(contains_duplicate_brute([1, 2, 3, 1]))  # True
print(contains_duplicate_brute([1, 2, 3]))     # False
````

## 2. Pattern — Does Any Duplicate Exist?

Use a set to remember previously seen values. For each number, check membership before adding it.

Average time O(n), extra space O(k), where k is the number of distinct values. This is the standard approach for Contains Duplicate.

````python
def contains_duplicate(nums):
    seen = set()
    for num in nums:
        if num in seen:
            return True
        seen.add(num)
    return False

print(contains_duplicate([1, 2, 3, 1]))  # True
print(contains_duplicate([1, 2, 3, 4]))  # False
````

## 3. Pattern — Return Unique Duplicate Values

Use `seen` for all values already encountered and `duplicates` for values encountered more than once. Each duplicated value appears once in the result.

Average time O(n), extra space O(n). Sort the result if the problem requires sorted output.

````python
def find_duplicates(nums):
    seen = set()
    duplicates = set()
    for num in nums:
        if num in seen:
            duplicates.add(num)
        else:
            seen.add(num)
    return list(duplicates)

print(sorted(find_duplicates([1, 2, 3, 2, 1, 1])))  # [1, 2]
````

## 4. Pattern — Every Repeated Occurrence After the First

For `[1, 2, 1, 1]`, the repeated occurrences are `[1, 1]`. Append the value each time it is already in `seen`. Do not use a duplicate set because that would remove repeated occurrences from the answer.

````python
def repeated_occurrences(nums):
    seen = set()
    result = []
    for num in nums:
        if num in seen:
            result.append(num)
        else:
            seen.add(num)
    return result

print(repeated_occurrences([1, 2, 1, 1]))  # [1, 1]
````

## 5. Pattern — Count Equal Index Pairs

If a value has appeared `count` times before the current position, the current value forms `count` new pairs. Add the previous frequency, then increment it.

A value with frequency f contributes f * (f - 1) // 2 pairs. Average time O(n), extra space O(k).

````python
def count_equal_pairs(nums):
    freq = {}
    pairs = 0
    for num in nums:
        pairs += freq.get(num, 0)
        freq[num] = freq.get(num, 0) + 1
    return pairs

print(count_equal_pairs([1, 2, 3, 1, 1, 3]))  # 4
````

## 6. Pattern — First Value to Repeat

This returns the first value whose second occurrence is encountered during a left-to-right scan. Be careful: this can differ from the value whose first original occurrence is earliest. Read the problem definition precisely.

````python
def first_value_to_repeat(nums):
    seen = set()
    for num in nums:
        if num in seen:
            return num
        seen.add(num)
    return None

print(first_value_to_repeat([10, 5, 3, 4, 3, 5, 6]))  # 3
````

## 7. Pattern — Find Duplicate Indices

If positions are required, store `value -> previous index` or `value -> list of indices`. A set cannot tell you where the previous occurrence was.

````python
def duplicate_index_pairs(nums):
    last_index = {}
    result = []
    for i, num in enumerate(nums):
        if num in last_index:
            result.append((last_index[num], i))
        last_index[num] = i
    return result

print(duplicate_index_pairs([4, 2, 4, 4]))  # [(0, 2), (2, 3)]
````

## 8. Sorting Alternative

Sorting makes equal values adjacent. Scan neighboring elements to detect duplicates.

Time O(n log n). Extra space depends on sorting and whether a copy is created. Sorting changes original order, so preserve indices if needed.

````python
def contains_duplicate_sorting(nums):
    arr = sorted(nums)
    for i in range(1, len(arr)):
        if arr[i] == arr[i - 1]:
            return True
    return False

print(contains_duplicate_sorting([3, 1, 4, 2, 1]))  # True
````

## 9. Edge Cases

Test: empty list, one element, all unique, all equal, negative numbers, zero, duplicates at the beginning/end, and values repeated three or more times. Set elements and dictionary keys must be hashable.

## 10. Complexity Comparison

| Approach | Time | Extra space |
|---|---:|---:|
| Nested loops | O(n²) | O(1) |
| Set detection | O(n) average | O(k), up to O(n) |
| Frequency map | O(n) average | O(k) |
| Sorting | O(n log n) | Depends on sorting/copy |

Hash lookup is O(1) on average, not guaranteed; worst-case operations can be O(n).

## 11. Common Mistakes

1. Using a set when occurrence counts matter.
2. Confusing unique duplicate values with every repeated occurrence.
3. Confusing first original occurrence with first repeat encountered.
4. Forgetting sorting changes order.
5. Not testing empty/single-element inputs.
6. Claiming worst-case O(1) for hash lookup.

## 12. Interview Thinking Checklist

1. Does the problem ask for existence, values, counts, or indices?
2. Should each duplicate appear once or on every repeat?
3. Is scan order important?
4. Is a set enough, or do I need a frequency/index map?
5. Can I solve it in one pass?
6. Would sorting be better under memory constraints?

Recognition words: duplicate, repeated, unique, frequency, occurrence, first repeating.

**Challenge:** Return the smallest value that appears at least twice. Return -1 if none exists. Consider negative values.

````python
def smallest_repeated_value(nums):
    seen = set()
    answer = None
    for num in nums:
        if num in seen:
            if answer is None or num < answer:
                answer = num
        else:
            seen.add(num)
    return -1 if answer is None else answer

print(smallest_repeated_value([5, 1, 3, 1, 5, 2]))  # 1
print(smallest_repeated_value([4, 2, 7]))            # -1
````
