## 1. What is Hashing?

Hashing is a technique that converts a value (key) into a number using a **hash function**. This number helps a data structure decide where to store or find the value.

Think of a library where every book has a code. Instead of checking every shelf, you use the code to go directly to the expected shelf.

In DSA, hashing is mainly used for fast **searching, insertion, deletion, and frequency counting**.

## 2. Why do we need Hashing?

Suppose we have a list and want to check whether a number exists.

- Linear search may check every element: **O(n)** time.
- A hash set usually checks membership in **O(1) average time**.

Hashing helps us avoid repeatedly scanning an entire collection.

**Important:** O(1) is the average-case expectation, not a guarantee for every operation.

```python
# Searching with a list: O(n) worst-case lookup
nums = [10, 20, 30, 40, 50]
target = 40

print(target in nums)  # True
```

```python
# Searching with a set: O(1) average-case lookup
nums_set = {10, 20, 30, 40, 50}
target = 40

print(target in nums_set)  # True
```

## 3. How does Hashing work?

A hash table uses a hash function to calculate a hash value for a key. The hash value helps determine where the key should be stored.

Conceptual flow:

1. Give the hash table a key.
2. The hash function calculates a hash value.
3. The table maps that value to a storage position (often called an index or bucket).
4. The table stores or searches for the key at the relevant position.

Real hash tables use additional rules to handle collisions and maintain performance. We will study those details separately.

```python
# Python provides hash() for hashable objects.
print(hash(10))
print(hash("python"))
print(hash((1, 2, 3)))
```

**Important:** Do not expect the printed hash value of a string to remain the same across different Python processes. Python randomizes hashes for some types, including strings, to improve security.

## 4. What is a Collision?

A **collision** happens when two different keys produce the same hash-table position.

A collision does not mean the hash table fails. Hash tables have collision-resolution techniques so multiple keys can still be stored correctly.

For now, remember:

- Different keys can have the same hash value or map to the same bucket.
- A good hash table handles collisions internally.
- Too many collisions can reduce performance.

We will study collision handling in an advanced concept later.

## 5. Hashing in Python: dict and set

Python gives us two important hash-based data structures:

- **dict:** stores key-value pairs.
- **set:** stores unique elements.

Both use hashing internally and provide average O(1) lookup, insertion, and deletion.

```python
# Dictionary: key -> value
student_marks = {
    "Aman": 85,
    "Riya": 92,
    "Sam": 78
}

print(student_marks["Riya"])  # 92
print("Aman" in student_marks)  # True
```

```python
# Set: unique values
nums = {10, 20, 20, 30, 30, 30}

print(nums)       # {10, 20, 30} (display order may vary)
print(20 in nums) # True
```

## 6. What is Hashable in Python?

A value is **hashable** if it has a hash value that remains stable during its lifetime and can be compared for equality.

Dictionary keys and set elements must be hashable.

Common examples:

- Usually hashable: `int`, `float`, `str`, `bool`, and tuples containing only hashable values.
- Unhashable: `list`, `dict`, and `set`.

Why are lists unhashable? A list can change after creation. If a mutable list were used as a key, changing it could make its storage location inconsistent.

```python
# Hashable examples
print(hash(100))
print(hash("hello"))
print(hash((1, "a")))

# Uncomment to see the error: lists are unhashable
# print(hash([1, 2, 3]))
```

## 7. Time Complexity of Hash-Based Operations

| Operation | Average | Worst case |
|---|---:|---:|
| Search / lookup | O(1) | O(n) |
| Insert | O(1) | O(n) |
| Delete | O(1) | O(n) |

Worst-case behavior can occur when collisions are excessive. Python's implementation includes strategies to keep normal operations efficient.

**Interview rule:** Say average O(1), not guaranteed O(1), when discussing dictionary and set operations.

## 8. When should you think about Hashing?

Recognize these clues in a problem:

- Find whether an element exists quickly.
- Detect duplicates.
- Count the frequency of elements or characters.
- Find a pair using a required complement.
- Group values by a common property.
- Remember information seen earlier while scanning an array or string.

Ask yourself: **Am I repeatedly searching or counting? Can a dictionary or set remember the information for me?**
