# 02. Hash Map (Python Dictionary)

A hash map stores data as **key → value** pairs. In Python, the built-in `dict` is a hash map implementation.

This notebook builds from fundamentals to interview-level usage: operations, complexity, frequency maps, index maps, safe access, iteration, pitfalls, and reusable patterns.

**Learning order:** WHAT → WHY → HOW → OPERATIONS → PATTERNS → PITFALLS → INTERVIEW THINKING

## 1. What is a Hash Map?

A hash map stores information using a key to find its associated value.

Example: a student name is the key and the student's marks are the value.

```python
marks = {"Aman": 85, "Riya": 92}
```

- `"Aman"` is a key.
- `85` is its value.
- Keys are unique. Assigning a value to an existing key updates that entry.
- Values can be repeated.

**When useful:** when you need to find, update, or remember information using a key.

````python
# Creating a dictionary
marks = {"Aman": 85, "Riya": 92, "Sam": 78}

print(marks["Riya"])       # 92
print(len(marks))           # 3
print("Aman" in marks)     # True (checks keys)
print("Aman" in marks.values())  # False (checks values)
````

## 2. Core Dictionary Operations

| Operation | Python | Average time |
|---|---|---:|
| Create | `d = {}` | — |
| Insert / update | `d[key] = value` | O(1) |
| Read | `d[key]` | O(1) |
| Safe read | `d.get(key, default)` | O(1) |
| Delete | `del d[key]` | O(1) |
| Remove and return | `d.pop(key)` | O(1) |
| Check key | `key in d` | O(1) |
| Number of entries | `len(d)` | O(1) |

These are average-case complexities. A single operation can take O(n) in the worst case due to collisions or resizing. Iterating over all entries takes O(n).

````python
# Insert, update, read, check, and delete
profile = {}
profile["name"] = "Asha"       # insert
profile["age"] = 21
profile["age"] = 22             # update

print(profile["name"])
print(profile.get("city", "Not provided"))
print("age" in profile)

removed_age = profile.pop("age")
print(removed_age)
print(profile)
````

## 3. Access: [] vs get()

`d[key]` raises a `KeyError` if the key does not exist.

`d.get(key, default)` returns the default instead of raising an error.

Use `[]` when the key is guaranteed to exist. Use `get()` when it may be missing.

**Interview habit:** choose access based on the problem's guarantee; do not hide a missing-key bug accidentally.

````python
scores = {"Aman": 90}

print(scores["Aman"])                 # 90
print(scores.get("Riya"))              # None
print(scores.get("Riya", 0))           # 0

# Uncomment to observe KeyError:
# print(scores["Riya"])
````

## 4. Iterating Through a Dictionary

- `d.keys()` gives keys.
- `d.values()` gives values.
- `d.items()` gives `(key, value)` pairs.

Python dictionaries preserve insertion order. In algorithm problems, rely on this only when order is relevant and the problem allows it.

````python
prices = {"apple": 4, "banana": 2, "orange": 3}

for fruit in prices:
    print(fruit)                    # keys

for price in prices.values():
    print(price)                    # values

for fruit, price in prices.items():
    print(fruit, price)             # key-value pairs
````

## 5. Pattern 1 — Frequency Map

**Problem signal:** count how many times each number or character appears.

Thinking:
1. Each distinct item becomes a dictionary key.
2. Its value stores the count.
3. For each item, increase its count by one.

This avoids repeatedly scanning the input to count each item.

````python
nums = [4, 2, 4, 1, 2, 4]

freq = {}
for num in nums:
    freq[num] = freq.get(num, 0) + 1

print(freq)  # {4: 3, 2: 2, 1: 1}
````

### Frequency map with `collections.Counter`

`Counter` is a convenient frequency-counting tool. Learn the manual dictionary pattern first because it explains the underlying idea and works in any interview setting.

Use `Counter` when concise counting is appropriate.

````python
from collections import Counter

nums = [4, 2, 4, 1, 2, 4]
frequency = Counter(nums)

print(frequency)       # Counter({4: 3, 2: 2, 1: 1})
print(frequency[4])    # 3
````

## 6. Pattern 2 — Last Seen Index

**Problem signal:** you need to remember where an item appeared most recently.

Store `value → index` while scanning from left to right. This is useful in duplicate-distance problems and substring problems.

Important: decide whether to update the stored index before or after using the previous index. The order can change the result.

````python
words = ["a", "b", "a", "c"]
last_seen = {}

for i, word in enumerate(words):
    if word in last_seen:
        print(word, "was previously at", last_seen[word])
    last_seen[word] = i

# Output includes: a was previously at 0
````

## 7. Pattern 3 — Complement Lookup

**Problem signal:** find two values that add up to a target.

For each number `x`, the required partner is `target - x`.

Instead of checking every pair, store previously seen values in a dictionary or set. For the classic Two Sum problem, a dictionary can store `value → index`.

The key interview question is: **Should I look for the complement before or after inserting the current value?** Usually check first so the same element is not used twice.

````python
nums = [2, 7, 11, 15]
target = 9
seen = {}  # value -> index
answer = None

for i, num in enumerate(nums):
    complement = target - num
    if complement in seen:
        answer = [seen[complement], i]
        break
    seen[num] = i

print(answer)  # [0, 1]
````

## 8. Pattern 4 — Grouping by a Key

**Problem signal:** put items into groups based on a shared property.

A dictionary can map a group identifier to a list of items. For example, group words by their sorted-character signature to group anagrams.

`setdefault()` can create a list the first time a key appears, but `defaultdict(list)` is often cleaner.

````python
words = ["eat", "tea", "tan", "ate", "nat", "bat"]
groups = {}

for word in words:
    key = "".join(sorted(word))
    groups.setdefault(key, []).append(word)

print(list(groups.values()))
````

## 9. `defaultdict` — Automatic Default Values

`defaultdict` creates a default value when a missing key is accessed. Common forms:

- `defaultdict(int)` → default `0`, useful for counts.
- `defaultdict(list)` → default `[]`, useful for grouping.
- `defaultdict(set)` → default `set()`, useful for unique grouped values.

Know ordinary dictionaries and `get()` first. `defaultdict` is a convenience, not a different algorithmic idea.

````python
from collections import defaultdict

freq = defaultdict(int)
for ch in "banana":
    freq[ch] += 1
print(dict(freq))

groups = defaultdict(list)
for word in ["eat", "tea", "tan"]:
    key = "".join(sorted(word))
    groups[key].append(word)
print(dict(groups))
````

## 10. Hashable Keys and Common Pitfalls

Dictionary keys must be hashable. Common hashable keys include integers, strings, and tuples whose elements are all hashable. Lists, dictionaries, and sets cannot be dictionary keys.

Common mistakes:

1. Writing `if value in d.values()` when you intended to check keys. This scans values and is O(n).
2. Using `d[key]` when the key may be absent, causing `KeyError`.
3. Updating a frequency with `d[key] += 1` before initializing the key.
4. Storing the wrong information: value, index, count, or frequency are different choices.
5. Forgetting that dictionary keys are unique; assigning the same key again overwrites its value.

````python
# Safe frequency update
nums = [1, 1, 2]
freq = {}
for x in nums:
    freq[x] = freq.get(x, 0) + 1
print(freq)

# Tuple key is valid when its elements are hashable
position_map = {(2, 3): "cell"}
print(position_map[(2, 3)])
````

## 11. How to Choose What the Dictionary Stores

Before coding, state the relationship clearly:

| Need | Dictionary design |
|---|---|
| Count occurrences | `value → count` |
| Find an earlier position | `value → index` |
| Store associated information | `key → data` |
| Group similar items | `group_key → list of items` |
| Track whether seen | use a set, or `value → True` |

The most important design decision is choosing the correct key and value.

## 12. Interview Thinking Checklist

When a problem involves repeated searching, counting, or remembering previous elements, ask:

1. What information must I remember?
2. What should be the dictionary key?
3. What should its value represent: count, index, boolean, or collection?
4. When should I insert or update the entry?
5. Can the current item match itself accidentally?
6. Is a set enough, or do I need a dictionary?
7. What is the average time and extra space complexity?

A common optimization is replacing an O(n) search inside a loop with an average O(1) dictionary lookup, reducing O(n²) to O(n).

## 14. Complexity Summary and Revision

- Dictionary lookup, insertion, and deletion: **O(1) average**, **O(n) worst case**.
- Iterating through a dictionary: **O(n)**.
- Building a frequency map for n items: **O(n) average time**, up to **O(k)** space for k distinct keys.
- A dictionary stores key-value relationships; a set stores unique values.
- Choose the key-value relationship before writing code.
