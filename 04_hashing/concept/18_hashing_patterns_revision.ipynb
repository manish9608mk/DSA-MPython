# 18. Hashing Patterns (Revision Sheet)

One page to recognize which hashing pattern a question needs.

**Learning order:** DECISION TABLE → KEY DESIGN → COMPLEXITY AND LIMITS → MISTAKES → INTERVIEW SCRIPT → PRACTICE MAP

## 1. Decision table

| Question looks like | Pattern | Notebook |
|---|---|---|
| how many times does X appear | frequency map (`Counter`) | 04 |
| have I seen this before / duplicates | set | 07 |
| find a partner (`target - x`) | complement lookup | 06 |
| group items that "look the same" | signature as dict key (sorted string, count tuple) | 08, 09 |
| subarray with a given sum (negatives allowed) | prefix sum + map | 10, 11 |
| longest substring with at most k distinct | window + counts | 12 |
| need a custom key (list, object) | tuple / `frozenset` / `__hash__` | 13 |
| substring / duplicate substring | rolling hash | 15 |
| one-to-one mapping | two maps | 17 |
| consecutive sequences | set + start check | 17 |
| pairs from several arrays | meet in the middle | 17 |
| nearest / ordered | **not hashing**: bisect / heap / tree | 16 |

## 2. Designing a key

- Anagrams: `tuple(sorted(word))` or a 26-count tuple (O(L) instead of O(L log L)).
- Shifted strings: normalize the first character difference.
- Coordinates / pairs: `(r, c)` tuple.
- Sets of values: `frozenset`.
- **Never** use a `list` or `dict` as a key (unhashable). Never mutate an object used as a key.

## 3. Complexity and limits

| | Average | Worst |
|---|---|---|
| insert / lookup / delete | O(1) | O(n) (collisions, rare in Python) |
| build from n items | O(n) | |
| memory | O(n) | |

Python's dict preserves **insertion order** (3.7+). Sets do not. Hash of a `str` is randomized per run (so set/dict iteration order of a `set` of strings can vary between runs).

## 4. Common mistakes

- `defaultdict` access creates the key silently (check with `in`, not `d[k]`).
- Mutating a dict while iterating over it.
- Forgetting `{0: 1}` in prefix-sum maps.
- Using `dict.get(k)` and treating `0` as "missing".
- Unhashable keys (`list`).
- Using hashing where ordering is needed.
- Hash collisions in custom rolling hashes (always verify).
- Counting an element with itself (complement of `x` is `x`).

## 5. Interview script

1. "Brute force is O(n²) because for each element I scan the rest."
2. "I store what I've seen in a hash map so each check is O(1)."
3. "The key is ... and the value is ... Time O(n), space O(n)."
4. Mention the trade-off: more memory for less time. Ask if sorting (O(1) space) is acceptable.
5. Test: empty, one element, duplicates, negative numbers.

## Quick Revision

- Hash = O(1) exact lookups, trading memory for time.
- Pick a good key (sorted tuple, count tuple, coordinates).
- Prefix map starts with `{0: 1}`. Complement lookup before inserting.
- Not for order/closeness → ordered structures.
