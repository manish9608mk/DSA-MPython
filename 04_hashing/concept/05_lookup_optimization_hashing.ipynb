# 05. Lookup Optimization (Hashing)

Lookup optimization means storing useful information in a set or dictionary so we do not repeatedly search the same data.

**Core idea:** Instead of searching again and again, remember what you have already seen.

## 1. What is Lookup?

Lookup means checking whether information exists or finding information associated with a key.

- Use a **set** when you only need existence.
- Use a **dictionary** when you need associated information such as an index or count.
- Use a **list** when order and positional access matter.

## 2. Linear Search vs Hash Lookup

List membership takes O(n) worst-case time. Set membership takes O(1) average time after building the set.

Building a set costs O(n) average time and O(n) space. Therefore, for one lookup the conversion may not help; for many lookups it can be valuable.

Always include preprocessing cost in your analysis.

````python
nums = [10, 20, 30, 40, 50]
target = 40

print(target in nums)  # list lookup: O(n)
lookup = set(nums)     # build once: O(n) average
print(target in lookup)  # set lookup: O(1) average
````

## 3. Pattern — Precompute Once, Query Many Times

If n values are stored and q queries must be answered:

- Repeated list search: O(n * q).
- Build a set once and query: O(n + q) average.
- Extra space: O(n).

Do not rebuild the set inside the query loop.

````python
def answer_queries(nums, queries):
    lookup = set(nums)
    return [query in lookup for query in queries]

print(answer_queries([5, 10, 15, 20], [10, 12, 20, 30]))
# [True, False, True, False]
````

## 4. Pattern — Replace Nested Loops with a Set

To check whether two arrays share a value, brute force compares every pair: O(n * m) time.

Optimized approach: put one array in a set, then scan the other. Average time becomes O(n + m), with O(n) or O(m) extra space depending on which array is stored.

````python
def have_common_element(a, b):
    lookup = set(a)
    for value in b:
        if value in lookup:
            return True
    return False

print(have_common_element([1, 3, 5], [2, 4, 5]))  # True
print(have_common_element([1, 3], [2, 4]))        # False
````

## 5. Pattern — Value to Index Mapping

Sometimes existence is not enough; we need the position. Store `value -> index` in a dictionary.

This pattern appears in Two Sum and problems involving previous positions.

````python
nums = [8, 3, 6, 2]
index_map = {}
for i, value in enumerate(nums):
    index_map[value] = i

print(index_map)
print(index_map[6])  # 2
````

## 6. Pattern — Complement Lookup (Two Sum)

For current value `x`, the partner needed to reach target is `target - x`.

1. Create `seen = {}` storing value to index.
2. For each value, calculate its complement.
3. Check whether the complement is already stored.
4. If found, return both indices.
5. Otherwise store the current value and index.

Check before inserting the current value so one element cannot be used twice.

Average time O(n); extra space O(n).

````python
def two_sum(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    return []

print(two_sum([2, 7, 11, 15], 9))  # [0, 1]
print(two_sum([3, 3], 6))          # [0, 1]
````

## 7. Pattern — Lookup with Counts

A set loses duplicate counts. If copies matter, use a dictionary storing `value -> remaining count`.

This is useful for matching elements and checking whether one collection can be formed from another.

````python
available = [2, 2, 3, 5]
required = [2, 3, 2]
count = {}
for x in available:
    count[x] = count.get(x, 0) + 1

possible = True
for x in required:
    if count.get(x, 0) == 0:
        possible = False
        break
    count[x] -= 1

print(possible)  # True
````

## 8. Store the Best Information Seen So Far

A dictionary can remember:

- `value -> latest index`
- `value -> earliest index`
- `prefix sum -> number of times seen`
- `character -> last seen index`

The correct stored information depends on the problem. Longest-distance problems may need the earliest index; shortest-distance problems may need the latest index.

Ask: what information from the past is sufficient to make the current decision?

````python
text = "abca"
last_seen = {}
for i, ch in enumerate(text):
    if ch in last_seen:
        print(ch, "was previously at", last_seen[ch])
    last_seen[ch] = i
print(last_seen)
````

## 9. Choose Set or Dictionary

| Requirement | Structure | Stored information |
|---|---|---|
| Check existence | Set | Value |
| Detect duplicates | Set | Seen values |
| Count occurrences | Dictionary | Value -> count |
| Find previous index | Dictionary | Value -> index |
| Complement and index | Dictionary | Value -> index |
| Match limited copies | Dictionary | Value -> remaining count |

Choose the simplest structure that stores exactly what the algorithm needs.

## 10. Complexity and Time-Space Trade-off

Let n be the collection size and q the number of queries.

| Approach | Time | Extra space |
|---|---:|---:|
| One list lookup | O(n) | O(1) |
| Build a set | O(n) average | O(n) |
| q list lookups | O(n * q) | O(1) |
| Set build + q lookups | O(n + q) average | O(n) |
| Two Sum with dictionary | O(n) average | O(n) |

Hashing trades memory for speed. Set and dictionary operations are O(1) on average, not guaranteed; worst case can be O(n).

## 11. Common Mistakes

1. Rebuilding a set inside the query loop.
2. Using a set when duplicate counts matter.
3. Using a list for many repeated membership checks.
4. Forgetting preprocessing time.
5. Storing an index when a count is required, or vice versa.
6. In Two Sum, inserting before checking the complement.
7. Claiming guaranteed O(1) instead of average O(1).
8. Ignoring whether output needs indices, values, unique values, or all pairs.

**Challenge:** Return True if any two different elements sum to target; do not return their indices.

````python
def has_pair_with_sum(nums, target):
    seen = set()
    for num in nums:
        complement = target - num
        if complement in seen:
            return True
        seen.add(num)
    return False

print(has_pair_with_sum([1, 4, 6, 8], 10))  # True
print(has_pair_with_sum([1, 2, 3], 10))     # False
````
