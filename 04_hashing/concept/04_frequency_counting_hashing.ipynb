# 04. Frequency Counting (Hashing)

Frequency counting records how many times each distinct value appears. It is used in arrays, strings, anagrams, duplicate problems, grouping, and sliding windows.

**Learning order:** WHAT → WHY → MANUAL MAP → COUNTER → PATTERNS → COMPLEXITY → INTERVIEW THINKING

## 1. What is Frequency Counting?

Frequency is the number of times an element appears.

For `[4, 2, 4, 1, 2, 4]`, the frequency map is `{4: 3, 2: 2, 1: 1}`.

A dictionary stores `element → count`. If there are n elements and k distinct values, building the map takes O(n) average time and O(k) extra space.

````python
nums = [4, 2, 4, 1, 2, 4]
frequency = {}
for num in nums:
    frequency[num] = frequency.get(num, 0) + 1
print(frequency)  # {4: 3, 2: 2, 1: 1}
````

## 2. Why is Frequency Counting Useful?

Counting each distinct value by scanning the whole input again can take O(n²). A single pass with a dictionary counts all values in O(n) average time.

**Recognition clues:** count, occurrences, duplicates, anagrams, most frequent, least frequent, same character composition.

## 3. Understand the Manual Frequency Map

```python
frequency = {}
for x in items:
    frequency[x] = frequency.get(x, 0) + 1
```

For `[2, 1, 2]`: first `2` gets count 1; `1` gets count 1; second `2` increases its count to 2. Final map: `{2: 2, 1: 1}`.

`get(x, 0)` returns zero when the key is missing.

````python
items = [2, 1, 2]
frequency = {}
for x in items:
    frequency[x] = frequency.get(x, 0) + 1
    print("Current:", x, frequency)
print("Final:", frequency)
````

## 4. Frequency Counting with Counter

`collections.Counter` is a convenient frequency-counting tool. Learn the manual pattern first so you understand the algorithm.

- `Counter(items)` builds counts.
- `counter[x]` returns the count (zero if missing).
- `counter.most_common(k)` returns the k most common items.

Use it when concise Python code is appropriate.

````python
from collections import Counter

nums = [4, 2, 4, 1, 2, 4]
frequency = Counter(nums)
print(frequency)
print(frequency[4])             # 3
print(frequency[99])            # 0
print(frequency.most_common(2))
````

## 5. Frequency Counting in Strings

Strings are sequences of characters, so the same pattern counts each character. This is useful for Valid Anagram, First Unique Character, character replacement, and grouping strings.

For a string of length n, counting takes O(n) average time. Space depends on the number of distinct characters.

````python
text = "banana"
frequency = {}
for ch in text:
    frequency[ch] = frequency.get(ch, 0) + 1
print(frequency)  # {'b': 1, 'a': 3, 'n': 2}
````

## 6. Pattern — Find the Most Frequent Element

Build a frequency map, then find the key with the largest count. If values tie, follow the problem's stated tie-breaking rule.

```python
most_frequent = max(frequency, key=frequency.get)
```

Building the map is O(n) average; finding the maximum among k keys is O(k).

````python
nums = [1, 3, 2, 3, 4, 3, 2]
freq = {}
for x in nums:
    freq[x] = freq.get(x, 0) + 1
most_frequent = max(freq, key=freq.get)
print(most_frequent)  # 3
````

## 7. Pattern — First Unique Character

**Signal:** return the first character whose frequency is exactly one.

Use two passes: first count all characters; then scan the original string in order and return the first index whose count is one. The second pass preserves original order.

````python
def first_unique_character(text):
    freq = {}
    for ch in text:
        freq[ch] = freq.get(ch, 0) + 1
    for i, ch in enumerate(text):
        if freq[ch] == 1:
            return i
    return -1

print(first_unique_character("leetcode"))  # 0
print(first_unique_character("aabb"))      # -1
````

## 8. Pattern — Compare Frequency Maps (Anagrams)

Two strings are anagrams when they contain exactly the same characters with the same frequencies.

1. If lengths differ, return False.
2. Count characters in the first string.
3. Decrease counts using the second string.
4. Reject a missing character or a negative count.

Time: O(n) average. Space: O(k), where k is the number of distinct characters.

````python
def is_anagram(s, t):
    if len(s) != len(t):
        return False
    freq = {}
    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1
    for ch in t:
        if ch not in freq:
            return False
        freq[ch] -= 1
        if freq[ch] < 0:
            return False
    return True

print(is_anagram("listen", "silent"))  # True
print(is_anagram("rat", "car"))        # False
````

## 9. Pattern — Same Frequencies in Two Arrays

Sometimes order does not matter, but multiplicity does. `[1, 2, 2, 3]` and `[2, 1, 3, 2]` have the same frequency distribution.

`Counter(a) == Counter(b)` compares every value's count.

````python
from collections import Counter

def same_frequencies(a, b):
    return Counter(a) == Counter(b)

print(same_frequencies([1, 2, 2, 3], [2, 1, 3, 2]))  # True
print(same_frequencies([1, 2, 2], [1, 2, 3]))        # False
````

## 10. Pattern — Top K Frequent Elements

**Signal:** return k values with the highest frequencies.

First build a frequency map. Then choose a retrieval strategy:

- Sorting: O(n + u log u), where u is the number of distinct values.
- Heap: commonly O(n + u log k).
- Bucket approach: O(n) time using frequency-indexed buckets.

Start with sorting; later we will study heap and bucket solutions. Respect tie rules if the problem defines them.

````python
from collections import Counter

def top_k_frequent_sorting(nums, k):
    freq = Counter(nums)
    ordered = sorted(freq, key=freq.get, reverse=True)
    return ordered[:k]

print(top_k_frequent_sorting([1, 1, 1, 2, 2, 3], 2))  # [1, 2]
````

## 11. Fixed Alphabet Optimization

If input is guaranteed to contain only lowercase English letters (`a` to `z`), use a list of length 26 instead of a dictionary.

`index = ord(ch) - ord('a')`

This uses O(26) = O(1) extra space. Do not use it unless the character constraints guarantee this alphabet. For unrestricted Unicode, use a dictionary or Counter.

````python
text = "banana"
freq = [0] * 26
for ch in text:
    freq[ord(ch) - ord('a')] += 1
print(freq[ord('a') - ord('a')])  # 3
print(freq[ord('n') - ord('a')])  # 2
````

## 12. Common Mistakes

1. Forgetting initialization; use `get(x, 0) + 1`.
2. Confusing distinct-key count with total elements.
3. Using a set when exact frequency matters.
4. Ignoring original order in first-unique problems.
5. Assuming Counter makes every top-k solution optimal.
6. Using a 26-element array without checking constraints.
7. Ignoring tie-breaking rules.
8. Checking only whether keys exist instead of comparing counts.

## 13. Interview Thinking Checklist

Ask:

1. What exactly am I counting: numbers, characters, pairs, or groups?
2. What should be the dictionary key?
3. Does original order matter?
4. Do I need all frequencies, a maximum, or only distinct count?
5. Is the alphabet restricted?
6. Can I compare two frequency maps?
7. Is sorting enough, or do I need a heap or buckets?
8. What are n (input length) and k (distinct values), and what are the time and space costs?

A strong explanation: "I scan the input once and maintain a frequency map. Each update is O(1) average, so building it takes O(n) average time."

## 15. Complexity Summary and Reusable Templates

Let n be the input length and k the number of distinct values.

| Task | Time | Extra space |
|---|---:|---:|
| Build frequency map | O(n) average | O(k) |
| Find max after counting | O(k) | O(k) map |
| Compare two frequency maps | O(n + m) average | O(k) |
| First unique character | O(n) average | O(k) |
| Sort distinct keys by frequency | O(n + k log k) | O(k) |

### Reusable templates

```python
# Manual frequency map
freq = {}
for x in items:
    freq[x] = freq.get(x, 0) + 1

# First item with frequency one

    if freq[x] == 1:
        ...

# Compare frequencies

```

**Key idea:** frequency counting converts repeated counting work into one pass with a hash map.
