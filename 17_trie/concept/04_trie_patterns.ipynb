# 04. Trie Patterns (Revision Sheet)

**Learning order:** DECISION TABLE → IMPLEMENTATION CHOICES → COMPLEXITY → MISTAKES → INTERVIEW SCRIPT → PRACTICE MAP

## 1. Decision table

| Question looks like | Pattern |
|---|---|
| words starting with a prefix, count words with prefix | basic trie, counts in nodes |
| autocomplete suggestions | walk to prefix + DFS (or top-k stored in nodes) |
| search with `.` wildcard | DFS branching on `.` |
| shortest root / replace by prefix | stop at first `is_end` |
| find many words in a grid | trie + backtracking |
| word is a suffix of a stream | reversed trie |
| maximum XOR, bits | binary trie |
| longest common prefix | walk while one child and no end |
| sum / count of values by prefix | store totals in nodes |

## 2. Implementation choices

| Choice | Pros | Cons |
|---|---|---|
| `children = {}` (dict) | flexible alphabet, less memory for sparse tries | slightly slower |
| `children = [None] * 26` | fast, simple | wastes memory, only a-z |
| nested dicts with a `"$"` key | very short code | harder to extend |
| compressed trie / radix tree | fewer nodes | more complex (used in routers, databases) |

## 3. Complexities

| Operation | Time | Space |
|---|---|---|
| insert / search / startsWith | O(L) | O(L) new nodes per insert |
| build for n words | O(total characters) | O(total characters) |
| autocomplete | O(L + size of subtree) | O(depth) |
| Word Search II | O(R·C·4^L) worst, pruned by the trie | O(total characters) |
| binary trie (max XOR) | O(32 · n) | O(32 · n) |

## 4. Common mistakes

- Missing `is_end`.
- `search` returning `True` for a prefix.
- Forgetting to restore state in the grid DFS.
- Adding duplicates in Word Search II.
- Using a trie when a hash set is enough.
- Not mentioning the memory cost.

## 5. Interview script

1. "I need fast prefix queries, so a trie."
2. "Insert in O(L), query in O(L). Space is O(total characters)."
3. Compare: "A hash set gives O(L) exact search but O(n·L) prefix search."
4. Test: empty string, a word that is a prefix of another, no matches.

## Quick Revision

- Trie = prefix tree, O(L) operations, `is_end` marks words.
- Wildcard = DFS. Grid of words = trie + backtracking. Stream suffix = reversed trie. XOR = binary trie.
- Memory is the cost.
