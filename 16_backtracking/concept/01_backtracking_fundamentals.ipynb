# 01. Backtracking: Fundamentals

Backtracking builds a solution **one choice at a time**. If a choice leads to a dead end (or after recording a full solution), we **undo** it and try the next choice.

**Learning order:** REAL LIFE → TEMPLATE → DECISION TREE → TRACE → SUBSETS → COMPLEXITY → BACKTRACKING VS OTHERS → INTERVIEW THINKING

## 1. Real-life examples

- **Maze:** at every junction you choose a path. If it ends in a wall, you **walk back** to the last junction and try another path.
- **Packing a suitcase** to fit a weight limit: try an item, if it does not work, take it out.
- **Sudoku:** place a digit, continue, if stuck, erase and try another digit.
- **Password cracking, scheduling, N-Queens.**

**Recognition clue:** "generate **all** subsets / permutations / combinations", "find **all** paths", "place things so that no rule is broken".

## 2. The template: choose → explore → un-choose

```python
def backtrack(state):
    if is_solution(state):
        save a COPY of the solution
        return
    for choice in available_choices(state):
        make(choice)                 # choose
        backtrack(new_state)         # explore
        undo(choice)                 # un-choose (backtrack)
```

The **undo** step restores the state so the next choice starts clean. Without it, choices leak between branches.

## 3. Subsets with the template

State: `path` (current subset) and `start` (the first index we may pick from, to avoid using the same element twice or producing the same subset in a different order).

````python
def subsets(nums):
    res, path = [], []
    def backtrack(start):
        res.append(path[:])                       # every path is a valid subset (copy it!)
        for i in range(start, len(nums)):
            path.append(nums[i])                  # choose
            backtrack(i + 1)                      # explore
            path.pop()                            # un-choose
    backtrack(0)
    return res

print(subsets([1, 2, 3]))   # [[], [1], [1, 2], [1, 2, 3], [1, 3], [2], [2, 3], [3]]
````

````python
# Trace: watch choose / un-choose on [1, 2]
def subsets_trace(nums):
    path = []
    def backtrack(start, depth):
        pad = "  " * depth
        print(f"{pad}subset {path}")
        for i in range(start, len(nums)):
            path.append(nums[i]);  print(f"{pad} choose {nums[i]}")
            backtrack(i + 1, depth + 1)
            path.pop();            print(f"{pad} undo {nums[i]}")
    backtrack(0, 0)

subsets_trace([1, 2])
````

## 4. The decision tree and complexity

Every call is a node, every choice an edge. The number of leaves is the number of solutions explored.

| Problem | Count | Time |
|---|---|---|
| subsets of n | 2ⁿ | O(n · 2ⁿ) (copying each) |
| permutations of n | n! | O(n · n!) |
| combinations C(n, k) | n! / (k!(n-k)!) | O(k · C(n, k)) |
| N-Queens | pruned, much less than nⁿ | exponential |

Backtracking is **exponential by nature**. We make it fast with **pruning**: stop a branch as soon as it cannot lead to a valid answer.

## 5. Backtracking vs recursion vs DP vs greedy

| Technique | Idea |
|---|---|
| Plain recursion | one smaller subproblem, combine |
| **Backtracking** | try all choices, undo, **list/search all** solutions |
| **DP** | same subproblems repeat → store results (count/optimize) |
| Greedy | one locally best choice, no going back |

If the question says "**return all**" → backtracking. If it says "**count / max / min**" and subproblems repeat → DP.

## 6. Interview thinking

**Say:** "At each step I choose from the remaining options, recurse, and undo the choice. I'll prune branches that cannot succeed. Time is exponential, space O(depth) for the recursion."

**Define out loud:** the **state** (what changes), the **choices**, the **goal** (when to save), the **constraints** (when to prune).

**Common mistakes:** saving `path` instead of `path[:]` (all answers become the same list), forgetting to undo, wrong `start` index (duplicates or missed answers), no pruning.

## Quick Revision

- Choose → explore → un-choose.
- Save a **copy** at the goal.
- `start` index avoids reordering duplicates in subsets/combinations.
- Exponential: prune early.
- "All solutions" → backtracking; "count/best with overlapping states" → DP.
