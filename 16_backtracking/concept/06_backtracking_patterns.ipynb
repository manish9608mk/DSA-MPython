# 06. Backtracking Patterns (Revision Sheet)

**Learning order:** TEMPLATE TABLE → DUPLICATE RULES → CHOOSING THE SHAPE → MISTAKES → INTERVIEW SCRIPT → PRACTICE MAP

## 1. Which template?

| Problem type | State | Loop | Recurse with |
|---|---|---|---|
| subsets | `path`, `start` | `i` from `start` | `i + 1`, record at every node |
| combinations (size k) | `path`, `start` | `i` from `start` | `i + 1`, record at size k |
| combination sum (reuse) | `path`, `start`, `remaining` | `i` from `start` | `i` |
| permutations | `path`, `used[]` | all `i`, skip used | same call, flip `used` |
| partitions of a string | `path`, `start` | `end` from `start+1` | `end` |
| grid path (word search) | position, index | 4 directions | mark / restore cell |
| placement (N-Queens, Sudoku) | row/cell index + sets | candidates for the cell | next cell |
| valid sequences (parentheses) | counts | allowed next chars | same string + 1 |

## 2. Duplicate handling rules

- **Combinations / subsets:** sort, then `if i > start and nums[i] == nums[i-1]: continue`.
- **Permutations:** sort, then `if used[i] or (i > 0 and nums[i] == nums[i-1] and not used[i-1]): continue`.
- The condition ensures the **same value is not chosen first twice at the same level**.

## 3. Mistakes checklist

- Appending `path` instead of `path[:]`.
- Not undoing every change (list, `used`, sets, board cell).
- Wrong `start` (`i` vs `i + 1`).
- Skipping duplicates with `i > 0` instead of `i > start`.
- No pruning → TLE.
- Not sorting before duplicate skipping or `break` pruning.
- Using backtracking when DP is expected (counting problems).
- Recursion depth issues for long inputs.

## 4. Interview script

1. "I'll build the answer step by step, choosing from ... at each step."
2. "State: ..., choices: ..., goal: ..., prune when ..."
3. "Time is O(branching ^ depth) in the worst case, pruning reduces it. Space O(depth)."
4. Test: empty input, one element, duplicates, no solution.
