# 05. Stack Patterns (Revision Sheet)

**Learning order:** DECISION TABLE → TEMPLATES → MISTAKES → INTERVIEW SCRIPT → PRACTICE MAP

## 1. Decision table

| Question looks like | Pattern |
|---|---|
| brackets / valid string | stack of openers |
| cancel adjacent items, backspace | pop when the top cancels |
| evaluate expression | operand stack (+ sign / bracket saving) |
| nested encoded string, path | stack of context |
| O(1) min / max | stack of `(value, min so far)` |
| next greater/smaller, days until | monotonic stack of indexes |
| largest rectangle, trapped water | monotonic stack with widths |
| contribution of each element | nearest smaller/greater on both sides |
| "smallest number after removing digits" | greedy monotonic stack |
| simulate recursion / DFS iteratively | explicit stack |

## 3. Common mistakes

- Popping an empty stack.
- Not checking that the stack is empty at the end.
- Pushing values when indexes are needed.
- `<` vs `<=` with equal values.
- Forgetting leftover stack elements / sentinel.
- Operand order in subtraction and division.
- Using `list.pop(0)` (that is a queue operation, O(n)).

## 4. Interview script

1. "The most recent unresolved item matters, so I'll use a stack."
2. "I scan once. A new element resolves items on top, I pop them."
3. "Each element is pushed and popped once → O(n) time, O(n) space."
4. Test: empty input, all increasing, all decreasing, duplicates, single element.

## Quick Revision

- Stack = LIFO, O(1) push/pop/peek at the end.
- Matching, cancelling, expression = plain stack.
- Next greater/smaller = monotonic stack of indexes, O(n).
- Rectangle / water / contribution = nearest smaller on both sides.
- Test empty, duplicates, sorted input.
