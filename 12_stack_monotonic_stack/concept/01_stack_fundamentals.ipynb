# 01. Stack: Fundamentals

A stack is **LIFO**: Last In, First Out. You only touch the **top**.

**Learning order:** REAL LIFE → OPERATIONS → PYTHON LIST AS STACK → VALID PARENTHESES → ADJACENT DUPLICATES → BACKSPACE → INTERVIEW THINKING

## 1. Real-life examples

- **Pile of plates:** you put a plate on top and take from the top.
- **Undo (Ctrl+Z):** the last change is undone first.
- **Browser back button:** the last page you visited is the first one you return to.
- **Function calls:** the call stack (recursion) is a real stack.

**Recognition clue:** "the most recent unmatched thing matters" (brackets, undo, nested structure).

## 2. Operations (all O(1))

| Operation | Python list |
|---|---|
| push | `stack.append(x)` |
| pop | `stack.pop()` |
| peek (top) | `stack[-1]` |
| is empty | `not stack` |

Never use `list.pop(0)` for a stack (that is the front, O(n)). Always work at the **end** of the list.

```python
class Stack:
    def __init__(self):
        self.items = []
    def push(self, x):
        self.items.append(x)
    def pop(self):
        return self.items.pop()
    def peek(self):
        return self.items[-1]
    def is_empty(self):
        return not self.items

s = Stack()
s.push(1); s.push(2); s.push(3)
print(s.pop(), s.peek(), s.is_empty())   # 3 2 False
```

## 3. Valid Parentheses

Push every **opening** bracket. On a **closing** bracket, the top must be its matching opener. At the end the stack must be empty.

```python
def is_valid(s):
    pairs = {")": "(", "]": "[", "}": "{"}
    stack = []
    for ch in s:
        if ch in pairs:                              # closing bracket
            if not stack or stack.pop() != pairs[ch]:
                return False
        else:
            stack.append(ch)
    return not stack

for t in ["()[]{}", "(]", "([)]", "{[]}", "("]:
    print(t, is_valid(t))
# ()[]{} True / (] False / ([)] False / {[]} True / ( False
```

## 4. Remove All Adjacent Duplicates

Scan left to right. If the current character equals the top, they cancel (pop). Otherwise push.

```python
def remove_duplicates(s):
    stack = []
    for ch in s:
        if stack and stack[-1] == ch:
            stack.pop()
        else:
            stack.append(ch)
    return "".join(stack)

print(remove_duplicates("abbaca"))   # ca
```

## 5. Backspace String Compare

Typing with `#` as backspace: a `#` pops the last typed character. Build both final strings with a stack and compare. (O(1)-space version uses two pointers from the right.)

```python
def build_text(s):
    stack = []
    for ch in s:
        if ch == "#":
            if stack:
                stack.pop()
        else:
            stack.append(ch)
    return stack

def backspace_compare(a, b):
    return build_text(a) == build_text(b)

print(backspace_compare("ab#c", "ad#c"), backspace_compare("a#c", "b"))   # True False
```

## 6. Interview thinking

**Signals:** brackets, nested structure, "undo", "most recent", evaluate expression, cancel adjacent pairs, simulate recursion without recursion.

**Say:** "I push items until I see something that resolves the top, then I pop. Each element is pushed and popped at most once, so O(n) time and O(n) space."

**Common mistakes:** popping from an empty stack (check first), forgetting the stack must be empty at the end, mixing up the order of the matching pair.

## Quick Revision

- LIFO. Push/pop/peek at the end of a list, all O(1).
- Brackets: push openers, match on closers, empty at the end.
- Adjacent duplicates / backspace: pop when the new item cancels the top.
- O(n) time because each item enters and leaves once.
