## 1. Binary lifting: jump 2^k steps up

`up[k][v]` = the `2^k`-th ancestor of `v`. Build: `up[k][v] = up[k-1][ up[k-1][v] ]`. Any jump of `d` steps = combine jumps by the **bits of `d`**. Build O(n log n), each query O(log n).

**LCA (many queries):**
1. Lift the deeper node up to the same depth.
2. If equal, done. Otherwise lift both together while their ancestors differ (from the biggest `k` down).
3. The LCA is the parent of either.

```python
from collections import defaultdict, deque

class LCA:
    def __init__(self, n, edges, root=0):
        graph = defaultdict(list)
        for u, v in edges:
            graph[u].append(v); graph[v].append(u)
        self.LOG = max(1, n.bit_length())
        self.depth = [0] * n
        self.up = [[root] * n for _ in range(self.LOG)]
        seen, q = {root}, deque([root])
        while q:                                            # BFS to fill parents and depths
            u = q.popleft()
            for v in graph[u]:
                if v not in seen:
                    seen.add(v)
                    self.depth[v] = self.depth[u] + 1
                    self.up[0][v] = u
                    q.append(v)
        for k in range(1, self.LOG):
            for v in range(n):
                self.up[k][v] = self.up[k - 1][self.up[k - 1][v]]

    def lca(self, a, b):
        if self.depth[a] < self.depth[b]:
            a, b = b, a
        diff = self.depth[a] - self.depth[b]
        for k in range(self.LOG):
            if diff >> k & 1:
                a = self.up[k][a]                           # lift a to b's depth
        if a == b:
            return a
        for k in range(self.LOG - 1, -1, -1):
            if self.up[k][a] != self.up[k][b]:
                a, b = self.up[k][a], self.up[k][b]
        return self.up[0][a]

#        0
#      /   \
#     1     2
#    / \     \
#   3   4     5
t = LCA(6, [(0, 1), (0, 2), (1, 3), (1, 4), (2, 5)])
print(t.lca(3, 4), t.lca(3, 5), t.lca(1, 4), t.lca(5, 5))   # 1 0 1 5
```

## 2. Landscape of "advanced" topics (what exists, when you meet it)

| Topic | Used for | Interview frequency |
|---|---|---|
| KMP / Z-function / Rabin-Karp | substring search in O(n + m) | medium (strings topic) |
| Manacher | longest palindromic substring O(n) | low |
| Suffix array / suffix automaton | many string queries | competitive programming |
| Heavy-light decomposition | path queries on trees | competitive programming |
| Network flow (Dinic, Ford-Fulkerson), matching | assignment, cuts | low (ML/OR roles) |
| 2-SAT | logic constraints | low |
| Convex hull, geometry sweep | geometry | low |
| FFT | polynomial multiplication | very low |
| Randomized algorithms, Bloom filters, HyperLogLog | big data | system design discussions |
| Union-Find variants, Mo's algorithm | offline queries | medium |

## 3. How to study these

1. **Priority 1 (interview-relevant):** Fenwick tree, segment tree, KMP, bridges, SCC. Be able to code them.
2. **Priority 2:** sparse table, binary lifting, Mo's algorithm, Rabin-Karp (hashing).
3. **Priority 3 (only if you like it or compete):** flows, suffix structures, HLD, FFT.

For product-company interviews, **most questions are solved with Priority 0 topics** (arrays, hashing, trees, graphs, DP, heaps). Advanced topics decide only a few rounds. Depth in the basics beats breadth here.
