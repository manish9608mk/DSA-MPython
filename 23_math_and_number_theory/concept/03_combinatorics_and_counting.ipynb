# 03. Combinatorics and Counting

Counting arrangements: permutations, combinations, Catalan numbers, and counting by formula instead of enumeration.

**Learning order:** BASIC FORMULAS → PASCAL'S TRIANGLE → PATHS BY nCr → CATALAN → UNIQUE BSTs → STARS AND BARS → INCLUSION-EXCLUSION → INTERVIEW THINKING

## 1. Formulas

- **Permutations** of `n` distinct items: `n!`. Choose and arrange `r` of them: `P(n, r) = n! / (n - r)!`.
- **Combinations** (order does not matter): `C(n, r) = n! / (r! (n - r)!)`.
- **Pascal's identity:** `C(n, r) = C(n-1, r-1) + C(n-1, r)`.
- With repeated items, divide by the factorial of each repeat count (arrangements of "AABC" = 4! / 2! = 12).

Python: `math.comb(n, r)`, `math.perm(n, r)`, `math.factorial(n)`.

```python
import math
print(math.comb(5, 2), math.perm(5, 2), math.factorial(5), math.factorial(4) // math.factorial(2))   # 10 20 120 12
```

## 2. Pascal's Triangle

Row `i` has `i + 1` numbers. Each inner number is the sum of the two above it. Real life: how many different ways to pick `r` friends from `n`.

```python
def pascal(num_rows):
    rows = []
    for i in range(num_rows):
        row = [1] * (i + 1)
        for j in range(1, i):
            row[j] = rows[i - 1][j - 1] + rows[i - 1][j]
        rows.append(row)
    return rows

print(pascal(5))   # [[1], [1, 1], [1, 2, 1], [1, 3, 3, 1], [1, 4, 6, 4, 1]]
```

## 3. Grid paths by formula

To go from the top-left to the bottom-right of an `m × n` grid using only right/down moves, you make `m - 1` downs and `n - 1` rights in some order: choose which of the `m + n - 2` moves are downs. **Answer: `C(m + n - 2, m - 1)`.** No DP needed.

```python
def unique_paths(m, n):
    return math.comb(m + n - 2, m - 1)

print(unique_paths(3, 7), unique_paths(3, 2))   # 28 3
```

## 4. Catalan numbers

`Cat(n) = C(2n, n) / (n + 1)`: 1, 1, 2, 5, 14, 42, ... They count:
- valid **parentheses** strings with `n` pairs,
- **binary search trees** with `n` nodes,
- ways to **triangulate** a polygon, Dyck paths, ...

Recurrence: `Cat(n) = sum Cat(i) · Cat(n - 1 - i)` (root splits the nodes into left and right).

```python
def catalan(n):
    return math.comb(2 * n, n) // (n + 1)

def num_trees(n):                                     # Unique Binary Search Trees (DP version)
    dp = [1, 1] + [0] * (n - 1)
    for nodes in range(2, n + 1):
        for root in range(1, nodes + 1):
            dp[nodes] += dp[root - 1] * dp[nodes - root]
    return dp[n]

print([catalan(i) for i in range(8)], num_trees(3), num_trees(5))   # [1, 1, 2, 5, 14, 42, 132, 429] 5 42
```

## 5. Stars and bars

Number of ways to distribute `n` identical items into `k` boxes (each box may be empty): `C(n + k - 1, k - 1)`. Think of `n` stars and `k - 1` bars separating the boxes.

Real life: distributing 10 identical chocolates among 3 children.

```python
def distribute(n, k):
    return math.comb(n + k - 1, k - 1)

print(distribute(10, 3), distribute(3, 2))   # 66 4
```

## 6. Inclusion-Exclusion

To count items with **at least one** of several properties, add single counts, subtract pair overlaps, add triple overlaps, ...

Example: numbers up to 100 divisible by 2 or 3 or 5:
`100/2 + 100/3 + 100/5 - 100/6 - 100/10 - 100/15 + 100/30 = 74`.

```python
def count_divisible_any(n, nums):
    total = 0
    for mask in range(1, 1 << len(nums)):
        lcm_val, bits = 1, 0
        for i, x in enumerate(nums):
            if mask >> i & 1:
                lcm_val = lcm_val * x // math.gcd(lcm_val, x)
                bits += 1
        total += (n // lcm_val) * (1 if bits % 2 else -1)
    return total

print(count_divisible_any(100, [2, 3, 5]))   # 74
```

## 7. Interview thinking

**Signals:** "number of ways/paths/arrangements", "valid parentheses count", "distribute identical items", "at least one of".

**Say:** "This is counting, not enumeration. The count is `C(...)`, which I compute with factorials (mod p if large) or with DP via Pascal's identity."

**Common mistakes:** confusing permutations and combinations, forgetting repeated elements, overflow without modulo, using recursion that enumerates instead of counting.
